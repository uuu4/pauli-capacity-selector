## Bolum 2 - Cekirdek  (`qk_core` + `phaseb_core` + `gate4_core` + `t11_core`, TEK KAYNAK)

```python
"""Kuantum kernel cekirdegi -- TEK KAYNAK.

Bu dosya defterlere OLDUGU GIBI GOMULUR (build_*_nb.py `open(...).read()` ile alir),
boylece Colab'de self-contained kalir ama kod tek yerde durur.

BEKLENEN ORTAM: `xp`, `GPU`, `CDT`, `FDT`, `to_xp`, `to_np` defterin 0. hucresinde
tanimlanmis olmali (cupy/numpy dagitimi).

Dogrulanmis (bkz. defterlerin Bolum 4):
  fwht == H^{otimes n} · pauli_expect_fast == naif yol (2.2e-16) · Parseval ·
  GFQK_{w<=n} == fidelity kernel · dort eksende S_req monoton · Katman B/C ayrimlari
"""


# ============================================================================
# DEVRE -- yogun 2^n x 2^n matris KURULMAZ
# ============================================================================

def fwht(a, normalize=True):
    """Son eksende H^{otimes n}. a: (..., 2^n). O(N n 2^n), yogun matris yok.

    YERINDE kelebek (butterfly). Onceki surum her adimda `moveaxis`+`stack`
    yapiyordu; profil bunun tek basina Faz B suresinin ~%15'i oldugunu gosterdi
    (7600 `stack` cagrisi). H tek-qubit kapilari komutatif oldugu icin bit
    sirasindan bagimsiz; dogrulama [1a] H^{otimes n}'e karsi kilitliyor.
    """
    dim = a.shape[-1]
    n = int(np.log2(dim))
    shp = a.shape
    b = xp.array(a, copy=True).reshape(-1, dim)
    h = 1
    while h < dim:
        b = b.reshape(-1, dim // (2*h), 2, h)
        u = xp.array(b[:, :, 0, :], copy=True)
        v = b[:, :, 1, :]
        b[:, :, 0, :] = u + v
        b[:, :, 1, :] = u - v
        b = b.reshape(-1, dim)
        h *= 2
    b = b.reshape(shp)
    return b * (2.0 ** (-n / 2)) if normalize else b


def sample_X(rng, N, n, rep=1, jitter=0.0):
    """Veri ureteci -- FAZ D icin REDUNDANS KOLU.

    `rep=1, jitter=0.0` -> eski yolun BIREBIR aynisi (ayni tek `rng` cagrisi,
    dolayisiyla Faz A/B/C/C2 sonuclari degismez).

    `rep>1`: n koordinat, `d_eff = max(3, n//rep)` GIZLI degiskenden turetilir:

        x_j = z_{j mod d_eff} + jitter * u_j,    z~U(0,2pi),  u~U(-pi,pi)

    Bu redundansi VERIDE uretir, DEVREDE degil. Amac kritik: devrenin ifade gucu
    (EC'nin kaynagi) dokunulmadan kalirken `psi` sutunlari korele hale gelsin.
    Devreyi degistirseydik EC'yi de degistirir ve testi anlamsizlastirirdik.

    `d_eff >= 3` tutulur cunku etiket bataryasi ilk 3 koordinati kullanir
    (`make_labels`, `n_fixed=3`) -- gorevin icsel zorlugu sabit kalmali.

    UYARI (raporlanmali): `rep>1` veri manifoldunun icsel boyutunu dusurur, yani
    ogrenme gorevini kolaylastirabilir. Bu, A7-vs-A3 karsilastirmasi icin ZARARSIZ
    (iki kol da ayni veriyi gorur) ama mutlak dogruluklar Faz C2 ile KIYASLANAMAZ.
    """
    if rep == 1 and jitter == 0.0:
        return rng.uniform(0, 2 * np.pi, size=(N, n))      # BIREBIR eski yol
    d = max(3, n // rep)
    z = rng.uniform(0, 2 * np.pi, size=(N, d))
    X = z[:, np.arange(n) % d]
    if jitter > 0.0:
        X = X + jitter * rng.uniform(-np.pi, np.pi, size=(N, n))
    return X


def sign_tables(n):
    z = np.arange(2 ** n)
    S = 1.0 - 2.0 * ((z[:, None] >> np.arange(n - 1, -1, -1)[None, :]) & 1)
    pairs = [(i, j) for i in range(n) for j in range(i + 1, n)]
    SP = np.stack([S[:, i] * S[:, j] for i, j in pairs], 1) if pairs else np.zeros((2 ** n, 0))
    return S, SP, pairs


_SIGN_DEV = {}


def sign_tables_dev(n):
    """`sign_tables`'in cihazdaki (FDT, transpoze) kopyasi, SON `n` icin onbellekli.

    2026-10-13 (kosu 2, n=16/20): eski `encode` tabloyu HER cagrida yeniden kurup
    cihaza tasiyordu; n=20'de bu cagri basina ~1.5 GB (SP: 2^20 x 190). Degerler
    birebir ayni -- yalniz bir kez kurulur. Tek `n` tutulur: n=20 tablosu ~0.8 GB.
    """
    if n not in _SIGN_DEV:
        _SIGN_DEV.clear()
        S, SP, pairs = sign_tables(n)
        _SIGN_DEV[n] = (to_xp(S, FDT).T, to_xp(SP, FDT).T, pairs)
    return _SIGN_DEV[n]


def encode(X, n, depth, ent, bw, norm='none'):
    """ZZ-feature map, `depth` tekrar. Doner: (M, 2^n) statevector.

    faz(z) = -( bw * sum_i x_i s_i(z)  +  ent * bw^2 * sum_{i<j} x_i x_j s_ij(z) )

    `norm` -- `n`-BAGIMLI KONFOUND kontrolu:
      'none' : ham. Terim sayilari n ve n(n-1)/2 oldugu icin faz sacilimi
               sirasiyla ~sqrt(n) ve ~n buyur -> BUYUK n'de devre DAHA COK karisir.
               Bu, "n buyudu" ile "encoding guclendi" etkilerini karistirir.
      'sqrt' : her iki terim kendi terim sayisinin karekokune bolunur -> faz
               sacilimi n'den BAGIMSIZ. "Ayni harita, farkli n" yorumu gecerli olur.

    ent=0 -> faz tamamen tek-qubit -> durum CARPIM kalir (Meyer-Wallach tam 0),
    `norm` bunu degistirmez (tek-qubit terim carpimsal yapiyi bozmaz).
    """
    St, SPt, pairs = sign_tables_dev(n)
    Xp = np.stack([X[:, i] * X[:, j] for i, j in pairs], 1) if pairs else np.zeros((len(X), 0))
    c1, c2 = 1.0, 1.0
    if norm == 'sqrt':
        c1 = 1.0 / np.sqrt(n)
        c2 = 1.0 / np.sqrt(max(len(pairs), 1))
    elif norm != 'none':
        raise ValueError(norm)
    phase = -(c1 * bw * (to_xp(X, FDT) @ St)
              + c2 * ent * (bw ** 2) * (to_xp(Xp, FDT) @ SPt))
    D = xp.exp(1j * phase).astype(CDT)
    psi = xp.zeros((len(X), 2 ** n), dtype=CDT)
    psi[:, 0] = 1.0
    for _ in range(depth):
        psi = fwht(psi) * D
    return psi


def meyer_wallach(psi, n):
    """Q = 2(1 - (1/n) sum_k tr(rho_k^2)).  0 = carpim, 1 = maksimal dolanik."""
    M = psi.shape[0]
    tot = xp.zeros(M, dtype=FDT)
    for k in range(n):
        a = xp.moveaxis(psi.reshape(M, *([2] * n)), k + 1, 1).reshape(M, 2, -1)
        rho = xp.einsum('mir,mjr->mij', a, a.conj())
        tot += xp.real(xp.einsum('mij,mji->m', rho, rho)).astype(FDT)
    return float(to_np(xp.mean(2.0 * (1.0 - tot / n))))


# ============================================================================
# PAULI DOMENI -- flip-maskesi + Walsh-Hadamard hizli yolu
# ============================================================================

def _popcount(v):
    v = np.asarray(v, dtype=np.int64).copy()
    r = np.zeros_like(v)
    while np.any(v):
        r += v & 1
        v >>= 1
    return r


def _spread4(v, n):
    v = np.asarray(v, dtype=np.int64)
    out = np.zeros_like(v)
    for k in range(n):
        out += ((v >> k) & 1) * (4 ** k)
    return out


def pool_layout_from_indices(idx, n):
    """Verilen KEYFI Pauli indeks kumesi icin layout (agirlik havuzu degil).
    `qglobal` etiketi gibi tek tuk operator gerektiren yerlerde kullanilir."""
    idx = np.unique(np.asarray(idx, dtype=np.int64))
    f_ = np.zeros(len(idx), np.int64); g_ = np.zeros(len(idx), np.int64)
    for k in range(n):
        d = (idx // (4 ** k)) % 4                  # 0=I 1=X 2=Y 3=Z
        f_ |= np.where((d == 1) | (d == 2), 1 << k, 0)
        g_ |= np.where((d == 2) | (d == 3), 1 << k, 0)
    blocks, fs = [], []
    for f in np.unique(f_):
        m = f_ == f
        blocks.append((g_[m], idx[m])); fs.append(f)
    w = np.zeros(len(idx), np.int64)
    for k in range(n):
        w += (((idx // (4 ** k)) % 4) != 0)
    o = np.argsort(idx)
    return np.array(fs), blocks, idx[o], w[o]


def pool_layout(n, h, max_flips=None, seed=0, keep_low=None):
    """Agirlik <= h havuzu.

    `max_flips`: flip-maske sayisi bunu asarsa RASTGELE alt orneklenir. Maliyet
    flip-maske sayisiyla DOGRU ORANTILI (her biri bir WHT), operator sayisiyla degil;
    o yuzden kisitlama burada yapilir. Alt ornekleme TUM kollara ayni havuzu verir,
    yani adildir -- ve A3'u (top-p) A1'den (rastgele) DAHA COK zorlar, cunku en iyi
    operator disarida kalabilir. Bizim iddiamiz acisindan MUHAFAZAKAR.

    `keep_low` (v2, 2026-10-15 denetimi): popcount <= keep_low olan flip'lerin HEPSI
    tutulur, `max_flips` yalniz geri kalanlardan (popcount > keep_low) cekilir. Neden:
    H=2 kollari H=4 icin kirpilmis havuzdan cekiliyordu (n=12'de 2-body 154/594, ZZ
    asiri temsil) ve n=14, depth=1, ent=0'da H=2 havuzunun rank'i 8'e dusup Faz B'yi
    dusurdu. `None` -> eski davranis (eski kosular birebir yeniden uretilir).

    Doner: (flip_maskeleri, [(g_dizisi, pauli_indeksleri) her f icin],
            tum_indeksler_sirali, tum_agirliklar_sirali).
    """
    allz = np.arange(2 ** n)
    pc = _popcount(allz)
    fs = np.where(pc <= h)[0]
    if keep_low is not None:
        lo, hi = fs[pc[fs] <= keep_low], fs[pc[fs] > keep_low]
        if max_flips is not None and len(hi) > max_flips:
            hi = hi[np.random.default_rng(seed + 4242).choice(len(hi), max_flips, replace=False)]
        fs = np.sort(np.concatenate([lo, hi]))
    elif max_flips is not None and len(fs) > max_flips:
        keep = np.random.default_rng(seed + 4242).choice(len(fs), max_flips, replace=False)
        fs = np.sort(fs[np.union1d(keep, 0)])      # f=0 (Z-only aile) HER ZAMAN kalir
    blocks, idx_all, w_all = [], [], []
    for f in fs:
        g = np.where(pc[allz | f] <= h)[0]            # agirlik = |f OR g|
        pidx = 2 * _spread4(f & g, n) + _spread4(f & ~g, n) + 3 * _spread4(g & ~f, n)
        blocks.append((g, pidx))
        idx_all.append(pidx)
        w_all.append(pc[g | f])
    idx_all = np.concatenate(idx_all)
    w_all = np.concatenate(w_all)
    o = np.argsort(idx_all)
    return fs, blocks, idx_all[o], w_all[o]


def pauli_expect_fast(psi, n, layout):
    """<P_{f,g}> = Re[ (-i)^{|f&g|} * WHT_z( conj(psi_z) psi_{z^f} )[g] ]

    Ayni f'yi paylasan tum operatorler tek gather + tek WHT ile hesaplanir.
    n=12, w<=4: 46 666 operator yerine 794 donusum.
    """
    fs, blocks, _, _ = layout
    z = xp.arange(2 ** n)
    conj = psi.conj()
    cols, keys = [], []
    for f, (g, pidx) in zip(fs, blocks):
        Wt = fwht(conj * psi[:, z ^ int(f)], normalize=False)
        ph = ((-1j) ** to_xp(_popcount(np.bitwise_and(int(f), g)))).astype(CDT)
        cols.append(xp.real(Wt[:, to_xp(g.astype(np.int64))] * ph[None, :]).astype(FDT))
        keys.append(pidx)
    E = xp.concatenate(cols, axis=1)
    return E[:, to_xp(np.argsort(np.concatenate(keys)))]


def pauli_expect_naive(psi, n, idx):
    """Referans yol -- SADECE dogrulama icin (yavas)."""
    flip = np.zeros(len(idx), np.int64)
    phm = np.zeros(len(idx), np.int64)
    nY = np.zeros(len(idx), np.int64)
    for k in range(n):
        d = (np.asarray(idx) // (4 ** k)) % 4
        flip |= np.where((d == 1) | (d == 2), 1 << k, 0)
        phm |= np.where((d == 2) | (d == 3), 1 << k, 0)
        nY += (d == 2)
    z = np.arange(2 ** n, dtype=np.int64)
    P = to_np(psi)
    out = np.zeros((P.shape[0], len(idx)))
    for t in range(len(idx)):
        zf = z ^ flip[t]
        sgn = 1.0 - 2.0 * (_popcount(zf & phm[t]) & 1)
        out[:, t] = np.real(np.sum(P.conj() * sgn[None, :] * P[:, zf], 1) * (1j) ** int(nY[t]))
    return out


# ============================================================================
# KERNEL -- birim kosegene normalize + tahmin edicinin shot varyansi
# ============================================================================

def kernel_gfqk(E_pool, w_pool, H, p):
    """GFQK_{w<=H}, gurultulu, birim kosegene normalize. Doner (K~, sigma2_shot).

    Gurultu: qubit basina depolarizing, Pauli-transfer ozdegeri lam=1-4p/3 -> EGZAKT.
    Shot varyansi: classical shadow, operator basina 3^w.
      k = sum_i a_i b_i, Var(a^_i) ~ 3^{w_i}
      -> Var(k^) ~ 2 sum_i 3^{w_i} <psi~_i^2>,   <psi~_i^2> = mean(E[:,i]^2)
    `c`'nin BIRINCI kuvveti kullanilir; c^2 yazmak pay/paydada sadelesip gurultu
    eksenini ters cevirir (bkz. 02_PROJECT_STATUS.md hata #16).
    """
    w = np.asarray(w_pool)
    m = to_xp(w <= H)
    ww = to_xp(w[w <= H].astype(np.float64), FDT)
    E = E_pool[:, m] * xp.sqrt((1.0 - 4.0 * p / 3.0) ** (2 * ww))[None, :]
    K = E @ E.T
    dg = xp.maximum(xp.diag(K), 1e-30)
    s2 = float(to_np(2.0 * xp.sum((3.0 ** ww) * xp.mean(E ** 2, 0)) / xp.mean(dg) ** 2))
    return K / xp.sqrt(xp.outer(dg, dg)), s2


def kernel_fidelity(psi):
    """A0 referansi. diag zaten 1; tahmin edici compute-uncompute BINOM -> sigma2 None.
    Bu, GFQK'dan FARKLI bir olcum protokoludur; mutlak shot sayilari ayni para birimi
    degil (bkz. hata #17)."""
    return xp.abs(psi.conj() @ psi.T) ** 2, None


# ============================================================================
# TESHIS -- Katman A-shot / A-learn / B / C
# ============================================================================

def effrank(ev):
    ev = xp.clip(ev.real, 0, None)
    s1, s2 = ev.sum(), (ev ** 2).sum()
    return float(to_np(s1 * s1 / s2)) if float(to_np(s2)) > 0 else 0.0


def diagnose(K, sigma2_shot=None, budgets=(1e3, 1e5, 1e7)):
    """S_req = sigma2_shot / Var[k~]  --  `1/Var` DEGIL (hata #15)."""
    M = K.shape[0]
    off = K[~xp.eye(M, dtype=bool)]
    var = float(to_np(off.var()))
    mu = float(to_np(off.mean()))
    kc = xp.clip(off, 0.0, 1.0)
    s2 = float(to_np(xp.mean(kc * (1 - kc)))) if sigma2_shot is None else float(sigma2_shot)
    er = effrank(xp.linalg.eigvalsh(K))
    d = dict(offdiag_var=var, offdiag_mean=mu, sigma2_shot=s2, M=M,
             s_req=(s2 / var) if var > 0 else float('inf'),
             effrank_K=er, effrank_frac=er / M)
    for S in budgets:
        d[f'resolv@{S:.0e}'] = float(to_np(xp.mean(xp.abs(off - mu) > 3 * np.sqrt(s2 / S))))
    return d


def layers_ac(d, budget, mean_gate, effrank_gate):
    """Doner (durum, a_shot, a_learn). Katman C ONCE: K->11^T ise bandwidth cokusu, EC DEGIL.

    UYARI: a_learn (effrank/M -> 1) EC'ye OZGU DEGILDIR -- carpim durumlarinda da
    tetiklenir ve n>=14'te 1.0000'da doyar. Bolge listesi Katman B (ustel onayi) ile
    KAPILANMALIDIR (bkz. 02_PROJECT_STATUS.md 6.96).
    """
    if d['offdiag_mean'] >= mean_gate:
        return 'COLLAPSE', False, False
    a_shot = d['s_req'] > budget
    a_learn = d['effrank_frac'] > effrank_gate
    if a_shot and a_learn:
        return 'EC-BOTH', True, True
    if a_shot:
        return 'EC-SHOT', True, False
    if a_learn:
        return 'EC-LEARN', False, True
    return 'NO-EC', False, False


def layer_b(ns, vals, n_boot=2000, seed=0):
    """log V ~ a + b*n (ustel)  vs  log V ~ a' + b'*log n (polinom), AIC ile.
    `exponential` = egim GA ust siniri < 0  VE  AIC ustelde daha kucuk.

    UYARI: `exponential=False`, "EC yok" DEMEK DEGILDIR -- egri henuz bukulme
    bolgesinde olabilir (asimptota girmemis). Bkz. 6.96 acik #1.
    """
    ns = np.asarray(ns, float)
    v = np.log(np.maximum(np.asarray(vals, float), 1e-300))
    k = len(ns)
    if k < 3:
        return None

    def fit(x, y):
        b, a = np.polyfit(x, y, 1)
        r = y - (a + b * x)
        ss = ((y - y.mean()) ** 2).sum()
        return b, (1 - (r ** 2).sum() / ss if ss > 0 else 0.0), (r ** 2).sum()

    b_e, r2_e, ss_e = fit(ns, v)
    b_p, r2_p, ss_p = fit(np.log(ns), v)
    aic = lambda ss: k * np.log(max(ss, 1e-300) / k) + 4
    rng = np.random.default_rng(seed)
    bs = []
    while len(bs) < n_boot:                       # dejenere cekilis (tek n) egimi bozar
        i = rng.integers(0, k, k)
        if len(np.unique(ns[i])) >= 2:
            bs.append(np.polyfit(ns[i], v[i], 1)[0])
    bs = np.array(bs)
    return dict(slope=float(b_e), slope_lo=float(np.percentile(bs, 2.5)),
                slope_hi=float(np.percentile(bs, 97.5)), r2_exp=float(r2_e),
                r2_poly=float(r2_p), d_aic=float(aic(ss_e) - aic(ss_p)),
                exponential=bool(np.percentile(bs, 97.5) < 0 and aic(ss_e) < aic(ss_p)))


"""Faz B cekirdegi -- ogrenme tarafi. `qk_core.py` ile birlikte deftere gomulur.

qk_core: devre, Pauli domeni, kernel, EC teshisi (Katman A/B/C)
phaseb : etiketler, kol secimi (A1-A5), merge, KRR, istatistik, Katman D

BEKLENEN ORTAM: qk_core'un tanimladigi her sey + `xp/GPU/CDT/FDT/to_xp/to_np`.
"""


# ============================================================================
# 1) ETIKET BATARYASI -- SABIT koordinat alt kumesinde
# ============================================================================
# KRITIK: her etiket sadece ilk `n_fixed` koordinata bakar. n buyurken eklenen
# qubitler SEYIRCI olur -> gorevin icsel zorlugu SABIT kalir -> dogruluk dususu
# SADECE EC'ye atfedilebilir. (v1'de 'parity' tum n'e bakiyordu, iki etki
# karisiyordu -- 02_PROJECT_STATUS.md 5.7 madde 2.)
#
# `qglobal` BILEREK istisna: n ile zorlasir, "EC referansi" olarak raporlanir.

LABELS = ['lin3', 'quad3', 'parity3', 'trig3', 'qlocal', 'qglobal']


def make_labels(X, kind, n, seed, states=None, n_fixed=3):
    rng = np.random.default_rng(seed + 5000)
    k = min(n_fixed, X.shape[1])
    x = X[:, :k]
    if kind == 'lin3':
        vals = np.sin(x) @ rng.normal(size=k)
    elif kind == 'quad3':
        vals = np.sin(x[:, 0])*np.sin(x[:, 1]) + 0.5*np.cos(x[:, 2 % k])
    elif kind == 'parity3':
        # ZATEN +-1 degerli -> medyan-merkezleme UYGULANMAZ (self-test yakaladi:
        # medyan +-1 cikinca sign(vals-medyan) her seyi tek sinifa cevirir).
        y = np.prod(np.sign(np.sin(x)), axis=1).astype(float)
        y[y == 0] = 1.0
        return y
    elif kind == 'trig3':
        vals = np.sin(x[:, 0] + x[:, 1])*np.cos(x[:, 2 % k])
    elif kind == 'qlocal':                         # ilk 2 qubitte etkili gozlenebilir
        assert states is not None
        d = 2**n
        A = rng.normal(size=(4, 4)) + 1j*rng.normal(size=(4, 4))
        O = (A + A.conj().T)/2
        s = to_np(states).reshape(-1, 4, d // 4)
        vals = np.real(np.einsum('mir,ij,mjr->m', s.conj(), O, s))
    elif kind == 'qglobal':
        # GLOBAL gozlenebilir: O = sum_j c_j P_j, P_j = UNIFORM RASTGELE AGIRLIK-n Pauli.
        #
        # ONCEKI SURUM 2^n x 2^n YOGUN MATRIS URETIYORDU -> n=16'da 68 GB, Colab
        # oturumu cokuyordu; n=12'de de tohum basina ~3 x (M x 4096 x 4096) einsum
        # ile surenin cogunu yiyordu. Bu surum matematiksel olarak ayni sinifta
        # (tum qubitlere dokunan gozlenebilir) ama maliyeti O(R M n 2^n).
        assert states is not None
        R = 8
        # agirlik-n Pauli: her qubitte X/Y/Z (identity YOK) -> basamak in {1,2,3}
        idx = np.array([sum(int(dd)*(4**k) for k, dd in enumerate(rng.integers(1, 4, n)))
                        for _ in range(R)], dtype=np.int64)
        cj = rng.normal(size=R)
        # R cekilis CAKISABILIR (n=8'de ~%0.4). `pool_layout_from_indices` unique
        # uyguladigi icin sutun sayisi R'den az olur -> eski surumde IndexError.
        # Cakisan terimlerin katsayilarini TOPLA: sum_j c_j P_j degismez.
        uniq, inv = np.unique(idx, return_inverse=True)
        c_agg = np.zeros(len(uniq))
        np.add.at(c_agg, inv, cj)
        lay = pool_layout_from_indices(uniq, n)
        E = to_np(pauli_expect_fast(states, n, lay))   # sutunlar artan Pauli indeksi
        vals = E @ c_agg                               # == uniq sirasi, yeniden siralama YOK
    else:
        raise ValueError(kind)
    y = np.sign(vals - np.median(vals))
    y[y == 0] = 1.0
    return y


# ============================================================================
# 2) KOL SECIMI  A1-A5
# ============================================================================

def capacity(E_sel):
    """gamma_i = E_x[psi_i(x)^2]  -- ETIKETSIZ kapasite (bizim kriterimiz)."""
    return to_np(xp.mean(E_sel**2, axis=0))


def select_random(n_pool, p, rng, n_draws=5):
    """A1 -- Gan et al.'in fiilen yaptigi: uniform rastgele."""
    return [rng.choice(n_pool, size=min(p, n_pool), replace=False) for _ in range(n_draws)]


def select_hbody(w_pool, p, rng, n_draws=5):
    """A2_low -- DUSUK AGIRLIK ONCELIKLI rastgele secim: agirlik<=H operatorleri UNIFORM
    ele alir, en kucuk H'den baslar. KAPASITE BILGISI KULLANMAZ.

    ⚠️ 2026-10-15 DUZELTME: Bu, Gan+'in H-body LPQK'si DEGILDIR (2311.13552 Eq. 42-43:
    agirligi TAM H olan Pauli'ler, I yok; deneyde tam 2-body havuzdan rastgele p).
    Faz B-D ve T1.1/FM kosu 1'de "A2_hbody" adiyla bu fonksiyon kullanildi ve Gan+'in
    deney rakibinden GUCLU cikti (runs/rival_check_v1). Gan+'a sadik kol: `select_hbody_exact`.

    v1 HATASI: kapasite-sirali listeden `cand[:p]` aliyordu -> A2 gizlice kapasite
    kullaniyordu ve H=H_max'te A3 ile OZDESLESIYORDU (hata #12).
    """
    w = np.asarray(w_pool)
    for H in range(1, int(w.max()) + 1):
        cand = np.where(w <= H)[0]
        if len(cand) >= p:
            if len(cand) == p:
                return [cand]
            return [rng.choice(cand, size=p, replace=False) for _ in range(n_draws)]
    cand = np.arange(len(w))
    return [rng.choice(cand, size=min(p, len(cand)), replace=False) for _ in range(n_draws)]


def select_hbody_exact(w_pool, h, p, rng, n_draws=5):
    """A2_hb{h} -- Gan+'a SADIK H-body (2311.13552 Eq. 42-43, Sec. IV): agirligi TAM `h`
    olan operatorlerden UNIFORM rastgele p. I (agirlik 0) hicbir zaman girmez.

    Gan+ deneyde `h = 2`, "averaged over 10 different random selections of p features".
    Havuz `p`'den kucukse BOS liste doner (kol o satirda tanimsiz); esitse tek secim.
    """
    cand = np.where(np.asarray(w_pool) == h)[0]
    if len(cand) < p:
        return []
    if len(cand) == p:
        return [cand]
    return [rng.choice(cand, size=p, replace=False) for _ in range(n_draws)]


def select_topp(gamma, p):
    """A3 -- BIZIM kriterimiz: kapasiteye gore top-p (veri-bagimsiz, etiketsiz)."""
    return np.argsort(gamma)[::-1][:p]


def weight_normalized_capacity(gamma, w):
    """gamma~_i = gamma_i / E[gamma | agirlik(i)]  -- AGIRLIK YANLILIGI duzeltmesi.

    TARAMA BULGUSU (runs/screen_v1): 157 sinyalli hucrede top-16'nin agirlik profili
        BIZ (gamma):  w1:0.74  w2:0.19  w3:0.01  w4:0.00
        ORACLE (KTA): w1:0.11  w2:0.42  w3:0.31  w4:0.16
    Sinyal w2-w4'te, biz w1'e bakiyoruz. Sebep: yapili durumlarda dusuk agirlikli
    gozlenebilirlerin beklenen degerleri DOGAL OLARAK buyuk; gamma mutlak buyuklugu
    olcuyor, bilgi tasiyip tasimadigini degil. Ayrica bu, A3'un A2 (H-body, o da
    dusuk agirlik seciyor) ile ozdeslesmesini ve ona kaybetmesini acikliyor.

    Duzeltme: her operatoru KENDI AGIRLIK SINIFINDAKI tipik degere gore sirala.
    Etiketsiz kalir (sadece gamma ve agirlik kullanir) -> A5'in aksine ust sinir degil.
    """
    g = np.asarray(gamma, dtype=float)
    w = np.asarray(w)
    out = np.zeros_like(g)
    for ww in np.unique(w):
        m = (w == ww)
        mu = g[m].mean()
        out[m] = g[m]/mu if mu > 0 else 0.0
    return out


def select_topp_wnorm(gamma, w, p):
    """A6 -- agirlik-normalize kapasiteye gore top-p."""
    return np.argsort(weight_normalized_capacity(gamma, w))[::-1][:p]


# --- KAPI 3: HOPE'un gercek kurali  argmin J_k / dP_k -------------------------

def select_greedy_residual(E_sel, p, cost=None, block=4096):
    """A7/A8 -- ARTIK kapasiteye gore acgozlu secim (HOPE'un rate-distortion kurali).

    `gamma_i = E[psi_i^2]` bir operatorun MUTLAK katkisini olcer. HOPE'un kurali ise
    marjinaldir: bir birimin degeri, ZATEN TUTULANLARIN USTUNE ne kattigidir.
    Burada bu, secilenlerin gerdigi altuzaya dik bilesenin kapasitesidir:

        gamma_i^res = || psi_i - P_S psi_i ||^2 / M ,   S = simdiye kadar secilenler

    Her adimda `argmax gamma^res` secilir (= `argmin J_k/dP_k`, dP_k = 1 operator).
    `argmax gamma` bunun S=bos ozel halidir -- yani A3, bu kuralin ILK ADIMIDIR ve
    redundansi hic gormez. Tarama v2 bu korlugun bedelini olctu: A6 agirlik profilini
    duzeltti ama ortusme %6'da kaldi (bkz. 02_PROJECT_STATUS.md 6.995).

    `cost` verilirse marjinal kapasite ONA BOLUNUR (A8): tam `J_k/dP_k` formu,
    `dP_k` = operatorun olcum maliyeti (classical shadow: 3^w).
    UYARI: 3^w cok hizli buyudugu icin A8'in dusuk agirliga cokmesi BEKLENIR --
    yani A3'un hatasini tekrarlayabilir. Ikisi de olculup raporlanir.

    ETIKETSIZ kalir: yalnizca secim setindeki `psi` degerlerini kullanir.
    """
    R = xp.array(E_sel, copy=True)                    # (M, K) artiklar
    M, K = R.shape
    nrm = xp.sum(R**2, axis=0) / M
    c = None if cost is None else to_xp(np.asarray(cost, dtype=np.float64), FDT)
    chosen = []
    for _ in range(min(p, K)):
        score = nrm if c is None else nrm / xp.maximum(c, 1e-30)
        j = int(to_np(xp.argmax(score)))
        if float(to_np(nrm[j])) <= 1e-24:             # kalan her sey zaten kapsanmis
            break
        chosen.append(j)
        q = R[:, j] / xp.sqrt(xp.sum(R[:, j]**2))
        for s in range(0, K, block):                  # bellek dostu izdusum
            sl = slice(s, min(s+block, K))
            R[:, sl] -= xp.outer(q, q @ R[:, sl])
        nrm = xp.sum(R**2, axis=0) / M
        nrm[to_xp(np.asarray(chosen))] = 0.0
    # v2 (2026-10-15): rank < p ise kalan yer secilmemislerden ham kapasite sirasiyla
    # doldurulur. Eskiden `break` p'den kisa liste donduruyordu -> kollar ust uste
    # konamayip Faz B v2 dustu (n=14, depth=1, ent=0, H=2: rank 8). Doldurulanlar
    # secilenlerin gerdigi altuzayda (artik ~0), kernel'e yeni yon katmaz.
    if len(chosen) < min(p, K):
        cap = to_np(xp.sum(E_sel**2, axis=0)).astype(float)
        cap[np.asarray(chosen, dtype=int)] = -np.inf
        chosen += [int(j) for j in np.argsort(-cap, kind='stable')[:min(p, K) - len(chosen)]]
    return np.asarray(chosen)


def select_kta(E_sel, y_sel, p):
    """A5 -- DENETIMLI ust sinir (kernel-target alignment).
    KTA_i = (sum_m psi_i(x_m) y_m)^2 / sum_m psi_i(x_m)^2.  Etiket KULLANIR."""
    y = to_xp(np.asarray(y_sel, dtype=np.float64), FDT)
    num = (y[None, :] @ E_sel).ravel()**2
    den = xp.sum(E_sel**2, axis=0) + 1e-30
    return to_np(xp.argsort(num/den)[::-1][:p])


# ============================================================================
# 3) MERGE -- genellestirilmis Eckart-Young, SADECE etiketsiz sette ogrenilir
# ============================================================================

def greedy_merge(E_sel, idx, keep):
    """En korele cifti tekrar tekrar birlestirir; `keep` yon kalana dek.

    G = [[g_i, m_ij],[m_ij, g_j]] -> birlesik kapasite = lam1, distorsiyon = lam2,
    yon = ust ozvektor.  Doner: (C, distorsiyon)  --  C: (len(idx), keep) katsayi
    matrisi; ozellikler F = E[:, idx] @ C.

    KRITIK: C SADECE etiketsiz secim setinden ogrenilir, sonra tr/te'ye UYGULANIR.
    """
    E = E_sel[:, to_xp(np.asarray(idx))]
    G = to_np(E.T @ E) / E.shape[0]                  # (K,K) Gram
    K = G.shape[0]
    C = np.eye(K)
    dist = 0.0
    while C.shape[1] > keep:
        Gc = C.T @ G @ C                             # mevcut yonlerin Gram'i
        d = np.sqrt(np.maximum(np.diag(Gc), 1e-30))
        R = np.abs(Gc / np.outer(d, d))              # normalize korelasyon
        np.fill_diagonal(R, -1.0)
        i, j = np.unravel_index(np.argmax(R), R.shape)
        sub = Gc[np.ix_([i, j], [i, j])]
        ev, V = np.linalg.eigh(sub)                  # artan sirada
        dist += float(ev[0])                         # lam2 = kaybedilen kapasite
        new = C[:, i]*V[0, 1] + C[:, j]*V[1, 1]      # lam1 yonu
        C = np.delete(C, [i, j], axis=1)
        C = np.column_stack([C, new])
    return C, dist


# ============================================================================
# 4) KERNEL + KRR  (ozayrisimla lambda taramasi ~bedava)
# ============================================================================

def kernel_pair(F_a, F_b, diag_a=None, diag_b=None):
    """Birim kosegene normalize kernel. F: (M, p) ozellikler."""
    da = xp.maximum(xp.sum(F_a**2, 1), 1e-30) if diag_a is None else diag_a
    db = xp.maximum(xp.sum(F_b**2, 1), 1e-30) if diag_b is None else diag_b
    return (F_a @ F_b.T) / xp.sqrt(xp.outer(da, db))


LAMS = (1e-4, 1e-3, 1e-2, 1e-1, 1.0)


def krr_multi(Ktr, Kte, Ytr, Yte, lams=LAMS):
    """Kernel ridge, TUM etiketler tek cagride, `lambda` EGZAKT LOO-CV ile secilir.

    Neden boyle: kernel etiketten BAGIMSIZ, yani 6 etiket ayni ayrisimi paylasabilir.
    Ktr = U diag(ev) U^T BIR KEZ hesaplanir; sonra her lambda ve her etiket icin
    cozum sadece O(M^2) matris-vektor.

    LOO kapali formu (kat dongusu YOK, hem daha hizli hem daha iyi istatistik):
        H = U diag(ev/(ev+lam)) U^T,   h_i = H_ii
        yhat_i^{-i} = (yhat_i - h_i y_i) / (1 - h_i)

    Ilk surumde her (etiket, lambda, kat) icin ayri `solve` vardi -> ~112k cagri,
    pilot saatlerce surdu. Bu surum ayni sonucu ~2 ozayrisimla veriyor.

    Doner: {etiket_adi: (dogruluk_yuzde, correct_vec)}
    """
    Ktr = Ktr if Ktr.ndim == 3 else Ktr[None]      # (B,M,M)
    Kte = Kte if Kte.ndim == 3 else Kte[None]      # (B,Mte,M)
    names = list(Ytr.keys())
    Y = to_xp(np.column_stack([Ytr[k] for k in names]).astype(np.float64), FDT)   # (M,L)
    B, L = Ktr.shape[0], len(names)

    ev, U = xp.linalg.eigh(Ktr)                    # YIGIN halinde: tek cagri
    ev = xp.clip(ev.real, 0, None)                 # (B,M)
    UtY = xp.matmul(xp.swapaxes(U, 1, 2), Y[None])  # (B,M,L)  = U^T Y
    U2 = U**2
    best_acc = xp.full((B, L), -1.0, dtype=FDT)
    best_lam = np.zeros((B, L))
    for lam in lams:
        d = 1.0/(ev + lam)
        f = (ev*d)[:, :, None]                     # (B,M,1)
        yhat = xp.matmul(U, f*UtY)
        h = xp.matmul(U2, (ev*d)[:, :, None])
        loo = (yhat - h*Y[None]) / xp.maximum(1.0 - h, 1e-6)
        acc = xp.mean(xp.sign(loo) == Y[None], axis=1)     # (B,L)
        upd = acc > best_acc
        best_lam = np.where(to_np(upd), lam, best_lam)
        best_acc = xp.where(upd, acc, best_acc)

    lam_x = to_xp(best_lam, FDT)                   # (B,L)
    d = 1.0/(ev[:, :, None] + lam_x[:, None, :])   # (B,M,L)
    A = xp.matmul(U, d*UtY)                        # (B,M,L)  katsayilar
    P = to_np(xp.sign(xp.matmul(Kte, A)))          # (B,Mte,L)
    out = {}
    for b in range(B):
        for j, k in enumerate(names):
            c = (P[b, :, j] == np.asarray(Yte[k])).astype(np.float64)
            out[(b, k) if B > 1 else k] = (float(c.mean()*100), c)
    return out


def krr_accuracy(Ktr, Kte, ytr, yte, lams=LAMS):
    """Tek etiketlik ince sarmalayici (dogrulama testleri icin)."""
    r = krr_multi(Ktr, Kte, {'y': ytr}, {'y': yte}, lams)
    return r['y']


# ============================================================================
# 5) KATMAN D -- ogrenilebilirlik on-kosulu  (ON-KAYITLI)
# ============================================================================
# Faz A/A2/A2R2'nin tamami bu adima bagli. "w<=2 kernel konsantre olmuyor" tek
# basina UCUZ bir sonuc -- sabit bir kernel de konsantre olmaz. Bulgunun anlam
# kazanmasi icin ayni kernel'in HALA OGRENDIGI gosterilmelidir.

D_LIVE   = 5.0    # bir hucrenin karar tablosuna girmesi icin: en iyi kol >= sans+5
D_A0DEAD = 3.0    # "A0 olu" esigi: A0 < sans+3


def layer_d(acc_by_arm, chance=50.0):
    """A0 x Lego 2x2 tablosu. Doner: (durum, canli_mi, altin_mi)

    |                | Lego canli                  | Lego olu        |
    |----------------|-----------------------------|-----------------|
    | **A0 olu**     | `EC-ACTIVE-LEGO-SURVIVES` ✅ | `DEAD`          |
    | **A0 canli**   | `EC-MILD`                   | `LEGO-LOSES` ⚠  |

    `EC-ACTIVE-LEGO-SURVIVES` -> ALTIN, EC-mitigasyon iddiasi icin GECERLI tek rejim.
    `EC-MILD`     -> A0 saglikli; buradaki kazanc EC-mitigasyon KANITI DEGIL.
    `LEGO-LOSES`  -> tam kernel'de sinyal VAR, biz kaybediyoruz. **Bizim ALEYHIMIZE
                     sonuc, durustce raporlanmali.** (v1'in yaniltici 'EC-ACTIVE-ALL-DEAD'
                     etiketi bu durumu gizliyordu.)
    `DEAD`        -> hicbir kol ogrenmiyor, hucre bilgi tasimiyor -> dislanir.
    `AMBIGUOUS`   -> A0, [sans+3, sans+5) bandinda; ne olu ne canli sayilir, dislanir.
    """
    a0 = acc_by_arm.get('A0', 0.0)
    lego = max(v for k, v in acc_by_arm.items() if k != 'A0')
    a0_dead = a0 < chance + D_A0DEAD
    a0_alive = a0 >= chance + D_LIVE
    lego_alive = lego >= chance + D_LIVE
    if not (a0_alive or lego_alive):
        return 'DEAD', False, False
    if a0_dead and lego_alive:
        return 'EC-ACTIVE-LEGO-SURVIVES', True, True
    if a0_alive and lego_alive:
        return 'EC-MILD', True, False
    if a0_alive and not lego_alive:
        return 'LEGO-LOSES', True, False
    return 'AMBIGUOUS', False, False


# --- HATA #29 duzeltmesi (2026-08-17) -------------------------------------
# Faz B FULL'de altin hucrelerin %87'sinde `>=sans+5` sartini SADECE `A5_super`
# (etiket kullanan oracle) sagliyordu. Yani sart "bilgi hala orada" diyordu,
# "bizim ETIKETSIZ yontemimiz calisiyor" DEMIYORDU. KATI ALTIN bunu kapatir.
# Faz C'de ON-KAYITLI birincil kapi budur (01_TEST_PROTOCOL.md 6.37).
STRICT_UNSUP = ('A1_random', 'A2_hbody', 'A3_topp', 'A4_merge', 'A6_wnorm', 'A7_resid',
                'A8_cost',   # A8 KAPI 4'te acildi (2026-08-21); eski kosularda bu
                             # anahtar hic uretilmedigi icin Faz A-D sonuclari AYNI
                'A2_hb1', 'A2_hb2', 'A2_low')   # v2 rakipleri (2026-10-15), ayni mantik


def layer_d_strict(acc_by_arm, chance=50.0):
    """Katman D, ama "Lego hayatta" yarisini SADECE ETIKETSIZ bir kol saglayabilir.

    `layer_d` ile ayni 2x2; tek fark `A5_super` (ve baska her denetimli kol)
    hayatta-kalma oyuna KATILMAZ. `A0` her zaman kalir (karsilastirma ekseni).
    """
    sub = {k: v for k, v in acc_by_arm.items() if k == 'A0' or k in STRICT_UNSUP}
    if len(sub) < 2:
        raise ValueError('layer_d_strict: etiketsiz kol yok')
    return layer_d(sub, chance)


def redundancy(E_sel, idx):
    """Secilen `psi` sutunlari arasinda ortalama |Pearson korelasyonu| (kosegen haric).

    A7/A4'un mekanizmasi REDUNDANSA dayanir: artik-kapasite ve merge, ancak
    sutunlar korele oldugunda ham `gamma`'dan farklilasir. Bu olcum E2'nin
    (en guclu etki rejiminin taranmasi) OLCULEBILIR halidir -- tarama v3
    "A7 == A3" dedi ama redundansin O REJIMDE var olup olmadigini olcmemisti.
    """
    F = E_sel[:, to_xp(np.asarray(idx))]
    F = F - xp.mean(F, 0)
    s = xp.sqrt(xp.sum(F**2, 0))
    C = to_np((F.T @ F) / xp.maximum(xp.outer(s, s), 1e-30))
    k = C.shape[0]
    if k < 2:
        return 0.0
    return float(np.abs(C[~np.eye(k, dtype=bool)]).mean())


# ============================================================================
# 6) ISTATISTIK -- 01_TEST_PROTOCOL.md esikleri, DEGISTIRILMEDI
# ============================================================================

WIN_THR, ELIM_THR = 1.0, 0.3


def paired_bootstrap(A, B, n_boot=4000, seed=0):
    """A, B: (n_seed, n_test) dogru/yanlis matrisleri. Tohum duzeyinde BLOK
    yeniden orneklemeli eslestirilmis bootstrap."""
    A = np.asarray(A, float); B = np.asarray(B, float)
    D = A - B
    rng = np.random.default_rng(seed)
    ns, nt = D.shape
    point = float(D.mean()*100)
    st = np.empty(n_boot)
    for t in range(n_boot):
        s = rng.integers(0, ns, ns); i = rng.integers(0, nt, nt)
        ws = np.bincount(s, minlength=ns)
        wi = np.bincount(i, minlength=nt)
        st[t] = (ws @ D @ wi) / (ns * nt) * 100.0
    return point, float(np.percentile(st, 2.5)), float(np.percentile(st, 97.5))


def verdict(point, lo, hi, win=WIN_THR, elim=ELIM_THR):
    """Iki esik ailesi (hata #3): kazanc ve eleme esikleri TERS yone calisir."""
    if lo > 0 and point >= win:
        return 'KAZANC'
    if hi < elim:
        return 'ELEME-ADAYI'
    return 'SONUCSUZ'


COMPARISONS = {
    # --- protokolun ana sarti: A3/A4 hem A1'i hem A2'yi yensin, >=2 etikette ---
    'A3 top-p vs A1 rand':  ('A3_topp',  'A1_random'),
    'A3 top-p vs A2 hbody': ('A3_topp',  'A2_hbody'),
    'A4 merge vs A1 rand':  ('A4_merge', 'A1_random'),
    'A4 merge vs A2 hbody': ('A4_merge', 'A2_hbody'),   # v1'de EKSIKTI (hata #13)
    'A4 merge vs A3 top-p': ('A4_merge', 'A3_topp'),
    # --- Kapi 3 kollari (tarama v3'ten sonra eklendi; ELEME icin E2 eksik) ---
    'A7 resid vs A2 hbody': ('A7_resid', 'A2_hbody'),
    'A7 resid vs A1 rand':  ('A7_resid', 'A1_random'),
    'A7 resid vs A3 top-p': ('A7_resid', 'A3_topp'),
    'A7 resid vs A4 merge': ('A7_resid', 'A4_merge'),   # redundansa iki farkli cevap
    'A6 wnorm vs A3 top-p': ('A6_wnorm', 'A3_topp'),
    # --- Faz C: EC-mitigasyon iddiasinin ASIL cumlesi (2026-08-17) ---
    # "A0 olu, biz canli" bir REJIM etiketiydi (Katman D); bu ise DOGRUDAN,
    # eslestirilmis, GA'li bir karsilastirma. Iddia icin en gucluk tek satir.
    'A3 top-p vs A0 fidel': ('A3_topp',  'A0'),
    'A7 resid vs A0 fidel': ('A7_resid', 'A0'),
    # --- ust sinir ---
    # HATA #32 (2026-08-19): tek yonlu karsilastirma "A3 anlamli USTUN" vakalarini
    # gorunmez kiliyordu -- elle hesaplamak zorunda kaldim. Simdi iki yonlu.
    'A3 top-p vs A7 resid': ('A3_topp',  'A7_resid'),
    'A5 super vs A3 top-p': ('A5_super', 'A3_topp'),
    'A5 super vs A7 resid': ('A5_super', 'A7_resid'),
}


"""KAPI 4 cekirdegi -- KAYNAK MUHASEBESI.  TEK KAYNAK.

qk_core : devre, Pauli domeni, kernel, EC teshisi (Katman A/B/C)
phaseb  : etiketler, kol secimi (A1-A7), merge, KRR, istatistik, Katman D
gate4   : shot muhasebesi -- shadow varyansi, gurultulu kernel, S*, butce

BEKLENEN ORTAM: qk_core + phaseb_core'un tanimladigi her sey +
`np/xp/GPU/CDT/FDT/to_xp/to_np` (defterin 0. hucresi).

================================ PARA BIRIMI =================================
1 shot = 1 DEVRE CALISTIRMASI. Iki olcum protokolu AYNI birimde sayilir ama
kernel matrisine FARKLI carpanla donusur:

  shadow   (A1-A8): S snapshot x NOKTA sayisi. Tum (x,y) ciftleri
                    post-processing'den bedava gelir  ->  maliyet M ile LINEER.
  fidelity (A0)   : compute-uncompute, her CIFT icin ayri devre
                                                     ->  maliyet M^2 ile.

Bu `M` carpani Kapi 4'un en buyuk tek kalemidir ve hata #17'nin ("A0'in mutlak
shot sayilari GFQK ile ayni para birimi degil") cozumudur: birim ayni, carpan
farkli, ikisi de acikca yazili.

Ikinci eksen (devre derinligi) `twoq_gates` ile ayri raporlanir; ucuncu eksen
(model karmasikligi `p`) kollar arasi kiyasta SABIT tutulur. Protokol geregi
uc eksen BIRLIKTE degil AYRI AYRI raporlanir (01_TEST_PROTOCOL.md, KAPI 4).
==============================================================================
"""

SHADOW_BASE = 3.0


def op_cost(w):
    """`dP_k` -- bir Pauli'nin classical-shadow olcum maliyeti: `3^agirlik`.

    A8'in maliyet vektoru budur (`select_greedy_residual(..., cost=op_cost(w))`,
    yani HOPE'un tam `argmin J_k/dP_k` kurali). Kapi 1-3'te `dP_k = 1` alinmisti;
    protokol A8'i "shot muhasebesi Kapi 4'un konusu" diye burada aciyor.
    """
    return SHADOW_BASE ** np.asarray(w, dtype=float)


def dir_var(w, C=None, model='l1'):
    """Yonlerin TEK SNAPSHOT varyansi. Doner `(p,)`.

    `C=None`   -> ciplak Pauli'ler: `v_k = 3^{w_k}`.
    `C` verili -> birlestirilmis yon `f_k = sum_i C_ik P_i` (A4):
      'l1' (BIRINCIL, muhafazakar): shadow normu alt-toplamsal,
           `Var <= (sum_i |C_ik| 3^{w_i/2})^2`
      'l2' (IYIMSER alt sinir): bilesen tahminleri BAGIMSIZmis gibi,
           `Var = sum_i C_ik^2 3^{w_i}`

    Ikisi de hesaplanip raporlanir; `l1/l2` orani merge'in SHOT EKSENINDEKI
    bedelidir (olcut G2). Protokolun 2. sorusu -- "bilesen Pauli'leri ayri ayri
    mi gerekiyor?" -- cevabi: shadow altinda EK SNAPSHOT gerekmez (ayni anlik
    goruntuler tum Pauli'leri besler), ama birlesik yonun VARYANSI buyuyebilir.
    Yani merge'in maliyeti "sifir" degil: "ekstra devre yok, ekstra varyans var".
    """
    c = op_cost(w)
    if C is None:
        return c
    A = np.abs(np.asarray(C, dtype=float))
    if model == 'l1':
        return (A.T @ np.sqrt(c)) ** 2
    if model == 'l2':
        return (A.T ** 2) @ c
    raise ValueError(model)


def sigma2_arm(F, v):
    """`S=1` shot varyansi, birim-kosegene normalize kernel icin. `qk_core.
    kernel_gfqk` ile AYNI form: `2 * sum_k v_k <f_k^2> / <diag>^2`.

    F: `(M, p)` yon degerleri · v: `(p,)` tek-snapshot varyanslari.
    `S` shot'ta varyans `sigma2/S`, dolayisiyla `S_req = sigma2 / Var[k~]`
    (`1/Var` DEGIL -- hata #15).
    """
    mf2 = to_np(xp.mean(F ** 2, 0)).astype(float)
    dg = float(to_np(xp.mean(xp.sum(F ** 2, 1))))
    return float(2.0 * np.sum(np.asarray(v, dtype=float) * mf2) / max(dg, 1e-30) ** 2)


def noisy_features(F, v, S, rng):
    """`f_hat = f + N(0, v/S)`. Doner `(F_hat, diag_hat)`; diag YANLILIK DUZELTILMIS.

    `E[sum_k f_hat_k^2] = sum_k f_k^2 + sum_k v_k/S` -> duzeltilmezse dusuk
    normlu noktalar yapay olarak birbirine benzer gorunur (hata #16 ile ayni
    aile: gurultu ekseni ters donerdi). Cikarilir, pozitife kirpilir.

    BILINEN BASITLESTIRME (on-kayitli): ayni shadow kumesi TUM operatorleri
    besledigi icin `eps`'ler gercekte KORELELIDIR; burada bagimsiz alinir. Bu
    hata TUM kollarda ayni yonde -> `S*` ORANLARI icin zararsiz, MUTLAK `S*`
    icin iyimser. Kapi 4'un karari oranlar uzerinden verilir.
    """
    sd = np.sqrt(np.asarray(v, dtype=float) / float(S))
    G = F + to_xp(rng.standard_normal(size=tuple(F.shape)) * sd[None, :], FDT)
    return G, xp.maximum(xp.sum(G ** 2, 1) - float(np.sum(sd ** 2)), 1e-30)


def noisy_fidelity(K, S, rng, symmetric=False):
    """A0'in gurultu modeli: compute-uncompute BINOM tahmin edici.

    `k_hat ~ Binom(S, k)/S` -> `Var = k(1-k)/S`. Shadow'daki `3^w` yuku YOK;
    A0'in bedeli varyansta degil, CIFT SAYISINDA (bkz. `budget`).

    `symmetric=True`: egitim blogu icin. Girdi bloklari bagimsiz gurultulendigi
    icin `K_hat` simetrik cikmaz; `krr_multi` `eigh` kullandigindan simetrize
    edilir. Bu ayni zamanda fiziksel olarak da dogru: `(x,y)` ve `(y,x)` AYNI
    devredir, iki kez olculmez -- yani simetrizasyon shot da kazandirmaz,
    `budget('fidelity', ...)` zaten ust ucgeni sayar.
    """
    Kc = np.clip(to_np(K).astype(float), 0.0, 1.0)
    Kh = rng.binomial(int(S), Kc) / float(int(S))
    if symmetric:
        Kh = (Kh + Kh.T) / 2.0
    return to_xp(Kh, FDT)


def s_star(S_grid, accs, target, interp=True):
    """Hedefi saglayan ve DAHA BUYUK her `S`'te de saglayan en kucuk `S`.

    Monotonluk sarti, tek bir gurultu dalgalanmasinin `S*`'i yapay olarak
    kucultmesini engeller (kol karsilastirmasinda bu dogrudan sahte kazanc
    olurdu). `S_grid` ARTAN olmali. Hicbiri saglamiyorsa `inf`.

    `interp=True` (varsayilan): kesisme `log10(S)`'te DOGRUSAL ara degerle
    bulunur. GEREKCE: grid adimi yarim dekad (~3.16x); grid'e yapistirilmis
    `S*` yalnizca `{1, 3.16, 10, ...}` oranlarini uretebilir, yani "2x tasarruf"
    gibi bir esik OLCULEMEZ hale gelir. Ara deger cozunurlugu surekli yapar.
    Gridin ILK noktasinda zaten saglaniyorsa `S_grid[0]` doner (yani "<= S_min",
    ust sinir olarak okunur); `inf` girdisinde ara deger yapilmaz.
    """
    a = list(accs)
    for i, S in enumerate(S_grid):
        if not all(x >= target for x in a[i:]):
            continue
        if i == 0 or not interp or np.isinf(S):
            return float(S)
        s0, a0, a1 = S_grid[i - 1], a[i - 1], a[i]      # a0 < target <= a1
        if a1 <= a0:
            return float(S)
        f = (target - a0) / (a1 - a0)
        return float(10.0 ** (np.log10(s0) + f * (np.log10(S) - np.log10(s0))))
    return float('inf')


def budget(protocol, S, m_tr, m_te, m_sel=0):
    """Tum kernel matrisini kurmak icin TOPLAM devre calistirmasi.

    `m_sel` = kriter (`gamma`) icin olculen SECIM seti. A1/A2 kriter kullanmaz
    -> `m_sel=0`; A3/A4/A7/A8 kullanir. Ayni anlik goruntuler havuzdaki TUM
    `M_H` operatoru besledigi icin ek maliyet operator sayisinda LOGARITMIK
    (bkz. 01_TEST_PROTOCOL.md 2.5) -- yani `m_sel` NOKTA sayisiyla dogrusal,
    havuz buyuklugunden bagimsiz sayilir. Kriter maliyetinin karari cevirip
    cevirmedigi G5'te ayrica raporlanir.
    """
    if protocol == 'shadow':
        return float(S) * (m_tr + m_te + m_sel)
    if protocol == 'fidelity':
        return float(S) * (m_tr * (m_tr - 1) / 2.0 + m_te * m_tr)
    raise ValueError(protocol)


def twoq_gates(n, depth, ent, protocol):
    """Devre basina 2-qubit kapi sayisi (ikinci maliyet ekseni).

    ZZ-feature map: katman basina `n(n-1)/2` ZZ dondurmesi, `ent=0` ise hic.
    Fidelity compute-uncompute -> 2 kat. Shadow'un rastgele olcum katmani TEK
    QUBIT'tir, 2q sayisina katki YOK.
    """
    g = depth * (n * (n - 1) // 2) if ent > 0 else 0
    return 2 * g if protocol == 'fidelity' else g


def weight_mix(w, idx):
    """Secimin ortalama agirligi + `3^w` cinsinden ortalama maliyeti.

    G3'un teshisi: A8'in dusuk agirliga COKUP cokmedigi (yani A2'ye yakinsayip
    yakinsamadigi) buradan gorulur -- `select_greedy_residual` docstring'inde
    on-kayitli olarak beklenen sonuc budur.
    """
    ww = np.asarray(w, dtype=float)[np.asarray(idx, dtype=int)]
    return float(ww.mean()), float(op_cost(ww).mean())




"""T1.1 cekirdegi -- GERCEK VERI (Kapi 6, dis gecerlilik).  TEK KAYNAK.

qk_core : devre, Pauli domeni, kernel, EC teshisi (Katman A/B/C)
phaseb  : etiketler, kol secimi (A1-A8), merge, KRR, istatistik, Katman D
gate4   : shot muhasebesi -- shadow varyansi, gurultulu kernel, S*, butce
t11     : gercek veri yukleme + on-isleme + M-bagimsiz EC kapisi + Huang `g`

BEKLENEN ORTAM: qk_core + phaseb_core + gate4_core'un tanimladigi her sey +
`np/xp/GPU/CDT/FDT/to_xp/to_np` (defterin 0. hucresi).

On-kayit: `01_TEST_PROTOCOL.md` 6.42 (K1-K13) + kilit sonrasi ek D1.
Buradaki her fonksiyon o bolumun bir maddesini uygular; madde numarasi
docstring'de yazilidir. Kilitten sonra kural degistirilmez -- kod kurala uyar.

================================ SENTETIKTEN FARK ============================
Sentetik tasarimin uc varsayimi gercek veride cokuyordu (6.42 tablosu):
  her tohumda YENI X   -> ayni havuzun yeniden bolunmesi   (K3)
  sans tam %50         -> egitim cogunluk orani            (K1)
  etiket ekseni        -> her setin tek gercek `y`'si      (K8)
Bu dosya bu ucunun karsiligini kurar; geri kalan boru hatti AYNEN devralinir.
==============================================================================
"""

# ============================================================================
# 1) VERI -- K8
# ============================================================================

DATASETS = ('ionosphere', 'breast_cancer', 'fashion_mnist')

# Fashion-MNIST (2026-10-12, T1.1 sonuclari GORULDUKTEN SONRA eklendi -- on-kayitli degil).
# Huang+ 2021 / Shaydulin & Wild 2021 / Wang-Qi-Wang 2024 cizgisi: elbise vs gomlek,
# egitim CSV'sinden sinif basina 400 (toplam 800, dengeli). Kaynak TEK: Kaggle
# `zalando-research/fashionmnist` -> `fashion-mnist_train.csv`; satir sirasi kaynaga
# bagli oldugu icin OpenML'e dusulmez (ayni tohum farkli ornek secerdi).
FM_CLASSES = (3, 6)      # 3 = Dress (+1), 6 = Shirt (-1)
FM_PER_CLASS = 400
FM_SEED = 20261012       # karar tarihi


def _find_local(name, roots=('/kaggle/input', '.', 'data')):
    """Veri dosyasini diskte ARA -- mount yolu VARSAYILMAZ.

    Kaggle dataset'leri her zaman ayni yere baglanmaz; yol hardcode edilirse
    kosu baska bir oturumda sessizce cokup `fetch_openml`e duser (internet
    kapaliysa oturum olur). `rglob` ile bulunur, bulunamazsa None.
    """
    import os
    pats = (name + '.csv', name + '.data', name.replace('_', '') + '.csv')
    for r in roots:
        if not os.path.isdir(r):
            continue
        for dp, _, fns in os.walk(r):
            for fn in fns:
                if fn.lower() in pats:
                    return os.path.join(dp, fn)
    return None


def load_real(name):
    """Doner `(X, y)`; `X` (N, d) float, `y` (N,) +-1.  K8.

    `breast_cancer` sklearn icinde gelir (indirme YOK). `ionosphere` once diskte
    aranir (Kaggle input), yoksa OpenML'den cekilir. `as_frame=False` -> pandas
    yolu hic acilmaz; bu ortam bagimliligini bilerek dusuk tutuyoruz.

    `y` uzerinde +-1 gosterimi zorunlu: `krr_multi` `sign` ile calisiyor ve
    `make_labels` da +-1 uretiyordu -- boru hattinin geri kalani degismesin.
    """
    if name == 'breast_cancer':
        from sklearn.datasets import load_breast_cancer
        d = load_breast_cancer()
        X, t = d.data.astype(float), d.target
    elif name == 'ionosphere':
        p = _find_local('ionosphere')
        if p is not None:
            import csv
            with open(p, newline='', encoding='utf-8') as f:
                rows = [r for r in csv.reader(f) if r]
            if not rows[0][-1].strip().lower() in ('g', 'b'):
                rows = rows[1:]                       # baslik satiri
            X = np.array([[float(v) for v in r[:-1]] for r in rows])
            t = np.array([1 if r[-1].strip().lower() == 'g' else 0 for r in rows])
        else:
            from sklearn.datasets import fetch_openml
            d = fetch_openml('ionosphere', version=1, as_frame=False, parser='liac-arff')
            X = np.asarray(d.data, dtype=float)
            t = np.asarray([str(v).strip().lower() in ('g', 'good', '1') for v in d.target])
    elif name == 'fashion_mnist':
        p = _find_local('fashion-mnist_train')
        if p is None:
            raise FileNotFoundError('fashion-mnist_train.csv yok -- Kaggle dataset '
                                    'zalando-research/fashionmnist eklenmeli')
        import csv
        a, b = FM_CLASSES
        rows = {a: [], b: []}
        with open(p, newline='', encoding='utf-8') as f:
            rd = csv.reader(f)
            next(rd)                                       # baslik: label,pixel1..784
            for r in rd:
                c = int(r[0])
                if c in rows:
                    rows[c].append([float(v) for v in r[1:]])
        rng = np.random.default_rng(FM_SEED)
        Xs, ts = [], []
        for c in (a, b):
            A = np.asarray(rows[c])
            Xs.append(A[np.sort(rng.choice(len(A), FM_PER_CLASS, replace=False))])
            ts.append(np.full(FM_PER_CLASS, 1 if c == a else 0))
        X, t = np.vstack(Xs), np.concatenate(ts)
    else:
        raise ValueError(name)
    y = np.where(np.asarray(t).astype(int) == 1, 1.0, -1.0)
    return np.asarray(X, dtype=float), y


def chance_level(y_tr):
    """K1 -- sans = EGITIM kumesindeki cogunluk sinifi orani (yuzde).

    Test kumesinin dagilimina BAKILMAZ: baksaydik esik (`D_LIVE`, `TAU2`) test
    verisinin bir fonksiyonu olurdu ve "sabit tahminci canli sayildi" hatasi
    olcum tarafindan uretilirdi. Dengeli dogruluk secilmedi -- onceki fazlarla
    karsilastirilabilirligi bozar (6.42 K1).
    """
    y = np.asarray(y_tr)
    return 100.0 * max(float(np.mean(y > 0)), float(np.mean(y <= 0)))


def stratified_splits(y, n_splits, seed0, train_frac=0.7):
    """K2 + K3 + K11 -- %70/30 tabakali bolme, `n_splits` tekrar, tohum `seed0+i`.

    Tabakali olmasinin sayisal gerekcesi (6.42 K2): `ionosphere`'de rastgele
    bolmede test cogunluk oraninin std'si ~3.9 puan, `D_LIVE=5` ile AYNI
    buyuklukte -> hicbir sey ogrenmeyen sabit tahminci canli gorunebilirdi.
    Tabakalida fark ~0.07 puan (P1 bunu assert eder).

    Tekrarlar BAGIMSIZ DEGILDIR (ayni havuz) -> bootstrap GA'lari iyimser
    olabilir; makalede acikca yazilir (K3).
    """
    from sklearn.model_selection import StratifiedShuffleSplit
    out = []
    for i in range(n_splits):
        ss = StratifiedShuffleSplit(n_splits=1, train_size=train_frac,
                                    random_state=seed0 + i)
        tr, te = next(ss.split(np.zeros(len(y)), np.asarray(y)))
        out.append((tr, te))
    return out


def preprocess(X_tr, X_te, n, seed=0):
    """K9 -- standardizasyon -> PCA(n) -> kantil -> U(0, 2pi).  SIRA KILITLI.

    Her adim YALNIZ egitim kumesinde fit edilir (P2 tek `fit` cagrisi sayar).

    Neden kantil (6.42 K9): `encode` periyodik degildir (`x_i·x_j` terimi,
    `qk_core.py:99`), yani `bw`'nin anlami girdinin OLCEGINE baglidir. EC-LEARN
    aileleri `X ~ U(0, 2pi)` ile kalibre edildi. Olculen fark: `U(0,2pi)`'de faz
    std'si tek-qubit 1.25 / cift terim 6.96; `N(0,1)`'de 0.70 / 0.32 -- cift
    (dolanikli) terim ~22x kuculuyor. Olcek eslenmezse "gercek veride EC yok"
    sonucu VERIDEN degil OLCEKTEN gelirdi.

    Bilinen sinir: PCA bilesenleri bagimsiz degildir; yalniz MARJINALLER
    eslenir, ortak dagilim degil. Bu yuzden K10 EC kapisini yeniden olcer.

    `ionosphere`'in 2. oznitelii sabittir (varyans 0) -> `StandardScaler` sifira
    bolmez (sklearn sabit sutunu 0'a esler), ama PCA oncesi bunu acikca test
    ediyoruz (P4).
    """
    from sklearn.preprocessing import StandardScaler, QuantileTransformer
    from sklearn.decomposition import PCA
    sc = StandardScaler().fit(X_tr)
    A, B = sc.transform(X_tr), sc.transform(X_te)
    pca = PCA(n_components=n, random_state=seed).fit(A)
    A, B = pca.transform(A), pca.transform(B)
    # n_quantiles <= egitim boyutu; esitken egitim marjinali ~birebir uniform
    qt = QuantileTransformer(n_quantiles=min(len(A), 1000),
                             output_distribution='uniform',
                             subsample=10**9, random_state=seed).fit(A)
    return qt.transform(A) * 2.0 * np.pi, np.clip(qt.transform(B), 0.0, 1.0) * 2.0 * np.pi


# ============================================================================
# 2) EC KAPISI -- K13: `M`-bagimsiz hale getirme
# ============================================================================
# `effrank_frac = effrank/M` esigi (0.90) ASLINDA bir `m2` esigidir ve `M` ile
# olceklenir. Kapali form (turetildi + sayisal dogrulandi, 6.42 K13):
#     effrank_frac = 1 / (1 + (M-1) m2),   m2 = kosegen-disi k^2 ortalamasi
# (fidelity kernel'de diag(K)=1 -> Sum(lam)=M, Sum(lam^2)=M + M(M-1) m2).
# Ayni kernel (`m2 = 3.5e-4`) M=400'de 0.877 (NO-EC), M=245'te 0.921 (EC-LEARN).
# Sabit 0.90 birakilirsa `ionosphere` GEVSEK kapidan gecer -> EC olmayan birimler
# G1'e girer -> BIZIM LEHIMIZE yanlilik. O yuzden `m2` cinsinden sabitlenir.

M_ANCHOR = 400          # kalibrasyon capasi: esigin konuldugu Faz A2 grid'i
FRAC_ANCHOR = 0.90      # `EFFRANK_GATE`, Faz A2'den beri degismedi


def m2_star(anchor_M=M_ANCHOR, anchor_frac=FRAC_ANCHOR):
    """Kalibre edilmis esigin `m2` karsiligi: (1/0.90 - 1)/399 = 2.785e-4."""
    return (1.0 / anchor_frac - 1.0) / (anchor_M - 1.0)


def effrank_gate_for(M):
    """`M` boyutundaki bir kernel icin esdeger `effrank_frac` esigi.

    M=400 -> 0.9000 (sentetik kararlar BIREBIR ayni) · M=398 -> 0.9005 ·
    M=245 -> 0.9364.  Esik gercek veriye BAKILARAK secilmiyor; mevcut
    kalibrasyondan ve cebirden turetiliyor -> on-kayit ihlali DEGIL (K13).
    """
    return 1.0 / (1.0 + (M - 1.0) * m2_star())


def frac_to_m2(frac, M):
    """Ters donusum -- raporlamada `m2`'yi `M`'den bagimsiz gostermek icin."""
    return (1.0 / max(frac, 1e-12) - 1.0) / (M - 1.0)


# ============================================================================
# 2.5) KLASIK REFERANS -- RBF
# ============================================================================
# `runs/g1rerun_full_v1/classical_vs_quantum.py`'nin protokolu devralinir
# (medyan sezgisi x carpan izgarasi), TEK farkla: bant genisligi burada LOO ile
# secilir. Gerekce: D1.1 `K_C`'yi "boru hattindaki RBF, LOO-secili bant
# genisligiyle" diye tanimliyor; test kumesine bakarak secilen bir `gamma` ile
# `g` hesaplamak metrigi test verisine baglardi.
#
# Eski kosuyla kiyaslanabilirlik icin defter AYRICA test-en-iyisini de raporlar
# (`rbf_best_test`): o sayi klasigin LEHINE yanli, yani bizim ALEYHIMIZE --
# muhafazakar taraf.

RBF_MULTS = (0.1, 0.25, 0.5, 1.0, 2.0, 4.0, 10.0)


def rbf_K(A, B, gamma):
    d2 = (A*A).sum(1)[:, None] + (B*B).sum(1)[None, :] - 2.0*A @ B.T
    return np.exp(-gamma * np.maximum(d2, 0.0))


def rbf_gamma_grid(X_tr, mults=RBF_MULTS):
    """Medyan sezgisi: `gamma = mult / medyan(||x-x'||^2)`."""
    Z = X_tr[:200]
    med2 = float(np.median(((Z[:, None] - Z[None, :])**2).sum(-1)))
    return [m / max(med2, 1e-12) for m in mults]


def loo_lambda(K, y, lams=None):
    """`(lam, loo_acc)` -- egzakt LOO ile secilen ridge parametresi.

    NEDEN AYRI BIR FONKSIYON: `phaseb_core.krr_multi` bu secimi zaten yapiyor
    ama `lam`'i DISARI VERMIYOR, yalniz test dogrulugunu donduruyor. D1.1 `s_K`
    icin "o kolun KENDI LOO-secili lambda'si" gerekiyor. `krr_multi`'nin imzasini
    degistirmek Faz A-D'nin ALTI yayimlanmis kosusunun ortak yolunu oynatir --
    on kadar satiri burada tekrar etmek o riskten ucuz. Kapali form AYNI:
        H = U diag(ev/(ev+lam)) U^T,  yhat_i^{-i} = (yhat_i - h_i y_i)/(1 - h_i)
    """
    lams = LAMS if lams is None else lams
    A = np.asarray(to_np(K), dtype=float)
    yv = np.asarray(to_np(y), dtype=float).ravel()
    ev, U = np.linalg.eigh((A + A.T)/2.0)
    ev = np.clip(ev, 0.0, None)
    UtY = U.T @ yv
    U2 = U**2
    best = (None, -1.0)
    for lam in lams:
        f = ev/(ev + lam)
        h = U2 @ f
        loo = (U @ (f*UtY) - h*yv)/np.maximum(1.0 - h, 1e-6)
        a = float(np.mean(np.sign(loo) == yv))
        if a > best[1]:
            best = (lam, a)
    return best


def rbf_select_loo(X_tr, y_tr, mults=RBF_MULTS, lams=None):
    """`(gamma, lam, loo_acc)` -- bant genisligi + ridge, ikisi de LOO ile."""
    best = (None, None, -1.0)
    for g in rbf_gamma_grid(X_tr, mults):
        lam, a = loo_lambda(rbf_K(X_tr, X_tr, g), y_tr, lams)
        if a > best[2]:
            best = (g, lam, a)
    return best


# ============================================================================
# 3) HUANG METRIKLERI -- kilit sonrasi ek D1 (ikincil, BAGLAYICI DEGIL)
# ============================================================================
# Kaynak: Huang+ 2021 (arXiv:2011.01938), Ek F/G. Formuller PDF'ten dogrulandi
# (papers/Huang_PowerOfData_2011.01938.pdf, 2026-09-17).
# UYARI (bilgi notundan): EC altinda K_Q ~ I -> `g` BUYUK cikar ama `s_Q ~ N`
# oldugu icin kuantum model hicbir sey ogrenemez. `g` TEK BASINA OKUNMAZ;
# her zaman `s_Q`, `s_C` ile birlikte raporlanir (D1.2 md.4).

# D2 (2026-09-21, Latif onayi): izgara ASAGI uzatildi. Huang'in KURALI
# degismedi ("g_tra<=0.045 saglayan en buyuk lam"), yalniz aday kumesi genisledi.
# Gerekce OLCULDU (hicbir `g` DEGERI gorulmeden, yalniz tanimsizlik gorulerek):
#   ionosphere RBF Gram, iz-normalize: ozdeger_min = 4.9e-07, kosul sayisi 3.2e+08
#   g_tra ~ lam/(ozdeger_min+lam)  ->  sart lam <= 0.045*ozdeger_min ~ 2.2e-08
# Eski izgaranin en kucugu 1e-05, yani gerekenden ~3 mertebe buyuktu -> `g` TUM
# birimlerde tanimsiz cikiyordu (6/6 duman testinde). Bu, verinin degil izgaranin
# ozelligiydi. UYARI: bu kadar kucuk `lam`'da regularize `g` ham `g`'ye yaklasir
# -- regularizasyonun onlemek istedigi seye. Bu yuzden `g` YANINDA daima
# `K_C`'nin en kucuk ozdegeri ve kosul sayisi raporlanir (defter Bolum 8).
LAM_GRID_G = (1e-9, 1e-8, 1e-7, 1e-5, 1e-4, 1e-3, 0.01, 0.025, 0.05, 0.1)
G_TRA_MAX = 0.045       # Huang eq. F20 / L9-L10 -- DEGISMEDI


def trace_normalize(K, N=None):
    """`K <- N K / Tr(K)` -- Huang Thm.1'in normalizasyon sarti (D1.1).

    A7 gibi kosegeni 1 OLMAYAN kollar icin zorunlu; atlanirsa `g` kollar arasi
    kiyaslanabilir olmaz (olcek farki metrige sizar).
    """
    A = np.asarray(to_np(K), dtype=float)
    N = A.shape[0] if N is None else N
    tr = float(np.trace(A))
    return A * (N / tr) if tr > 0 else A


def _psd_sqrt(A):
    """Simetrik PSD karekok -- `scipy.linalg.sqrtm` DEGIL.

    `sqrtm` genel (kompleks olabilen) yol; burada matrisler simetrik PSD ve
    M <= 400, yani `eigh` hem ucuz hem gercek-degerli kaliyor. Negatif ozdegerler
    sayisal artiktir, sifira kirpilir.
    """
    ev, V = np.linalg.eigh((A + A.T) / 2.0)
    return (V * np.sqrt(np.clip(ev, 0.0, None))) @ V.T


def _lmax(A):
    return float(np.max(np.linalg.eigvalsh((A + A.T) / 2.0)))


def geometric_difference(K_Q, K_C, lam_grid=LAM_GRID_G, g_tra_max=G_TRA_MAX):
    """Regularize geometrik fark `g` + secilen `lambda`.  D1.1.

        g(lam)     = sqrt(|| sqrt(K_Q) sqrt(K_C) (K_C+lam I)^-2 sqrt(K_C) sqrt(K_Q) ||_inf)
        g_tra(lam) = lam * sqrt(|| sqrt(K_Q) (K_C+lam I)^-2 sqrt(K_Q) ||_inf)

    `lambda` ETIKETE BAKMADAN secilir: `g_tra <= 0.045` saglayan EN BUYUK `lam`
    (Huang eq. F20). Hicbiri saglamazsa `g` TANIMSIZ raporlanir -- doldurulmaz,
    en yakin degere yuvarlanmaz (D1.1 son madde).

    Ham `K_C^-1` KULLANILMAZ: RBF Gram matrisi neredeyse tekildir, ham ters
    `g`'yi yapay sisirir.

    Doner: `(g, lam)` ya da `(nan, nan)`.
    """
    Q = trace_normalize(K_Q)
    C = trace_normalize(K_C)
    sQ, sC = _psd_sqrt(Q), _psd_sqrt(C)
    ev, V = np.linalg.eigh((C + C.T) / 2.0)
    ev = np.clip(ev, 0.0, None)
    best = (np.nan, np.nan)
    for lam in sorted(lam_grid):                       # artan -> en buyugu kalir
        R = (V * (1.0 / (ev + lam) ** 2)) @ V.T        # (K_C + lam I)^-2
        if lam * np.sqrt(max(_lmax(sQ @ R @ sQ), 0.0)) <= g_tra_max:
            best = (float(np.sqrt(max(_lmax(sQ @ sC @ R @ sC @ sQ), 0.0))), float(lam))
    return best


def spectrum_health(K):
    """`(ozdeger_min, kosul_sayisi)` -- iz-normalize `K` icin.

    `g` ile BIRLIKTE raporlanmasi sart (D2): kucuk `lam`'da regularize `g` ham
    `g`'ye yaklasir, yani sayinin ne kadar guvenilir oldugu `K_C`'nin kosul
    sayisina bagli. Tek basina `g` bunu gizler.
    """
    ev = np.clip(np.linalg.eigvalsh(trace_normalize(K)), 0.0, None)
    lo, hi = float(ev.min()), float(ev.max())
    return lo, (hi/lo if lo > 0 else float('inf'))


def model_complexity(K, y, lam):
    """`s_K = y^T (K+lam I)^-1 K (K+lam I)^-1 y` -- egitilmis modelin ||w||^2.  D1.1.

    Huang eq. F1. `lam` = o kolun KENDI LOO-secili degeri (`phaseb_core.krr_multi`),
    `g`'nin `lam`'i DEGIL: biri model karmasikligini, digeri geometrik farki
    olcuyor ve ayri secim kurallari var.

    Test hatasi kabaca `sqrt(s_K/N)` ile sinirli; `s_K ~ N` ise model hicbir sey
    ogrenmemis demektir (EC'nin imzasi).
    """
    A = trace_normalize(K)
    yv = np.asarray(to_np(y), dtype=float).ravel()
    B = np.linalg.solve(A + lam * np.eye(A.shape[0]), yv)
    return float(B @ A @ B)




def enc(X, n, depth, ent, bw):
    return encode(X, n, depth, ent, bw=bw, norm=NORM)


def batch_for(n):
    # grup x 2^n <= 2^25 genlik (~256 MB complex64). n <= 16'da 512 (T1.1 ile ayni);
    # n = 20'de 32 -- 512 ile kodlama ara bellekleri ~17 GB'a cikiyordu.
    return int(min(512, max(8, 2 ** 25 >> n)))


def pauli_expect_batched(X, n, depth, ent, bw, layout, batch=None):
    batch = batch or batch_for(n)
    out = []
    for i in range(0, len(X), batch):
        psi = enc(X[i:i+batch], n, depth, ent, bw)
        out.append(pauli_expect_fast(psi, n, layout))
        del psi; free_mem()
    return xp.concatenate(out, axis=0)


def fidelity_blocks(Xa, Xb, n, depth, ent, bw, batch=None):
    batch = batch or batch_for(n)
    def st(X):
        return [enc(X[i:i+batch], n, depth, ent, bw) for i in range(0, len(X), batch)]
    Sa = st(Xa)
    Sb = Sa if Xb is Xa else st(Xb)          # egitim-egitim blogu: bir kez kodla
    K = xp.concatenate([xp.concatenate([xp.abs(a.conj() @ b.T)**2 for b in Sb], 1)
                        for a in Sa], 0)
    del Sa, Sb; free_mem()
    return K


def encoded_features_and_fidelities(Xt, Xe, n, depth, ent, bw, layout, batch=None):
    """Xt ve Xe durum vektorlerini YALNIZ BIR KEZ kodlar; ayni durumlardan hem
    Pauli beklentilerini (`Et, Ee`) hem de A0 fidelity matrislerini (`Ktt, Ket`) uretir.
    Sonuclar `pauli_expect_batched` + `fidelity_blocks` ile bit-bit aynidir;
    ancak devre kodlama sayisi yariya iner (n=20'de sureyi ~%47 kisaltir).
    """
    batch = batch or batch_for(n)
    St = [enc(Xt[i:i+batch], n, depth, ent, bw) for i in range(0, len(Xt), batch)]
    Se = [enc(Xe[i:i+batch], n, depth, ent, bw) for i in range(0, len(Xe), batch)]
    Et = xp.concatenate([pauli_expect_fast(psi, n, layout) for psi in St], axis=0)
    Ee = xp.concatenate([pauli_expect_fast(psi, n, layout) for psi in Se], axis=0)
    Ktt = xp.concatenate([xp.concatenate([xp.abs(a.conj() @ b.T)**2 for b in St], 1)
                          for a in St], 0)
    Ket = xp.concatenate([xp.concatenate([xp.abs(a.conj() @ b.T)**2 for b in St], 1)
                          for a in Se], 0)
    del St, Se; free_mem()
    return Et, Ee, Ktt, Ket
```

## Bolum 3 - DOGRULAMA: devralinan testler + **P1-P6** (gecmezse ASAGI DEVAM ETME)

`[1]`-`[9]` Kapi 4 defterinden devralindi (muhasebe tutarliligi, shot varyansi,
`s_star` semantigi, para birimi). `P1`-`P6` on-kayitin pilot kontrolleridir
(6.42 K12) ve **gercek veri uzerinde** koşar.

```python
def _validate():
    # [0] on-kayit hucresindeki `DS` ile cekirdegin `DATASETS`'i AYNI olmali;
    #     ayrisirlarsa on-kayit ile kosan kod farkli veri setlerini gosterir.
    assert set(DS) <= set(DATASETS), (DS, DATASETS)
    print('[0] OK  on-kayit DS, t11_core.DATASETS icinde')
    for ds in DS:
        X_, y_ = load_real(ds)
        tr_, _ = stratified_splits(y_, 1, SEED0, TRAIN_FRAC)[0]
        print(f'    {ds:<14} N={len(y_):<4} d={X_.shape[1]:<3} M_tr={len(tr_):<4} '
              f'sans={chance_level(y_[tr_]):.1f}%  EC kapisi={effrank_gate_for(len(tr_)):.4f}')

    rng = np.random.default_rng(0); n = 6; M = 300
    X = rng.uniform(0, 2*np.pi, size=(M, n))
    lay = pool_layout(n, 3)
    E = pauli_expect_batched(X, n, 2, 1.0, 0.3, lay, batch=64)
    wp = np.asarray(lay[3])

    # ---- devralinan cekirdek testleri --------------------------------------
    Ea = pauli_expect_fast(enc(X, n, 2, 1.0, 0.3), n, lay)
    assert np.abs(to_np(E-Ea)).max() < 1e-4
    print('[1] OK  parcali Pauli yolu == tek seferlik')

    for H in (2, 3):
        m = wp <= H
        _, s2_ref = kernel_gfqk(E, wp, H, 0.0)
        assert abs(sigma2_arm(E[:, to_xp(m)], op_cost(wp[m])) - s2_ref)/s2_ref < 1e-4
    print('[2] OK  sigma2_arm == kernel_gfqk.s2  (H=2,3)')

    idx = select_topp(capacity(E), 16)
    F = E[:, to_xp(np.asarray(idx))]; v = dir_var(wp[idx])
    Kc0 = to_np(kernel_pair(F, F)); off = ~np.eye(M, dtype=bool)
    for S in (1e3, 1e4):
        d = []
        for _ in range(20):
            G, dg = noisy_features(F, v, S, rng)
            d.append(to_np(kernel_pair(G, G, dg, dg))[off] - Kc0[off])
        r = float(np.var(np.concatenate(d))) / (sigma2_arm(F, v)/S)
        assert 0.5 < r < 2.0, (S, r)
    print(f'[3] OK  ampirik/analitik shot varyansi = {r:.2f}')

    assert np.allclose(dir_var(wp[idx], np.eye(16), 'l1'), 3.0**wp[idx])
    Cm, _ = greedy_merge(E, select_topp(capacity(E), 32), 16)
    v1 = dir_var(wp[select_topp(capacity(E), 32)], Cm, 'l1')
    v2 = dir_var(wp[select_topp(capacity(E), 32)], Cm, 'l2')
    assert np.all(v1 >= v2 - 1e-9)
    print(f'[4] OK  merge varyansi l1>=l2  (medyan bedel {np.median(v1/v2):.2f}x)')

    Kf, _ = kernel_fidelity(enc(X[:120], n, 2, 1.0, 0.3))
    K0 = to_np(Kf)
    d = [to_np(noisy_fidelity(Kf, 1e3, rng)) - K0 for _ in range(40)]
    r = float(np.var(np.concatenate([x.ravel() for x in d])))/(np.mean(K0*(1-K0))/1e3)
    assert 0.5 < r < 2.0, r
    Ks = to_np(noisy_fidelity(Kf, 1e3, rng, symmetric=True))
    assert np.allclose(Ks, Ks.T)
    print(f'[5] OK  A0 binom varyansi k(1-k)/S (oran {r:.2f}), egitim blogu simetrik')

    g = [10, 100, 1000, 10000]
    assert s_star(g, [60, 40, 40, 60], 55, interp=False) == 10000.0
    assert s_star(g, [40, 40, 40, 40], 55) == float('inf')
    assert abs(s_star(g, [40, 40, 60, 60], 55) - 10**2.75) < 1e-6
    assert (S_GRID[1]/S_GRID[0]) > G1_MIN_RATIO, 'grid adimi esikten kucuk'
    print('[6] OK  s_star monotonluk + log-ara deger')

    Eo = np.linalg.qr(rng.normal(size=(M, 4)))[0]*np.sqrt(M)
    Eo = to_xp(Eo*np.array([3.0, 2.9, 2.8, 2.7]), FDT)
    w4 = np.array([4, 1, 1, 1])
    assert 0 in set(map(int, select_greedy_residual(Eo, 2)))
    assert 0 not in set(map(int, select_greedy_residual(Eo, 2, cost=op_cost(w4))))
    print('[7] OK  A8 pahali yonu atliyor')

    # [8] PARA BIRIMI -- `m_sel` artik `M_tr` (K4), kapali form dogrulanir
    mtr, mte = 245, 106
    bs, bf = budget('shadow', 1, mtr, mte), budget('fidelity', 1, mtr, mte)
    assert abs(bf/bs - (mtr*(mtr-1)/2 + mte*mtr)/(mtr+mte)) < 1e-9 and bf/bs > 10
    assert budget('shadow', 1, mtr, mte, m_sel=mtr) == bs + mtr
    print(f'[8] OK  fidelity/shadow devre orani = {bf/bs:.0f}x  (M={mtr+mte})')

    a = {'A0': 51.0, 'A1_random': 51.0, 'A2_hbody': 51.0, 'A3_topp': 51.0,
         'A4_merge': 51.0, 'A7_resid': 51.0, 'A8_cost': 57.0, 'A5_super': 58.0}
    assert layer_d_strict(a, 50.0)[2] is True
    print('[9] OK  A8 etiketsiz kol olarak KATI ALTIN sayiliyor')

    # [10] v2 -- Gan+'a sadik H-body: yalniz TAM agirlik, I yok; havuzda I yok
    wv = np.array([1, 1, 1, 2, 2, 2, 2, 3])
    r10 = np.random.default_rng(0)
    s10 = select_hbody_exact(wv, 2, 3, r10, 5)
    assert len(s10) == 5 and all(set(wv[x]) == {2} and len(set(x)) == 3 for x in s10)
    assert select_hbody_exact(wv, 1, 4, r10) == [] and len(select_hbody_exact(wv, 1, 3, r10)) == 1
    w6 = np.asarray(pool_layout(6, 2)[3])
    assert 0 in set(w6) and 0 not in set(w6[(w6 <= 2) & (w6 >= 1)])
    assert all(0 not in set(wv[x]) for x in select_hbody(wv, 3, r10, 5))
    print('[10] OK  Gan+ H-body tam agirlikta, I hicbir havuzda yok')

    # ---- P1-P6: ON-KAYITIN PILOT KONTROLLERI (6.42 K12), GERCEK VERI ---------
    for ds in DS:
        Xr, yr = load_real(ds)
        sp = stratified_splits(yr, N_SPLIT, SEED0, TRAIN_FRAC)

        # P1 -- tabakali bolme
        dev = max(abs(chance_level(yr[tr]) - chance_level(yr[te])) for tr, te in sp)
        assert dev < 1.0, (ds, dev)

        # P2 -- sizinti yok: indeks kesisimi bos, on-isleme YALNIZ egitimde fit
        for tr, te in sp:
            assert len(np.intersect1d(tr, te)) == 0, ds
        tr0, te0 = sp[0]
        A1_, _ = preprocess(Xr[tr0], Xr[te0], 8)
        A2_, _ = preprocess(Xr[tr0], Xr[te0][::-1], 8)
        assert np.allclose(A1_, A2_), f'{ds}: egitim donusumu test kumesine bagli'

        # P3 -- kantil cikisi [0,2pi] ve egitim marjinali uniform (KS < 0.05)
        assert A1_.min() >= 0.0 and A1_.max() <= 2*np.pi + 1e-9, ds
        ks = max(float(np.max(np.abs(np.sort(A1_[:, j])/(2*np.pi)
                                     - (np.arange(len(A1_))+0.5)/len(A1_))))
                 for j in range(A1_.shape[1]))
        assert ks < 0.05, (ds, ks)

        # P4 -- sabit oznitelik (ionosphere 2. sutun) NaN/inf uretmiyor
        assert np.isfinite(A1_).all(), ds
        if ds == 'ionosphere':
            assert float(Xr[tr0][:, 1].std()) == 0.0, 'sabit sutun kaybolmus -> test bos'

        # P6 -- sans EGITIM cogunlugu (50 DEGIL) ve bolmeler arasi kararli
        chs = [chance_level(yr[tr]) for tr, _ in sp]
        if abs(float(np.mean(yr > 0)) - 0.5) > 0.01:      # dengeli sette (FM) sans zaten 50
            assert all(abs(c - 50.0) > 1.0 for c in chs), ds
        assert max(chs) - min(chs) < 1.0, (ds, max(chs)-min(chs))
        print(f'[P1-P4,P6] OK  {ds:<14} sans={np.mean(chs):.1f}%  '
              f'egitim/test sapma {dev:.3f}p  KS={ks:.4f}')

    # P5 -- sayisal saglik: gercek veride tek birim, hicbir kolda NaN yok
    Xr, yr = load_real(DS[0])
    tr0, te0 = stratified_splits(yr, 1, SEED0, TRAIN_FRAC)[0]
    At, Ae = preprocess(Xr[tr0], Xr[te0], 8)
    lay8 = pool_layout(8, 4, max_flips=max_flips_for(8), seed=SEED0)
    Et_ = pauli_expect_batched(At, 8, 8, 1.0, 0.15, lay8)
    Ee_ = pauli_expect_batched(Ae, 8, 8, 1.0, 0.15, lay8)
    w8 = np.asarray(lay8[3]); mi = np.where((w8 <= 4) & (w8 >= 1))[0]
    ii = to_xp(np.asarray(select_topp(capacity(Et_[:, to_xp(mi)]), 16)))
    Ft, Fe = Et_[:, to_xp(mi)][:, ii], Ee_[:, to_xp(mi)][:, ii]
    R = krr_multi(kernel_pair(Ft, Ft), kernel_pair(Fe, Ft), {'y': yr[tr0]}, {'y': yr[te0]})
    assert np.isfinite(R['y'][0]), 'NaN dogruluk'
    gg, ll, la = rbf_select_loo(At, yr[tr0])
    assert np.isfinite(gg) and np.isfinite(la)
    print(f'[P5] OK  gercek veri boru hatti NaN uretmiyor '
          f'(A3 test={R["y"][0]:.1f}%, RBF LOO={100*la:.1f}%)')

    print('\nDOGRULAMA + P1-P6 GECTI')


_validate()
```

## Bolum 4 - KOSU

`X_sel = X_tr` (K4): sentetikte ayri `N_SEL=1000` nokta vardi, gercek veride

```python
import pickle, shutil, pathlib

RUN_BUDGET_H = float(os.environ.get('HOPE_BUDGET_H', 11.25))   # Kaggle siniri 12 sa
DEADLINE = T_NB0 + RUN_BUDGET_H*3600
SPLIT_EST = {}           # n -> son bolme suresi x 1.15

# Onceki parcalarin checkpoint'leri: Kaggle'da `kernel_sources` -> /kaggle/input
for f in pathlib.Path('/kaggle/input').rglob('*'):
    if f.is_file() and f.parent.name in (CKPT, KDIR):
        dst = os.path.join(f.parent.name, f.name)
        if not os.path.exists(dst):
            shutil.copy(f, dst)


def run_unit(ds, X, y, n, depth, ent, bw, splits, log=None):
    # v3: agirlik <= 2 flip'lerinin HEPSI, yalniz 3-4 flip'liler kirpilir (denetim md. 4)
    layout = pool_layout(n, max(H_LIST), max_flips=max_flips_for(n), seed=SEED0, keep_low=2)
    w_pool = np.asarray(layout[3])
    key = unit_key(ds, n, depth, ent, bw)
    PER = [None]*len(splits)

    for si, (tr, te) in enumerate(splits):
        sp = os.path.join(CKPT, f'{key}__s{si:02d}.pkl')
        if os.path.exists(sp):
            with open(sp, 'rb') as fh:
                PER[si] = pickle.load(fh)
            continue
        if time.time() + SPLIT_EST.get(n, 0) > DEADLINE:
            continue                                           # sonraki parcaya kalir
        ts = time.time()
        accs = defaultdict(list); corr = defaultdict(list); sig = defaultdict(list)
        mix  = defaultdict(list); ec   = defaultdict(list); d1  = defaultdict(list)
        cls  = defaultdict(list); meta = defaultdict(list); XSAVE = {}
        Xt, Xe = preprocess(X[tr], X[te], n, seed=SEED0)      # K9, fit YALNIZ egitimde
        yt, ye = y[tr], y[te]
        Yt, Ye = {'y': yt}, {'y': ye}
        CH = chance_level(yt)                                  # K1
        meta['chance'].append(CH); meta['m_tr'].append(len(tr)); meta['m_te'].append(len(te))
        nrng = np.random.default_rng(SEED0 + 7717*si + 13*n + int(100*bw))
        rng  = np.random.default_rng(SEED0 + 9973*si + n + int(1000*bw))

        # ---- klasik referans: veriye bagli, DEVRE PARAMETRELERINDEN bagimsiz ----
        # ayni donusturulmus `X` (K9 "tek fark kernel" ilkesi)
        g_rbf, lam_rbf, loo_rbf = rbf_select_loo(Xt, yt)
        K_C = rbf_K(Xt, Xt, g_rbf)
        acc_rbf = krr_multi(to_xp(K_C, FDT), to_xp(rbf_K(Xe, Xt, g_rbf), FDT),
                            Yt, Ye, lams=(lam_rbf,))['y'][0]
        best_te = max(krr_multi(to_xp(rbf_K(Xt, Xt, gg), FDT),
                                to_xp(rbf_K(Xe, Xt, gg), FDT), Yt, Ye)['y'][0]
                      for gg in rbf_gamma_grid(Xt))
        cls['rbf_loo'].append(acc_rbf); cls['rbf_best_test'].append(best_te)
        cls['rbf_gamma'].append(g_rbf); cls['rbf_lam'].append(lam_rbf)
        cls['rbf_loo_acc'].append(100*loo_rbf)
        s_C = model_complexity(K_C, yt, lam_rbf) if D1_ON else np.nan
        # D2: `g` yaninda K_C'nin spektral sagligi DAIMA raporlanir -- kucuk
        # `lam`'da regularize `g` ham `g`'ye yaklasir, guvenilirligi kosula bagli.
        eig_lo, eig_cond = spectrum_health(K_C)
        cls['KC_eigmin'].append(eig_lo); cls['KC_cond'].append(eig_cond)

        Et, Ee, Ktt, Ket = encoded_features_and_fidelities(Xt, Xe, n, depth, ent, bw, layout)
        Es = Et                                                # K4

        # ---------- A0: H/p'den BAGIMSIZ ----------
        gate = effrank_gate_for(Ktt.shape[0])                  # K13
        d0 = diagnose(Ktt, budgets=(BUDGET,))
        st0, _, _ = layers_ac(d0, BUDGET, MEAN_GATE, gate)
        st_fix, _, _ = layers_ac(d0, BUDGET, MEAN_GATE, EFFRANK_GATE_FIXED)
        ec['status'].append(st0); ec['status_fixed'].append(st_fix)
        ec['effrank_frac'].append(d0['effrank_frac']); ec['gate'].append(gate)
        ec['m2'].append(frac_to_m2(d0['effrank_frac'], Ktt.shape[0]))
        ec['offdiag_mean'].append(d0['offdiag_mean']); ec['s_req_A0'].append(d0['s_req'])
        if D1_ON:
            try:
                _kq0 = to_np(Ktt)
                if not np.all(np.isfinite(_kq0)) or not np.all(np.isfinite(K_C)):
                    raise ValueError('non-finite kernel')
                lamA0, _ = loo_lambda(_kq0, yt)
                gA0, lgA0 = geometric_difference(_kq0, K_C)
                sQA0 = model_complexity(_kq0, yt, lamA0) if (lamA0 is not None and np.isfinite(lamA0)) else float('nan')
            except (np.linalg.LinAlgError, ValueError, TypeError) as _e:
                gA0, lgA0, sQA0 = float('nan'), float('nan'), float('nan')
            d1['A0|g'].append(gA0); d1['A0|lam_g'].append(lgA0)
            d1['A0|s_Q'].append(sQA0)
        for S in S_GRID:
            kt = Ktt if np.isinf(S) else noisy_fidelity(Ktt, S, nrng, symmetric=True)
            ke = Ket if np.isinf(S) else noisy_fidelity(Ket, S, nrng)
            a, c = krr_multi(kt, ke, Yt, Ye)['y']
            for H in H_LIST:
                for p in P_LIST:
                    accs[f'{H}|{p}|A0|{S}'].append(a)
                    if np.isinf(S):
                        corr[f'{H}|{p}|A0'].append(c)
            if not np.isinf(S):
                del kt, ke; free_mem()
        K_A0 = to_np(Ktt).astype(np.float32) if si == 0 else None
        del Ktt, Ket; free_mem()

        # ---------- Lego kollari ----------
        # KERNEL ARSIVI (D1.3 + D2): matrisler YALNIZ 0. bolme icin yazilir
        # (~350 MB; 30 bolmenin hepsi ~10 GB olurdu). Kaybedilen tek sey, ILERIDE
        # yeni bir kernel-duzeyi metrigini koşuyu tekrarlamadan 30 bolmede
        # hesaplayabilmek -- yani GA. `g`/`s_Q`/`s_C` degerleri zaten TUM
        # bolmeler icin `results.json`'a giriyor.
        # Buna karsilik on-islenmis `X_tr`/`y_tr` TUM bolmeler icin saklanir
        # (~0.6 MB/birim): PCA isaret konvansiyonu sklearn surumune bagli oldugu
        # icin, X olmadan "yeniden uret" birebir ayni kernel'i garanti etmez.
        XSAVE[f's{si}_Xtr'] = Xt.astype(np.float32)
        XSAVE[f's{si}_ytr'] = yt.astype(np.int8)
        KSAVE = {'K_C': K_C.astype(np.float32)} if si == 0 else None
        if KSAVE is not None and K_A0 is not None:
            KSAVE['A0'] = K_A0
        for H in H_LIST:
            # v2: I (agirlik 0) HICBIR kolun havuzunda yok -- bilgi tasimaz, Gan+ H-body'de de yok
            mi = np.where((w_pool <= H) & (w_pool >= 1))[0]
            if len(mi) < min(P_LIST):
                continue
            ix_H = to_xp(mi)
            Es_H, Et_H, Ee_H = Es[:, ix_H], Et[:, ix_H], Ee[:, ix_H]
            gam = capacity(Es_H); wH = w_pool[mi]; cst = op_cost(wH)

            for p in P_LIST:
                if p > len(mi):
                    continue
                sels = []
                for idx in select_random(len(mi), p, rng, N_DRAWS):
                    sels.append(('A1_random', np.asarray(idx), None))
                for idx in select_hbody(wH, p, rng, N_DRAWS):          # eski kol, yeni adi
                    sels.append(('A2_low', np.asarray(idx), None))
                for h, tag in ((1, 'A2_hb1'), (2, 'A2_hb2')):         # Gan+'a sadik (tam h)
                    for idx in select_hbody_exact(wH, h, p, rng, N_DRAWS):
                        sels.append((tag, np.asarray(idx), None))
                s3 = select_topp(gam, p);              sels.append(('A3_topp', np.asarray(s3), None))
                s7 = select_greedy_residual(Es_H, p);  sels.append(('A7_resid', np.asarray(s7), None))
                s8 = select_greedy_residual(Es_H, p, cost=cst)
                sels.append(('A8_cost', np.asarray(s8), None))
                cand = select_topp(gam, min(2*p, len(mi)))
                Cm, dist = greedy_merge(Es_H, cand, p)
                sels.append(('A4_merge', np.asarray(cand), Cm))
                # v3: oracle da gurultulu taramaya girer (eskiden yalniz S=inf -> S*=inf YAPI GEREGIYDI)
                sels.append(('A5_super', np.asarray(select_kta(Es_H, yt, p)), None))

                packs = []
                for tag, idx, Cc in sels:
                    ii = to_xp(idx)
                    if Cc is None:
                        Ft, Fe_ = Et_H[:, ii], Ee_H[:, ii]
                        v1 = dir_var(wH[idx]); v2 = v1
                    else:
                        Cx = to_xp(Cc, FDT)
                        Ft, Fe_ = Et_H[:, ii] @ Cx, Ee_H[:, ii] @ Cx
                        v1 = dir_var(wH[idx], Cc, 'l1')
                        v2 = dir_var(wH[idx], Cc, 'l2')
                    packs.append((tag, Ft, Fe_, v1))
                    s2 = sigma2_arm(Ft, v1)
                    # S_req / D1 icin gurultusuz kernel; LIN'de ortalama kosegene olcekli dogrusal kernel
                    # (`sigma2_arm` da ortalama kosegenle normalize -> ayni para birimi)
                    K0 = (kernel_pair(Ft, Ft) if NOISY_KERNEL != 'LIN'
                          else (Ft @ Ft.T) / xp.mean(xp.sum(Ft**2, 1)))
                    off = K0[~xp.eye(K0.shape[0], dtype=bool)]
                    vr = float(to_np(off.var()))
                    sig[f'{H}|{p}|{tag}|sigma2'].append(s2)
                    sig[f'{H}|{p}|{tag}|sigma2_l2'].append(sigma2_arm(Ft, v2))
                    sig[f'{H}|{p}|{tag}|s_req'].append(s2/vr if vr > 0 else np.inf)
                    mw, mc = weight_mix(wH, idx)
                    mix[f'{H}|{p}|{tag}|w'].append(mw); mix[f'{H}|{p}|{tag}|cost'].append(mc)
                    # ---- D1: g, s_Q. A1/A2 cok cekimlidir -> her cekilis ayri
                    # girdi olur ve ortalanir; KSAVE'e yalniz ILK cekilis yazilir.
                    if D1_ON:
                        try:
                            kq = to_np(K0)
                            if not np.all(np.isfinite(kq)) or not np.all(np.isfinite(K_C)):
                                raise ValueError('non-finite kernel')
                            lam_a, _ = loo_lambda(kq, yt)
                            gg_, lg_ = geometric_difference(kq, K_C)
                        except (np.linalg.LinAlgError, ValueError) as _e:
                            gg_, lg_, lam_a = float('nan'), float('nan'), float('nan')
                        d1[f'{H}|{p}|{tag}|g'].append(gg_)
                        d1[f'{H}|{p}|{tag}|lam_g'].append(lg_)
                        d1[f'{H}|{p}|{tag}|s_Q'].append(model_complexity(kq, yt, lam_a) if np.isfinite(lam_a) else float('nan'))
                        if KSAVE is not None and f'{H}|{p}|{tag}' not in KSAVE:
                            KSAVE[f'{H}|{p}|{tag}'] = kq.astype(np.float32)
                    del K0, off; free_mem()

                sig[f'{H}|{p}|A4_merge|dist'].append(float(dist))
                sig[f'{H}|{p}|A3_topp|rho'].append(redundancy(Es_H, s3))
                sig[f'{H}|{p}|overlap_A7_A8'].append(
                    len(set(map(int, s7)) & set(map(int, s8)))/float(p))

                for S in S_GRID:
                    Gt, Ge, dt, de, tg = [], [], [], [], []
                    for (tag, Ft, Fe_, v) in packs:
                        if np.isinf(S):
                            gt, dtt = Ft, xp.maximum(xp.sum(Ft**2, 1), 1e-30)
                            ge, dee = Fe_, xp.maximum(xp.sum(Fe_**2, 1), 1e-30)
                        else:
                            gt, dtt = noisy_features(Ft, v, S, nrng)
                            ge, dee = noisy_features(Fe_, v, S, nrng)
                        Gt.append(gt); Ge.append(ge); dt.append(dtt); de.append(dee); tg.append(tag)
                    F3, E3 = xp.stack(Gt), xp.stack(Ge)
                    D3, D4 = xp.stack(dt), xp.stack(de)
                    F3t = xp.swapaxes(F3, 1, 2)
                    if NOISY_KERNEL == 'LIN':        # Gan+ Eq. 44: F F^T / p, gurultulu normla bolme YOK
                        Kt = xp.matmul(F3, F3t)/F3.shape[2]
                        Ke = xp.matmul(E3, F3t)/F3.shape[2]
                    else:
                        if NOISY_KERNEL == 'RAW' and not np.isinf(S):   # duzeltmesiz (hep pozitif) norm
                            D3 = xp.maximum(xp.sum(F3**2, 2), 1e-30)
                            D4 = xp.maximum(xp.sum(E3**2, 2), 1e-30)
                        Kt = xp.matmul(F3, F3t)/xp.sqrt(D3[:, :, None]*D3[:, None, :])
                        Ke = xp.matmul(E3, F3t)/xp.sqrt(D4[:, :, None]*D3[:, None, :])
                    R = krr_multi(Kt, Ke, Yt, Ye)
                    for b, tag in enumerate(tg):
                        a, c = R[(b, 'y')]
                        accs[f'{H}|{p}|{tag}|{S}'].append(a)
                        if np.isinf(S):
                            corr[f'{H}|{p}|{tag}'].append(c)
                    del F3, E3, D3, D4, Kt, Ke, R; free_mem()
                del packs; free_mem()
            del Es_H, Et_H, Ee_H; free_mem()
        if KSAVE is not None:
            np.savez_compressed(os.path.join(KDIR, unit_key(ds, n, depth, ent, bw) + '.npz'),
                                **KSAVE)                       # D1.3 -- 0. bolme
        del Et, Ee; free_mem()
        PER[si] = dict(accs=accs, corr=corr, sig=sig, mix=mix, ec=ec, d1=d1, cls=cls,
                       meta=meta, X=XSAVE, s_C=s_C)
        with open(sp + '.tmp', 'wb') as fh:
            pickle.dump(PER[si], fh)
        os.replace(sp + '.tmp', sp)                            # yarim dosya kalmaz
        SPLIT_EST[n] = 1.15*(time.time() - ts)
        if log and (si == 0 or (si+1) % 5 == 0):
            print(f'      bolme {si+1}/{len(splits)}  ({time.time()-log:.0f}s)', flush=True)

    if any(r is None for r in PER):
        return None
    # Bolmeleri SIRAYLA birlestir -> tek parca koşudaki append sirasi birebir
    def _cat(nm):
        ks = dict.fromkeys(k for r in PER for k in r[nm])
        return {k: [x for r in PER for x in r[nm].get(k, [])] for k in ks}
    accs, corr, sig, mix = _cat('accs'), _cat('corr'), _cat('sig'), _cat('mix')
    ec, d1, cls, meta = _cat('ec'), _cat('d1'), _cat('cls'), _cat('meta')
    XSAVE = {k: v for r in PER for k, v in r['X'].items()}
    s_C = PER[-1]['s_C']

    # TUM bolmelerin on-islenmis girdisi -- kernel'i her sklearn surumunde
    # birebir yeniden uretebilmek icin (PCA isaret konvansiyonu surume bagli).
    np.savez_compressed(os.path.join(KDIR, unit_key(ds, n, depth, ent, bw) + '_X.npz'),
                        **XSAVE)

    out_a = {k: float(np.mean(v)) for k, v in accs.items()}
    out_s = {k: float(np.mean(v)) for k, v in sig.items()}
    out_m = {k: float(np.mean(v)) for k, v in mix.items()}
    out_l = {k: float(np.mean(v)) for k, v in cls.items()}
    out_d = {k: float(np.nanmean(v)) if np.any(np.isfinite(v)) else float('nan')
             for k, v in d1.items()}
    out_d.update({k + '|ndef': int(np.sum(np.isfinite(v))) for k, v in d1.items()
                  if k.endswith('|g')})
    out_e = {'status': Counter(ec['status']).most_common(1)[0][0],
             'status_fixed': Counter(ec['status_fixed']).most_common(1)[0][0],
             'effrank_frac': float(np.mean(ec['effrank_frac'])),
             'gate': float(np.mean(ec['gate'])), 'm2': float(np.mean(ec['m2'])),
             'offdiag_mean': float(np.mean(ec['offdiag_mean'])),
             's_req_A0': float(np.mean(ec['s_req_A0'])),
             'chance': float(np.mean(meta['chance'])),
             'm_tr': int(np.mean(meta['m_tr'])), 'm_te': int(np.mean(meta['m_te'])),
             's_C': float(s_C)}
    out_c = {}
    for k, v in corr.items():
        nd = len(v)//len(splits)
        out_c[k] = np.array(v).reshape(len(splits), nd, -1).mean(1)
    return out_a, out_s, out_m, out_e, out_c, out_l, out_d


RES_A, RES_S, RES_M, RES_E, RES_C, RES_L, RES_D = {}, {}, {}, {}, {}, {}, {}
DATA = {ds: load_real(ds) for ds in DS}
SPL = {ds: stratified_splits(DATA[ds][1], N_SPLIT, SEED0, TRAIN_FRAC) for ds in DS}
t00 = time.time()
INCOMPLETE = []
for ui, (ds, n, d, e, bw) in enumerate(ARM_UNITS):
    key = unit_key(ds, n, d, e, bw); path = os.path.join(CKPT, key + '.npz')
    if os.path.exists(path):
        z = np.load(path, allow_pickle=True)
        RES_A[key], RES_S[key], RES_M[key] = z['a'].item(), z['s'].item(), z['m'].item()
        RES_E[key], RES_C[key] = z['e'].item(), z['c'].item()
        RES_L[key], RES_D[key] = z['l'].item(), z['d'].item()
        print(f'[{ui+1}/{len(ARM_UNITS)}] {key}: checkpoint')
        continue
    t0 = time.time()
    X_, y_ = DATA[ds]
    try:
        out = run_unit(ds, X_, y_, n, d, e, bw, SPL[ds], log=t0)
    except Exception as _e:
        import traceback as _tb
        print(f'  !! {key} HATA: {_e}')
        print(_tb.format_exc()[-3000:])
        INCOMPLETE.append(key)
        continue
    if out is None:
        INCOMPLETE.append(key)
        continue
    a, s, m, ee, c, l, dd = out
    np.savez_compressed(path, a=a, s=s, m=m, e=ee, c=c, l=l, d=dd)
    RES_A[key], RES_S[key], RES_M[key] = a, s, m
    RES_E[key], RES_C[key], RES_L[key], RES_D[key] = ee, c, l, dd
    el = time.time()-t00
    print(f'[{ui+1}/{len(ARM_UNITS)}] {key}  {time.time()-t0:6.1f}s  ({ee["status"]}, '
          f'effrank/M={ee["effrank_frac"]:.3f} kapi={ee["gate"]:.3f})  '
          f'(toplam {el/60:.1f}dk, kalan ~{el/(ui+1)*(len(ARM_UNITS)-ui-1)/60:.1f}dk)',
          flush=True)
if INCOMPLETE:
    print(f'\n⚠️  EKSIK KOSU — {len(RES_A)}/{len(ARM_UNITS)} birim bitti, {len(INCOMPLETE)} eksik  {(time.time()-t00)/60:.1f} dk')
    for k in INCOMPLETE:
        nd = len([f for f in os.listdir(CKPT) if f.startswith(k + '__s') and f.endswith('.pkl')])
        print(f'  {k}: {nd}/{N_SPLIT} bolme')
    print('Analiz PARTIAL olarak calisacak — FULL sayilari gecersiz, sadece teshis icin. Kalan bolmeler icin sonraki parcayi bu ciktiyla (kernel_sources) devam ettir.')
    _IS_PARTIAL = True
    _PARTIAL_REASON = 'budget_or_error'
else:
    print(f'\nKOSU BITTI  {(time.time()-t00)/60:.1f} dk')
    _IS_PARTIAL = False
    _PARTIAL_REASON = ''
# K10 kontrolu icin bayrak: analiz hucreleri PARTIAL'de de calisir ama son JSON PARTIAL adiyla yazilir

# ---- K10: EC kapisi gercek veride YENIDEN olculdu (devralinmadi) ----
print('\nK10 -- EC durumu (birim basina yeniden olculdu):')
print(Counter(v['status'] for v in RES_E.values()).most_common())
dif = [k for k, v in RES_E.items() if v['status'] != v['status_fixed']]
print(f'K13 duyarlilik: `m2` kapisi ile sabit 0.90 kapisi {len(dif)} birimde AYRISIYOR')
for k in dif:
    print(f'  ! {k}: m2-kapisi={RES_E[k]["status"]}  0.90-kapisi={RES_E[k]["status_fixed"]} '
          f'(effrank/M={RES_E[k]["effrank_frac"]:.4f}, kapi={RES_E[k]["gate"]:.4f})')
```

## Bolum 5 - Satirlar + `S*`

Her satir = `(birim, H, p)`. Sentetikte etiket ekseni vardi; burada her veri
setinin **tek** gercek etiketi var, o yuzden satir sayisi 4 kat azdir ve
**kapsam sarti etiket degil VERI SETI uzerinden** okunur (K5).

Katman D **gurultusuz** dogruluklardan, `CHANCE` o veri setinin **egitim
cogunlugundan** (K1) hesaplanir.

```python
# PARTIAL kapisi: eksik birim varsa FULL degil PARTIAL
try:
    _IS_PARTIAL
except NameError:
    _IS_PARTIAL = len(RES_A) != len(ARM_UNITS)
if _IS_PARTIAL:
    print(f"⚠️  PARTIAL — {len(RES_A)}/{len(ARM_UNITS)} birim tamam, FULL tablosu gecersiz (sadece teshis).")
def _acc(key, H, p, arm, S):
    return RES_A[key].get(f'{H}|{p}|{arm}|{S}', np.nan)


rows = []
for (ds, n, d, e, bw) in ARM_UNITS:
    key = unit_key(ds, n, d, e, bw)
    if key not in RES_A:
        continue
    CH = RES_E[key]['chance']
    for H in H_LIST:
        for p in P_LIST:
            clean = {a: _acc(key, H, p, a, np.inf) for a in ARMS}
            # `A2_hb1` p > |1-body| iken tanimsiz (NaN) -- satiri dusurmez
            if any(np.isnan(v) for a, v in clean.items() if a != 'A2_hb1'):
                continue
            # v2 birincil rakip: Gan+ H-body ailesinin bu satirdaki GUCLUSU (bizim aleyhimize secim)
            HREF = ('A2_hb1' if np.isfinite(clean['A2_hb1']) and clean['A2_hb1'] > clean['A2_hb2']
                    else 'A2_hb2')
            clean['A2_hbody'] = clean[HREF]
            cfin = {a: v for a, v in clean.items() if np.isfinite(v)}
            st,  live,  gold  = layer_d(cfin, CH)
            sts, lives, golds = layer_d_strict(cfin, CH)
            tau  = clean['A2_hbody'] - TAU_DELTA
            tau2 = CH + 5.0
            r = dict(ds=ds, n=n, depth=d, ent=e, bw=bw, key=key, H=H, p=p,
                     chance=CH, status=st, live=live, gold=gold,
                     status_strict=sts, live_strict=lives, gold_strict=golds,
                     ec=RES_E[key]['status'], ec_fixed=RES_E[key]['status_fixed'],
                     tau=tau, tau2=tau2,
                     # K6: A2 sansa dusmusse birincil G1'e GIRMEZ
                     k6_ok=bool(clean['A2_hbody'] >= CH + K6_MIN_A2),
                     rbf=RES_L[key].get('rbf_loo', np.nan),
                     rbf_best=RES_L[key].get('rbf_best_test', np.nan),
                     s_C=RES_E[key].get('s_C', np.nan),
                     rho=RES_S[key].get(f'{H}|{p}|A3_topp|rho', np.nan),
                     merge_dist=RES_S[key].get(f'{H}|{p}|A4_merge|dist', np.nan),
                     overlap_A7_A8=RES_S[key].get(f'{H}|{p}|overlap_A7_A8', np.nan))
            r['hbody_ref'] = HREF
            for a in ARMS + ['A2_hbody']:
                src = HREF if a == 'A2_hbody' else a           # turetilmis kol -> kaynak kol
                curve = [_acc(key, H, p, src, S) for S in S_GRID]
                r[f'acc_{a}'] = clean[a]
                r[f'S_{a}']   = s_star(S_GRID, curve, tau)
                r[f'S2_{a}']  = s_star(S_GRID, curve, tau2)
                r[f'curve_{a}'] = curve
                # G1-ek (muhafazakar): hedef = ESKI kolun (A2_low) kendi dogrulugu - 1
                r[f'SL_{a}']  = s_star(S_GRID, curve, clean['A2_low'] - TAU_DELTA)
                r[f'g_{a}']   = RES_D[key].get(f'{H}|{p}|{src}|g', np.nan)
                r[f'sQ_{a}']  = RES_D[key].get(f'{H}|{p}|{src}|s_Q', np.nan)
            r['g_A0']  = RES_D[key].get('A0|g', np.nan)
            r['sQ_A0'] = RES_D[key].get('A0|s_Q', np.nan)
            for a in ARMS_SHOT + ['A2_hbody']:
                src = HREF if a == 'A2_hbody' else a
                r[f'sig_{a}']  = RES_S[key].get(f'{H}|{p}|{src}|sigma2', np.nan)
                r[f'sreq_{a}'] = RES_S[key].get(f'{H}|{p}|{src}|s_req', np.nan)
                r[f'w_{a}']    = RES_M[key].get(f'{H}|{p}|{src}|w', np.nan)
                r[f'c_{a}']    = RES_M[key].get(f'{H}|{p}|{src}|cost', np.nan)
            r['sig_A4_l2'] = RES_S[key].get(f'{H}|{p}|A4_merge|sigma2_l2', np.nan)
            rows.append(r)

LIVE  = [r for r in rows if r['live'] and r['ec'] == 'EC-LEARN']
OFFEC = [r for r in rows if r['live'] and r['ec'] != 'EC-LEARN']
GOLD  = [r for r in LIVE if r['gold_strict']]
G1SET = [r for r in LIVE if r['k6_ok']]                 # K6 uygulanmis birincil kume
K6OUT = [r for r in LIVE if not r['k6_ok']]
print(f'satir: {len(rows)}   CANLI (EC-LEARN): {len(LIVE)}   KATI ALTIN: {len(GOLD)}')
print(f'EC disi ama canli (ayri raporlanir): {len(OFFEC)}')
print(f'K6: birincil G1 kumesi {len(G1SET)} satir, dejenere hedef yuzunden '
      f'dislanan {len(K6OUT)} satir')
print('durum dagilimi:', Counter(r['status'] for r in rows).most_common())
print('veri seti (canli):', Counter(r['ds'] for r in LIVE).most_common())
print('birincil rakip (Gan+ H-body ailesinin guclusu), canli satirlarda:',
      Counter(r['hbody_ref'] for r in LIVE).most_common())

print('\nS* = inf orani (birincil G1 kumesinde):')
for a in ARMS + ['A2_hbody']:
    v = [r[f'S_{a}'] for r in G1SET]
    ninf = sum(np.isinf(x) for x in v)
    med = np.median([x for x in v if not np.isinf(x)]) if ninf < len(v) else np.nan
    print(f'  {a:<11} inf: {ninf:>3}/{len(v):<3}  sonlu medyan S* = '
          + ('-' if np.isnan(med) else f'{med:.0e}'))
```

## Bolum 6 - **G1 (BIRINCIL)**: sabit dogrulukta shot, gercek veride

`oran = S*(A2_hbody) / S*(kol)`. Dort sart (K5 ile kapsam yeniden tanimli):

1. medyan oran `>= 2.0`
2. %95 GA alt siniri `> 1.0`
3. canli satirlar **en az 2 VERI SETINDEN** gelmeli
4. en az 10 canli satir

**K6 aktif:** `A2 < sans+5` olan satirlar birincil degerlendirmeden dislanir
(Kapi 4'te bunu post-hoc yakalamistik, burada on-kayitli).

```python
def boot_median(v, n_boot=N_BOOT_G, seed=0):
    v = np.asarray(v, dtype=float)
    if len(v) == 0:
        return np.nan, np.nan, np.nan
    rng = np.random.default_rng(seed)
    st = np.median(v[rng.integers(0, len(v), (n_boot, len(v)))], axis=1)
    return float(np.median(v)), float(np.percentile(st, 2.5)), float(np.percentile(st, 97.5))


def ratios(rs, arm, ref='A2_hbody', field='S'):
    out, cap = [], 0
    for r in rs:
        a, b = r[f'{field}_{ref}'], r[f'{field}_{arm}']
        if np.isinf(b) and np.isinf(a):
            continue
        if np.isinf(b):
            out.append(1.0/RATIO_CAP); cap += 1
        elif np.isinf(a):
            out.append(RATIO_CAP); cap += 1
        elif b > 0:
            out.append(a/b)
    return np.array(out), cap


def g1_table(rs, title, ref='A2_hbody', field='S'):
    print('=' * 92); print(title); print('=' * 92)
    print(f"{'kol':<12}{'n':>5}{'medyan':>9}{'GA alt':>9}{'GA ust':>9}{'kirpik':>8}"
          f"{'veri seti':>12}  karar")
    res = {}
    for arm in sorted(CRIT_ARMS):
        v, cap = ratios(rs, arm, ref, field)
        nds = len({r['ds'] for r in rs})
        med, lo, hi = boot_median(v)
        core = len(v) >= G1_MIN_ROWS and med >= G1_MIN_RATIO and lo > G1_MIN_LO
        ok = core and nds >= G1_MIN_DS
        res[arm] = dict(n=len(v), median=med, lo=lo, hi=hi, capped=cap,
                        n_ds=nds, passed=bool(ok), passed_wo_k5=bool(core))
        # Tek veri setli kopyada (FM, ionosphere) K5 tanim geregi saglanamaz:
        # "gecmedi" yazmak oran/GA sartlarini da dusmus gibi gosterirdi.
        kar = 'GECTI' if ok else ('K5 bekliyor' if core and len(DS) < G1_MIN_DS else 'gecmedi')
        print(f'{arm:<12}{len(v):>5}{med:>9.2f}{lo:>9.2f}{hi:>9.2f}{cap:>8}'
              f'{nds:>12}  {kar}')
    return res


G1 = g1_table(G1SET, 'G1 -- BIRINCIL (canli & EC-LEARN & K6)')
print(f'\nsart: medyan>={G1_MIN_RATIO}, GA alt>{G1_MIN_LO}, '
      f'veri seti>={G1_MIN_DS} (K5), satir>={G1_MIN_ROWS}')
g1_passed = any(v['passed'] for v in G1.values())
g1_core = any(v['passed_wo_k5'] for v in G1.values())
if len(DS) < G1_MIN_DS:
    print(f'\nG1 SONUCU: bu defter TEK veri seti -> K5 (>= {G1_MIN_DS} veri seti) burada '
          f'uygulanamaz. Oran + GA + satir sartlari: {"SAGLANDI" if g1_core else "SAGLANMADI"}. '
          f'Birlesik G1 veri setleri bir arada ayrica hesaplanir.')
else:
    print(f'\nG1 SONUCU: {"GECTI" if g1_passed else "GECMEDI"}')

# v2 G1-ek: eski "dusuk agirlik oncelikli" kola (A2_low) gore -- MUHAFAZAKAR kiyas.
# Hedef A2_low'un kendi dogrulugu - 1 (`SL_` alanlari), K6 da A2_low'a gore.
G1SET_LOW = [r for r in LIVE if r['acc_A2_low'] >= r['chance'] + K6_MIN_A2]
G1_LOW = g1_table(G1SET_LOW, 'G1-ek -- dusuk agirlik oncelikli rakibe (A2_low) gore, muhafazakar',
                  ref='A2_low', field='SL')
print('\nDogruluk farki (gurultusuz, canli satirlar, medyan):')
for arm in ('A3_topp', 'A7_resid'):
    for ref in ('A2_hbody', 'A2_hb1', 'A2_hb2', 'A2_low', 'A1_random'):
        v = np.array([r[f'acc_{arm}'] - r[f'acc_{ref}'] for r in LIVE], dtype=float)
        v = v[np.isfinite(v)]
        if len(v):
            print(f'  {arm} - {ref:<9} {np.median(v):+6.2f}  (%{100*np.mean(v > 0):.0f} pozitif, n={len(v)})')

if K6OUT:
    print('\n--- K6 ile DISLANAN satirlar (ayri rapor, karar DEGISTIRMEZ) ---')
    _ = g1_table(K6OUT, 'K6-dislanan (A2 < sans+5, hedef dejenere)')
if OFFEC:
    print('\n--- EC kapisini GECMEYEN birimler (ayri rapor) ---')
    _ = g1_table(OFFEC, 'EC disi ama canli')
```

## Bolum 7 - G2 (merge bedeli) · G3 (A8) · G4 (vekil-ampirik) · G5 (butce)

```python
# ---- G2: merge'in shot bedeli ----
ov1 = [r['sig_A4_merge']/r['sig_A3_topp'] for r in G1SET
       if np.isfinite(r['sig_A4_merge']) and r['sig_A3_topp'] > 0]
ov2 = [r['sig_A4_l2']/r['sig_A3_topp'] for r in G1SET
       if np.isfinite(r['sig_A4_l2']) and r['sig_A3_topp'] > 0]
r43, _ = ratios(G1SET, 'A4_merge', 'A3_topp')
m43 = boot_median(r43)
print(f'G2  merge varyans sismesi: l1 medyan {np.median(ov1) if ov1 else np.nan:.2f}x  '
      f'l2 medyan {np.median(ov2) if ov2 else np.nan:.2f}x')
print(f'    S*(A3)/S*(A4) medyan {m43[0]:.2f} [{m43[1]:.2f}, {m43[2]:.2f}]  '
      + ('-> KAYNAKTA-GERI-VERILIYOR' if m43[0] < 1 else '-> merge shot ekseninde de kazandiriyor'))
G2 = dict(overhead_l1=float(np.median(ov1)) if ov1 else None,
          overhead_l2=float(np.median(ov2)) if ov2 else None,
          s_ratio_A3_over_A4=m43[0], lo=m43[1], hi=m43[2])

# ---- G3: A8 (on-kayitli beklenti: dusuk agirliga coker) ----
v8, _ = ratios(G1SET, 'A8_cost')
m8 = boot_median(v8)
loss8 = [r['acc_A7_resid'] - r['acc_A8_cost'] for r in G1SET]
w8m = np.mean([r['w_A8_cost'] for r in G1SET if np.isfinite(r['w_A8_cost'])]) if G1SET else np.nan
w2m = np.mean([r['w_A2_hbody'] for r in G1SET if np.isfinite(r['w_A2_hbody'])]) if G1SET else np.nan
w7m = np.mean([r['w_A7_resid'] for r in G1SET if np.isfinite(r['w_A7_resid'])]) if G1SET else np.nan
g3_ok = bool(len(v8) >= G1_MIN_ROWS and m8[0] >= G3_MIN_RATIO
             and np.median(loss8) <= G3_MAX_LOSS)
print(f'\nG3  A8: medyan oran {m8[0]:.2f} [{m8[1]:.2f}, {m8[2]:.2f}], '
      f'A7\'ye dogruluk kaybi medyan {np.median(loss8) if loss8 else np.nan:+.2f} puan '
      f'-> {"GECTI" if g3_ok else "GECMEDI"}')
print(f'    ortalama agirlik  A2={w2m:.2f}  A7={w7m:.2f}  A8={w8m:.2f}  '
      f'(A8 -> A2 ise beklenti dogrulandi: cokme)')
G3 = dict(median=m8[0], lo=m8[1], hi=m8[2], acc_loss=float(np.median(loss8)) if loss8 else None,
          w_A2=float(w2m), w_A7=float(w7m), w_A8=float(w8m), passed=g3_ok)

# ---- G4: analitik vekil vs ampirik S* ----
pairs = [(r[f'sreq_{a}'], r[f'S_{a}']) for r in G1SET for a in ARMS_SHOT
         if np.isfinite(r[f'sreq_{a}']) and np.isfinite(r[f'S_{a}']) and r[f'S_{a}'] > 0]
if len(pairs) > 3:
    A_ = np.log10([p[0] for p in pairs]); B_ = np.log10([p[1] for p in pairs])
    # std sifir ise korelasyon tanimsiz (sabit dizi)
    if float(np.std(A_)) == 0 or float(np.std(B_)) == 0:
        rho4 = lo4 = hi4 = float('nan'); g4_ok = False
        print(f'\nG4  korelasyon TANIMSIZ (sabit dizi, std=0)  (n={len(pairs)})')
    else:
        with np.errstate(invalid='ignore', divide='ignore'):
            rho4 = float(np.corrcoef(A_, B_)[0, 1])
            rng4 = np.random.default_rng(0)
            bs_raw = [float(np.corrcoef(A_[i], B_[i])[0, 1])
                      for i in (rng4.integers(0, len(A_), len(A_)) for _ in range(N_BOOT_G))]
            bs = [x for x in bs_raw if np.isfinite(x)]
            if not np.isfinite(rho4) or len(bs) < 100:
                lo4 = hi4 = float('nan'); g4_ok = False
                print(f'\nG4  korelasyon(log S_req, log S*) = TANIMSIZ [nan, nan]  (n={len(pairs)})')
            else:
                lo4, hi4 = float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))
                if not np.isfinite(lo4) or not np.isfinite(hi4):
                    g4_ok = False
                    print(f'\nG4  korelasyon(log S_req, log S*) = {rho4:+.3f} [nan, nan] -> TANIMSIZ  (n={len(pairs)})')
                else:
                    g4_ok = bool(lo4 > 0)
                    print(f'\nG4  korelasyon(log S_req, log S*) = {rho4:+.3f} [{lo4:+.3f}, {hi4:+.3f}] '
                          f'-> {"GECTI" if g4_ok else "GECMEDI"}  (n={len(pairs)})')
else:
    rho4 = lo4 = hi4 = np.nan; g4_ok = False
    print('\nG4  yeterli cift yok -> SONUCSUZ')
G4 = dict(rho=rho4, lo=lo4, hi=hi4, n=len(pairs), passed=g4_ok)

# ---- G5: kriter maliyeti dahil toplam butce (m_sel = |X_tr|, K4) ----
print('\nG5  TOPLAM DEVRE BUTCESI (kriter maliyeti DAHIL, m_sel = |X_tr|)')
print(f"{'kol':<12}{'medyan oran':>13}{'kriterli':>11}  not")
G5 = {}
for arm in sorted(CRIT_ARMS):
    num, den = [], []
    for r in G1SET:
        mtr, mte = RES_E[r['key']]['m_tr'], RES_E[r['key']]['m_te']
        sa, sb = r['S_A2_hbody'], r[f'S_{arm}']
        if not (np.isfinite(sa) and np.isfinite(sb)):
            continue
        num.append(budget('shadow', sa, mtr, mte))
        den.append(budget('shadow', sb, mtr, mte, m_sel=mtr))
    if num:
        rr = np.array(num)/np.array(den)
        mm = boot_median(rr)
        G5[arm] = dict(median=mm[0], lo=mm[1], hi=mm[2], n=len(rr))
        print(f'{arm:<12}{mm[0]:>13.2f}{"evet":>11}  GA [{mm[1]:.2f}, {mm[2]:.2f}], n={len(rr)}')
```

## Bolum 8 - KLASIK REFERANS + D1 (`g`, `s_Q`, `s_C`)

### Klasik referans -- MAKALEYE ZORUNLU
Sentetikte 72 hucrenin 72'sinde RBF her kuantum kolu yenmisti. Ayni karsilastirma
burada gercek veride yapilir. `rbf_loo` = LOO ile secilmis bant genisligi
(durust yol); `rbf_best_test` = test uzerinde en iyisi (klasigin LEHINE yanli,
yani bizim ALEYHIMIZE -- muhafazakar ust sinir olarak raporlanir).

### D1 -- ikincil, BAGLAYICI DEGIL
Yonlu ongoru (D1.2): kuantum kolun RBF'i `>= 1.0` puan yendigi her birim,
`g^2/N` dagiliminin **ust yarisinda** olmali **ve** o birimde `s_C > s_Q` olmali.
Hic avantaj birimi yoksa ongoru **test edilemez** -- "dogrulandi" YAZILMAZ.

```python
print('=' * 92)
print('KLASIK REFERANS -- RBF vs en iyi ETIKETSIZ kuantum kol (ayni bolme, ayni X)')
print('=' * 92)
print(f"{'birim':<34}{'EC':<10}{'sans':>6}{'A0':>7}{'kuantum':>9}{'RBF':>7}"
      f"{'RBF*':>7}{'fark':>8}")
UNS = [a for a in ARMS if a not in ('A0', 'A5_super')]
cmp_rows, n_q_win = [], 0
for r in sorted(rows, key=lambda r: (r['ds'], r['n'], r['bw'])):
    q = float(np.nanmax([r[f'acc_{a}'] for a in UNS]))      # A2_hb1 NaN olabilir
    dd = q - r['rbf']
    n_q_win += int(dd >= D1_ADV_MIN)
    cmp_rows.append(dict(key=r['key'], ds=r['ds'], H=r['H'], p=r['p'], ec=r['ec'],
                         chance=r['chance'], a0=r['acc_A0'], quantum=q,
                         rbf=r['rbf'], rbf_best=r['rbf_best'], diff=dd))
    if r['H'] == max(H_LIST) and r['p'] == max(P_LIST):
        print(f"{r['key']:<34}{r['ec']:<10}{r['chance']:>6.1f}{r['acc_A0']:>7.2f}"
              f"{q:>9.2f}{r['rbf']:>7.2f}{r['rbf_best']:>7.2f}{dd:>8.2f}")
n_tot = len(cmp_rows)
print(f'\nkuantum >= RBF + {D1_ADV_MIN} puan: {n_q_win}/{n_tot} satir')
print(f'fark medyani {np.median([c["diff"] for c in cmp_rows]):+.2f} puan  '
      f'(min {min(c["diff"] for c in cmp_rows):+.2f}, '
      f'max {max(c["diff"] for c in cmp_rows):+.2f})')

# ---- D1.2 yonlu ongoru ----
print('\n' + '=' * 92)
print('D1 -- Huang metrikleri (IKINCIL, baglayici degil)')
print('=' * 92)
D1OUT = {}
for ds in DS:
    sub = [r for r in rows if r['ds'] == ds]
    if not sub:
        continue
    N_ = RES_E[sub[0]['key']]['m_tr']
    gg = np.array([[r[f'g_{a}'] for a in UNS] for r in sub], dtype=float)
    ndef = int(np.sum(np.isfinite(gg)))
    # MEDYAN, ortalama DEGIL: bir bolmede `K_C` sayisal olarak tekil cikarsa
    # (`eigmin == 0`) kosul sayisi `inf` olur ve ortalamayi tek basina yutar --
    # pilotta ionosphere icin tam bu oldu (`kosul=inf`, eigmin>0 iken).
    _emin = np.array([RES_L[r['key']].get('KC_eigmin', np.nan) for r in sub], dtype=float)
    _econd = np.array([RES_L[r['key']].get('KC_cond', np.nan) for r in sub], dtype=float)
    eminv, econd = np.nanmedian(_emin), np.nanmedian(_econd)
    n_sing = int(np.sum(~np.isfinite(_econd)))
    print(f'\n{ds}: N={N_}, sqrt(N)={np.sqrt(N_):.1f}, `g` tanimli {ndef}/{gg.size} '
          f'(kol x satir)')
    # D2 sarti: `g` yalniz K_C'nin kosuluyla BIRLIKTE okunur
    print(f'  K_C spektral saglik (medyan): ozdeger_min={eminv:.2e}  kosul={econd:.1e}'
          + (f'  [{n_sing} birim sayisal olarak TEKIL]' if n_sing else ''))
    if ndef:
        print(f'  g: medyan {np.nanmedian(gg):.2f}  [{np.nanmin(gg):.2f}, {np.nanmax(gg):.2f}]'
              f'   g^2/N medyan {np.nanmedian(gg**2)/N_:.3f}')
    sQ = np.array([[r[f'sQ_{a}'] for a in UNS] for r in sub], dtype=float)
    sC = np.array([r['s_C'] for r in sub], dtype=float)
    print(f'  s_Q medyan {np.nanmedian(sQ):.1f} (N={N_})   s_C medyan {np.nanmedian(sC):.1f}')
    adv = [(r, a) for r in sub for a in UNS if r[f'acc_{a}'] - r['rbf'] >= D1_ADV_MIN]
    if not adv:
        print(f'  AVANTAJ BIRIMI YOK -> D1.2 ongorusu TEST EDILEMEZ (md.3). '
              f'"dogrulandi" yazilmaz.')
        # md.4: avantajin yoklugu `g`den mi konsantrasyondan mi?
        hi_g = np.nanmedian(gg) > np.sqrt(N_)/2 if ndef else False
        print(f'  aciklama: g {"buyuk" if hi_g else "kucuk"}, '
              f's_Q/N = {np.nanmedian(sQ)/N_:.2f} '
              f'-> {"konsantrasyon" if np.nanmedian(sQ)/N_ > 0.5 else "geometri"} kaynakli')
        D1OUT[ds] = dict(testable=False, n_def=ndef, g_med=float(np.nanmedian(gg)) if ndef else None,
                         sQ_over_N=float(np.nanmedian(sQ)/N_), sC_med=float(np.nanmedian(sC)))
    else:
        thr = np.nanmedian(gg)
        ok = sum(1 for (r, a) in adv
                 if np.isfinite(r[f'g_{a}']) and r[f'g_{a}'] >= thr and r['s_C'] > r[f'sQ_{a}'])
        print(f'  avantaj birimi {len(adv)}; ongoruyu saglayan {ok}/{len(adv)}')
        if ok < len(adv):
            print('  !! IHLAL (dusuk `g`de avantaj) -> ONCE UYGULAMA HATASI ARANIR (D1.2 md.2)')
        D1OUT[ds] = dict(testable=True, n_adv=len(adv), n_ok=ok,
                         g_med=float(thr), sC_med=float(np.nanmedian(sC)))
```

### SNR (v2, 2026-10-15)

Kol basina `S_req = sigma2_shot / Var[k~]` (kernel'in kosegen-disi sinyal varyansini shot
gurultusunun ustune cikarmak icin gereken shot). `S` shot'ta sinyal/gurultu orani
`SNR(S) = S / S_req`; `SNR = 1` cizgisi `S = S_req` noktasidir. Sol: birincil kumede kol basina
medyan SNR (golge = ceyrekler arasi). Sag: ayni satirlarda dogruluk vs SNR -- kollar `SNR ~ 1`
civarinda doymaya basliyorsa `S_req` dogru vekildir (G4'un gorsel hali).

```python
SNR_ARMS = ['A1_random', 'A2_hbody', 'A2_low', 'A3_topp', 'A7_resid', 'A8_cost']
SNR_LAB = {'A2_hbody': 'Gan+ H-body (guclusu)', 'A2_low': 'dusuk agirlik oncelikli'}
Sg = np.array([s for s in S_GRID if np.isfinite(s)], dtype=float)
SNR_OUT = {}
fig, ax = plt.subplots(1, 2, figsize=(13, 4.8))
for a in SNR_ARMS:
    sr = np.array([r[f'sreq_{a}'] for r in G1SET], dtype=float)
    ok = np.isfinite(sr) & (sr > 0)
    if not ok.any():
        continue
    snr = Sg[None, :] / sr[ok, None]                      # (satir, S)
    med = np.median(snr, 0)
    l, = ax[0].plot(Sg, med, 'o-', label=f'{SNR_LAB.get(a, a)} (S_req~{np.median(sr[ok]):.0f})')
    ax[0].fill_between(Sg, np.percentile(snr, 25, 0), np.percentile(snr, 75, 0),
                       color=l.get_color(), alpha=0.15)
    cur = np.array([r[f'curve_{a}'] for r in G1SET], dtype=float)[ok][:, :len(Sg)]
    ax[1].plot(med, np.nanmedian(cur, 0), 'o-', color=l.get_color(), label=SNR_LAB.get(a, a))
    SNR_OUT[a] = dict(S_req_median=float(np.median(sr[ok])), n=int(ok.sum()),
                      snr_median=list(map(float, med)), acc_median=list(map(float, np.nanmedian(cur, 0))))
ax[0].set_xscale('log'); ax[0].axhline(1, c='k', ls=':', lw=1)
ax[1].set_xscale('log'); ax[1].axvline(1, c='k', ls=':', lw=1)
ax[0].set_yscale('log'); ax[0].set_xlabel('shot S'); ax[0].set_ylabel('SNR = S / S_req')
ax[0].set_title('SNR vs shot (birincil kume, medyan + IQR)'); ax[0].legend(fontsize=7)
ax[1].set_xlabel('SNR (medyan)'); ax[1].set_ylabel('dogruluk %'); ax[1].set_title('dogruluk vs SNR')
if G1SET:
    ax[1].axhline(np.median([r['chance'] for r in G1SET]), ls=':', c='gray', lw=1)
ax[1].legend(fontsize=7)
plt.tight_layout(); plt.savefig(f'{TAG.upper()}_{MODE}_snr.png', dpi=130); plt.show()
json.dump(dict(S_grid=list(map(float, Sg)), arms=SNR_OUT),
          open(f'{TAG}_{MODE}_snr.json', 'w', encoding='utf-8'), indent=1)
```

## Bolum 10 - Kernel'leri KALICI sakla (Kaggle Dataset)

D1.3 defterin `K_Q` (tum kollar) ve `K_C` matrislerini `float32` yazmasini
sart kosuyor. Defter cikti klasoru oturuma bagli; **ozel bir Kaggle Dataset'e**
surumlenirse defter silinse bile veri kalir ve `python -m kaggle datasets
download` ile her an inebilir.

Calismasi icin Kaggle Secrets'ta `KAGGLE_USERNAME` ve `KAGGLE_KEY` tanimli
olmali. Tanimli degilse hucre **sessizce atlar** -- koşuyu asla dusurmez.

```python
import sys, subprocess

DS_SLUG = f'{TAG}-hope-kernels-{MODE.lower()}'

def push_kernels():
    try:
        from kaggle_secrets import UserSecretsClient
        sec = UserSecretsClient()
        os.environ['KAGGLE_USERNAME'] = sec.get_secret('KAGGLE_USERNAME')
        os.environ['KAGGLE_KEY'] = sec.get_secret('KAGGLE_KEY')
    except Exception as e:
        print('Kaggle Secrets yok -> kernel arsivi YALNIZ defter ciktisinda kalir:', e)
        return
    user = os.environ['KAGGLE_USERNAME']
    meta = {'title': f'HOPE T1.1 kernels ({MODE})', 'id': f'{user}/{DS_SLUG}',
            'licenses': [{'name': 'CC0-1.0'}]}
    json.dump(meta, open(os.path.join(KDIR, 'dataset-metadata.json'), 'w'))
    tot = sum(os.path.getsize(os.path.join(KDIR, f)) for f in os.listdir(KDIR))
    print(f'{KDIR}: {len(os.listdir(KDIR))} dosya, {tot/2**20:.0f} MB')
    for cmd in ([sys.executable, '-m', 'kaggle', 'datasets', 'version', '-p', KDIR,
                 '-m', f'T1.1 {MODE} {time.strftime("%Y-%m-%d %H:%M")}', '--dir-mode', 'zip'],
                [sys.executable, '-m', 'kaggle', 'datasets', 'create', '-p', KDIR,
                 '--dir-mode', 'zip']):
        r = subprocess.run(cmd, capture_output=True, text=True)
        print(r.stdout.strip() or r.stderr.strip())
        if r.returncode == 0:
            print(f'-> https://www.kaggle.com/datasets/{user}/{DS_SLUG}')
            return
    print('!! itilemedi; kernel arsivi defter ciktisinda duruyor')


push_kernels()
```

## PARA BIRIMI -- hata #17'nin cozumu

Faz A'dan beri "A0'in mutlak shot sayilari GFQK ile ayni para biriminde degil"
diye bir dipnot tasiyorduk. Kapi 4 bunu **kapatir**: birim ayni, carpan farkli
ve ikisi de acikca yazilir.

| protokol | kol | gurultu modeli | kernel matrisi icin devre sayisi |
|---|---|---|---|
| **shadow** | A1-A8 | `Var = 3^w / S` (agirlikta ustel) | `S x (M_tr + M_te [+ M_sel])` -- ciftler **bedava** |
| **fidelity** | A0 | `Var = k(1-k)/S` (binom, `3^w` yuku YOK) | `S x (M_tr(M_tr-1)/2 + M_te·M_tr)` -- her CIFT ayri devre |

Kapi 4'un en buyuk tek kalemi bu: `M = 1200` icin fidelity, shadow'un
**~333 kati** devre harcar. A0'in bedeli varyansta degil **cift sayisinda**.

**Ikinci eksen** (devre derinligi): 2-qubit kapi sayisi. ZZ-haritasi katman
basina `n(n-1)/2` ZZ dondurmesi; fidelity compute-uncompute oldugu icin **2 kat**.
Shadow'un rastgele olcum katmani tek-qubit -> 2q sayisina katkisi yok.

**Ucuncu eksen** (model karmasikligi `p`): kollar arasi kiyasta **sabit** tutulur,
`p` ekseni ayrica taranir. Protokol geregi uc eksen **birlikte degil, ayri ayri**
raporlanir.

```python
"""Kuantum kernel cekirdegi -- TEK KAYNAK.

Bu dosya defterlere OLDUGU GIBI GOMULUR (build_*_nb.py `open(...).read()` ile alir),
boylece Colab'de self-contained kalir ama kod tek yerde durur.

BEKLENEN ORTAM: `xp`, `GPU`, `CDT`, `FDT`, `to_xp`, `to_np` defterin 0. hucresinde
tanimlanmis olmali (cupy/numpy dagitimi).

Dogrulanmis (bkz. defterlerin Bolum 4):
  fwht == H^{otimes n} · pauli_expect_fast == naif yol (2.2e-16) · Parseval ·
  GFQK_{w<=n} == fidelity kernel · dort eksende S_req monoton · Katman B/C ayrimlari
"""


# ============================================================================
# DEVRE -- yogun 2^n x 2^n matris KURULMAZ
# ============================================================================

def fwht(a, normalize=True):
    """Son eksende H^{otimes n}. a: (..., 2^n). O(N n 2^n), yogun matris yok.

    YERINDE kelebek (butterfly). Onceki surum her adimda `moveaxis`+`stack`
    yapiyordu; profil bunun tek basina Faz B suresinin ~%15'i oldugunu gosterdi
    (7600 `stack` cagrisi). H tek-qubit kapilari komutatif oldugu icin bit
    sirasindan bagimsiz; dogrulama [1a] H^{otimes n}'e karsi kilitliyor.
    """
    dim = a.shape[-1]
    n = int(np.log2(dim))
    shp = a.shape
    b = xp.array(a, copy=True).reshape(-1, dim)
    h = 1
    while h < dim:
        b = b.reshape(-1, dim // (2*h), 2, h)
        u = xp.array(b[:, :, 0, :], copy=True)
        v = b[:, :, 1, :]
        b[:, :, 0, :] = u + v
        b[:, :, 1, :] = u - v
        b = b.reshape(-1, dim)
        h *= 2
    b = b.reshape(shp)
    return b * (2.0 ** (-n / 2)) if normalize else b


def sample_X(rng, N, n, rep=1, jitter=0.0):
    """Veri ureteci -- FAZ D icin REDUNDANS KOLU.

    `rep=1, jitter=0.0` -> eski yolun BIREBIR aynisi (ayni tek `rng` cagrisi,
    dolayisiyla Faz A/B/C/C2 sonuclari degismez).

    `rep>1`: n koordinat, `d_eff = max(3, n//rep)` GIZLI degiskenden turetilir:

        x_j = z_{j mod d_eff} + jitter * u_j,    z~U(0,2pi),  u~U(-pi,pi)

    Bu redundansi VERIDE uretir, DEVREDE degil. Amac kritik: devrenin ifade gucu
    (EC'nin kaynagi) dokunulmadan kalirken `psi` sutunlari korele hale gelsin.
    Devreyi degistirseydik EC'yi de degistirir ve testi anlamsizlastirirdik.

    `d_eff >= 3` tutulur cunku etiket bataryasi ilk 3 koordinati kullanir
    (`make_labels`, `n_fixed=3`) -- gorevin icsel zorlugu sabit kalmali.

    UYARI (raporlanmali): `rep>1` veri manifoldunun icsel boyutunu dusurur, yani
    ogrenme gorevini kolaylastirabilir. Bu, A7-vs-A3 karsilastirmasi icin ZARARSIZ
    (iki kol da ayni veriyi gorur) ama mutlak dogruluklar Faz C2 ile KIYASLANAMAZ.
    """
    if rep == 1 and jitter == 0.0:
        return rng.uniform(0, 2 * np.pi, size=(N, n))      # BIREBIR eski yol
    d = max(3, n // rep)
    z = rng.uniform(0, 2 * np.pi, size=(N, d))
    X = z[:, np.arange(n) % d]
    if jitter > 0.0:
        X = X + jitter * rng.uniform(-np.pi, np.pi, size=(N, n))
    return X


def sign_tables(n):
    z = np.arange(2 ** n)
    S = 1.0 - 2.0 * ((z[:, None] >> np.arange(n - 1, -1, -1)[None, :]) & 1)
    pairs = [(i, j) for i in range(n) for j in range(i + 1, n)]
    SP = np.stack([S[:, i] * S[:, j] for i, j in pairs], 1) if pairs else np.zeros((2 ** n, 0))
    return S, SP, pairs


_SIGN_DEV = {}


def sign_tables_dev(n):
    """`sign_tables`'in cihazdaki (FDT, transpoze) kopyasi, SON `n` icin onbellekli.

    2026-10-13 (kosu 2, n=16/20): eski `encode` tabloyu HER cagrida yeniden kurup
    cihaza tasiyordu; n=20'de bu cagri basina ~1.5 GB (SP: 2^20 x 190). Degerler
    birebir ayni -- yalniz bir kez kurulur. Tek `n` tutulur: n=20 tablosu ~0.8 GB.
    """
    if n not in _SIGN_DEV:
        _SIGN_DEV.clear()
        S, SP, pairs = sign_tables(n)
        _SIGN_DEV[n] = (to_xp(S, FDT).T, to_xp(SP, FDT).T, pairs)
    return _SIGN_DEV[n]


def encode(X, n, depth, ent, bw, norm='none'):
    """ZZ-feature map, `depth` tekrar. Doner: (M, 2^n) statevector.

    faz(z) = -( bw * sum_i x_i s_i(z)  +  ent * bw^2 * sum_{i<j} x_i x_j s_ij(z) )

    `norm` -- `n`-BAGIMLI KONFOUND kontrolu:
      'none' : ham. Terim sayilari n ve n(n-1)/2 oldugu icin faz sacilimi
               sirasiyla ~sqrt(n) ve ~n buyur -> BUYUK n'de devre DAHA COK karisir.
               Bu, "n buyudu" ile "encoding guclendi" etkilerini karistirir.
      'sqrt' : her iki terim kendi terim sayisinin karekokune bolunur -> faz
               sacilimi n'den BAGIMSIZ. "Ayni harita, farkli n" yorumu gecerli olur.

    ent=0 -> faz tamamen tek-qubit -> durum CARPIM kalir (Meyer-Wallach tam 0),
    `norm` bunu degistirmez (tek-qubit terim carpimsal yapiyi bozmaz).
    """
    St, SPt, pairs = sign_tables_dev(n)
    Xp = np.stack([X[:, i] * X[:, j] for i, j in pairs], 1) if pairs else np.zeros((len(X), 0))
    c1, c2 = 1.0, 1.0
    if norm == 'sqrt':
        c1 = 1.0 / np.sqrt(n)
        c2 = 1.0 / np.sqrt(max(len(pairs), 1))
    elif norm != 'none':
        raise ValueError(norm)
    phase = -(c1 * bw * (to_xp(X, FDT) @ St)
              + c2 * ent * (bw ** 2) * (to_xp(Xp, FDT) @ SPt))
    D = xp.exp(1j * phase).astype(CDT)
    psi = xp.zeros((len(X), 2 ** n), dtype=CDT)
    psi[:, 0] = 1.0
    for _ in range(depth):
        psi = fwht(psi) * D
    return psi


def meyer_wallach(psi, n):
    """Q = 2(1 - (1/n) sum_k tr(rho_k^2)).  0 = carpim, 1 = maksimal dolanik."""
    M = psi.shape[0]
    tot = xp.zeros(M, dtype=FDT)
    for k in range(n):
        a = xp.moveaxis(psi.reshape(M, *([2] * n)), k + 1, 1).reshape(M, 2, -1)
        rho = xp.einsum('mir,mjr->mij', a, a.conj())
        tot += xp.real(xp.einsum('mij,mji->m', rho, rho)).astype(FDT)
    return float(to_np(xp.mean(2.0 * (1.0 - tot / n))))


# ============================================================================
# PAULI DOMENI -- flip-maskesi + Walsh-Hadamard hizli yolu
# ============================================================================

def _popcount(v):
    v = np.asarray(v, dtype=np.int64).copy()
    r = np.zeros_like(v)
    while np.any(v):
        r += v & 1
        v >>= 1
    return r


def _spread4(v, n):
    v = np.asarray(v, dtype=np.int64)
    out = np.zeros_like(v)
    for k in range(n):
        out += ((v >> k) & 1) * (4 ** k)
    return out


def pool_layout_from_indices(idx, n):
    """Verilen KEYFI Pauli indeks kumesi icin layout (agirlik havuzu degil).
    `qglobal` etiketi gibi tek tuk operator gerektiren yerlerde kullanilir."""
    idx = np.unique(np.asarray(idx, dtype=np.int64))
    f_ = np.zeros(len(idx), np.int64); g_ = np.zeros(len(idx), np.int64)
    for k in range(n):
        d = (idx // (4 ** k)) % 4                  # 0=I 1=X 2=Y 3=Z
        f_ |= np.where((d == 1) | (d == 2), 1 << k, 0)
        g_ |= np.where((d == 2) | (d == 3), 1 << k, 0)
    blocks, fs = [], []
    for f in np.unique(f_):
        m = f_ == f
        blocks.append((g_[m], idx[m])); fs.append(f)
    w = np.zeros(len(idx), np.int64)
    for k in range(n):
        w += (((idx // (4 ** k)) % 4) != 0)
    o = np.argsort(idx)
    return np.array(fs), blocks, idx[o], w[o]


def pool_layout(n, h, max_flips=None, seed=0, keep_low=None):
    """Agirlik <= h havuzu.

    `max_flips`: flip-maske sayisi bunu asarsa RASTGELE alt orneklenir. Maliyet
    flip-maske sayisiyla DOGRU ORANTILI (her biri bir WHT), operator sayisiyla degil;
    o yuzden kisitlama burada yapilir. Alt ornekleme TUM kollara ayni havuzu verir,
    yani adildir -- ve A3'u (top-p) A1'den (rastgele) DAHA COK zorlar, cunku en iyi
    operator disarida kalabilir. Bizim iddiamiz acisindan MUHAFAZAKAR.

    `keep_low` (v2, 2026-10-15 denetimi): popcount <= keep_low olan flip'lerin HEPSI
    tutulur, `max_flips` yalniz geri kalanlardan (popcount > keep_low) cekilir. Neden:
    H=2 kollari H=4 icin kirpilmis havuzdan cekiliyordu (n=12'de 2-body 154/594, ZZ
    asiri temsil) ve n=14, depth=1, ent=0'da H=2 havuzunun rank'i 8'e dusup Faz B'yi
    dusurdu. `None` -> eski davranis (eski kosular birebir yeniden uretilir).

    Doner: (flip_maskeleri, [(g_dizisi, pauli_indeksleri) her f icin],
            tum_indeksler_sirali, tum_agirliklar_sirali).
    """
    allz = np.arange(2 ** n)
    pc = _popcount(allz)
    fs = np.where(pc <= h)[0]
    if keep_low is not None:
        lo, hi = fs[pc[fs] <= keep_low], fs[pc[fs] > keep_low]
        if max_flips is not None and len(hi) > max_flips:
            hi = hi[np.random.default_rng(seed + 4242).choice(len(hi), max_flips, replace=False)]
        fs = np.sort(np.concatenate([lo, hi]))
    elif max_flips is not None and len(fs) > max_flips:
        keep = np.random.default_rng(seed + 4242).choice(len(fs), max_flips, replace=False)
        fs = np.sort(fs[np.union1d(keep, 0)])      # f=0 (Z-only aile) HER ZAMAN kalir
    blocks, idx_all, w_all = [], [], []
    for f in fs:
        g = np.where(pc[allz | f] <= h)[0]            # agirlik = |f OR g|
        pidx = 2 * _spread4(f & g, n) + _spread4(f & ~g, n) + 3 * _spread4(g & ~f, n)
        blocks.append((g, pidx))
        idx_all.append(pidx)
        w_all.append(pc[g | f])
    idx_all = np.concatenate(idx_all)
    w_all = np.concatenate(w_all)
    o = np.argsort(idx_all)
    return fs, blocks, idx_all[o], w_all[o]


def pauli_expect_fast(psi, n, layout):
    """<P_{f,g}> = Re[ (-i)^{|f&g|} * WHT_z( conj(psi_z) psi_{z^f} )[g] ]

    Ayni f'yi paylasan tum operatorler tek gather + tek WHT ile hesaplanir.
    n=12, w<=4: 46 666 operator yerine 794 donusum.
    """
    fs, blocks, _, _ = layout
    z = xp.arange(2 ** n)
    conj = psi.conj()
    cols, keys = [], []
    for f, (g, pidx) in zip(fs, blocks):
        Wt = fwht(conj * psi[:, z ^ int(f)], normalize=False)
        ph = ((-1j) ** to_xp(_popcount(np.bitwise_and(int(f), g)))).astype(CDT)
        cols.append(xp.real(Wt[:, to_xp(g.astype(np.int64))] * ph[None, :]).astype(FDT))
        keys.append(pidx)
    E = xp.concatenate(cols, axis=1)
    return E[:, to_xp(np.argsort(np.concatenate(keys)))]


def pauli_expect_naive(psi, n, idx):
    """Referans yol -- SADECE dogrulama icin (yavas)."""
    flip = np.zeros(len(idx), np.int64)
    phm = np.zeros(len(idx), np.int64)
    nY = np.zeros(len(idx), np.int64)
    for k in range(n):
        d = (np.asarray(idx) // (4 ** k)) % 4
        flip |= np.where((d == 1) | (d == 2), 1 << k, 0)
        phm |= np.where((d == 2) | (d == 3), 1 << k, 0)
        nY += (d == 2)
    z = np.arange(2 ** n, dtype=np.int64)
    P = to_np(psi)
    out = np.zeros((P.shape[0], len(idx)))
    for t in range(len(idx)):
        zf = z ^ flip[t]
        sgn = 1.0 - 2.0 * (_popcount(zf & phm[t]) & 1)
        out[:, t] = np.real(np.sum(P.conj() * sgn[None, :] * P[:, zf], 1) * (1j) ** int(nY[t]))
    return out


# ============================================================================
# KERNEL -- birim kosegene normalize + tahmin edicinin shot varyansi
# ============================================================================

def kernel_gfqk(E_pool, w_pool, H, p):
    """GFQK_{w<=H}, gurultulu, birim kosegene normalize. Doner (K~, sigma2_shot).

    Gurultu: qubit basina depolarizing, Pauli-transfer ozdegeri lam=1-4p/3 -> EGZAKT.
    Shot varyansi: classical shadow, operator basina 3^w.
      k = sum_i a_i b_i, Var(a^_i) ~ 3^{w_i}
      -> Var(k^) ~ 2 sum_i 3^{w_i} <psi~_i^2>,   <psi~_i^2> = mean(E[:,i]^2)
    `c`'nin BIRINCI kuvveti kullanilir; c^2 yazmak pay/paydada sadelesip gurultu
    eksenini ters cevirir (bkz. 02_PROJECT_STATUS.md hata #16).
    """
    w = np.asarray(w_pool)
    m = to_xp(w <= H)
    ww = to_xp(w[w <= H].astype(np.float64), FDT)
    E = E_pool[:, m] * xp.sqrt((1.0 - 4.0 * p / 3.0) ** (2 * ww))[None, :]
    K = E @ E.T
    dg = xp.maximum(xp.diag(K), 1e-30)
    s2 = float(to_np(2.0 * xp.sum((3.0 ** ww) * xp.mean(E ** 2, 0)) / xp.mean(dg) ** 2))
    return K / xp.sqrt(xp.outer(dg, dg)), s2


def kernel_fidelity(psi):
    """A0 referansi. diag zaten 1; tahmin edici compute-uncompute BINOM -> sigma2 None.
    Bu, GFQK'dan FARKLI bir olcum protokoludur; mutlak shot sayilari ayni para birimi
    degil (bkz. hata #17)."""
    return xp.abs(psi.conj() @ psi.T) ** 2, None


# ============================================================================
# TESHIS -- Katman A-shot / A-learn / B / C
# ============================================================================

def effrank(ev):
    ev = xp.clip(ev.real, 0, None)
    s1, s2 = ev.sum(), (ev ** 2).sum()
    return float(to_np(s1 * s1 / s2)) if float(to_np(s2)) > 0 else 0.0


def diagnose(K, sigma2_shot=None, budgets=(1e3, 1e5, 1e7)):
    """S_req = sigma2_shot / Var[k~]  --  `1/Var` DEGIL (hata #15)."""
    M = K.shape[0]
    off = K[~xp.eye(M, dtype=bool)]
    var = float(to_np(off.var()))
    mu = float(to_np(off.mean()))
    kc = xp.clip(off, 0.0, 1.0)
    s2 = float(to_np(xp.mean(kc * (1 - kc)))) if sigma2_shot is None else float(sigma2_shot)
    er = effrank(xp.linalg.eigvalsh(K))
    d = dict(offdiag_var=var, offdiag_mean=mu, sigma2_shot=s2, M=M,
             s_req=(s2 / var) if var > 0 else float('inf'),
             effrank_K=er, effrank_frac=er / M)
    for S in budgets:
        d[f'resolv@{S:.0e}'] = float(to_np(xp.mean(xp.abs(off - mu) > 3 * np.sqrt(s2 / S))))
    return d


def layers_ac(d, budget, mean_gate, effrank_gate):
    """Doner (durum, a_shot, a_learn). Katman C ONCE: K->11^T ise bandwidth cokusu, EC DEGIL.

    UYARI: a_learn (effrank/M -> 1) EC'ye OZGU DEGILDIR -- carpim durumlarinda da
    tetiklenir ve n>=14'te 1.0000'da doyar. Bolge listesi Katman B (ustel onayi) ile
    KAPILANMALIDIR (bkz. 02_PROJECT_STATUS.md 6.96).
    """
    if d['offdiag_mean'] >= mean_gate:
        return 'COLLAPSE', False, False
    a_shot = d['s_req'] > budget
    a_learn = d['effrank_frac'] > effrank_gate
    if a_shot and a_learn:
        return 'EC-BOTH', True, True
    if a_shot:
        return 'EC-SHOT', True, False
    if a_learn:
        return 'EC-LEARN', False, True
    return 'NO-EC', False, False


def layer_b(ns, vals, n_boot=2000, seed=0):
    """log V ~ a + b*n (ustel)  vs  log V ~ a' + b'*log n (polinom), AIC ile.
    `exponential` = egim GA ust siniri < 0  VE  AIC ustelde daha kucuk.

    UYARI: `exponential=False`, "EC yok" DEMEK DEGILDIR -- egri henuz bukulme
    bolgesinde olabilir (asimptota girmemis). Bkz. 6.96 acik #1.
    """
    ns = np.asarray(ns, float)
    v = np.log(np.maximum(np.asarray(vals, float), 1e-300))
    k = len(ns)
    if k < 3:
        return None

    def fit(x, y):
        b, a = np.polyfit(x, y, 1)
        r = y - (a + b * x)
        ss = ((y - y.mean()) ** 2).sum()
        return b, (1 - (r ** 2).sum() / ss if ss > 0 else 0.0), (r ** 2).sum()

    b_e, r2_e, ss_e = fit(ns, v)
    b_p, r2_p, ss_p = fit(np.log(ns), v)
    aic = lambda ss: k * np.log(max(ss, 1e-300) / k) + 4
    rng = np.random.default_rng(seed)
    bs = []
    while len(bs) < n_boot:                       # dejenere cekilis (tek n) egimi bozar
        i = rng.integers(0, k, k)
        if len(np.unique(ns[i])) >= 2:
            bs.append(np.polyfit(ns[i], v[i], 1)[0])
    bs = np.array(bs)
    return dict(slope=float(b_e), slope_lo=float(np.percentile(bs, 2.5)),
                slope_hi=float(np.percentile(bs, 97.5)), r2_exp=float(r2_e),
                r2_poly=float(r2_p), d_aic=float(aic(ss_e) - aic(ss_p)),
                exponential=bool(np.percentile(bs, 97.5) < 0 and aic(ss_e) < aic(ss_p)))


"""Faz B cekirdegi -- ogrenme tarafi. `qk_core.py` ile birlikte deftere gomulur.

qk_core: devre, Pauli domeni, kernel, EC teshisi (Katman A/B/C)
phaseb : etiketler, kol secimi (A1-A5), merge, KRR, istatistik, Katman D

BEKLENEN ORTAM: qk_core'un tanimladigi her sey + `xp/GPU/CDT/FDT/to_xp/to_np`.
"""


# ============================================================================
# 1) ETIKET BATARYASI -- SABIT koordinat alt kumesinde
# ============================================================================
# KRITIK: her etiket sadece ilk `n_fixed` koordinata bakar. n buyurken eklenen
# qubitler SEYIRCI olur -> gorevin icsel zorlugu SABIT kalir -> dogruluk dususu
# SADECE EC'ye atfedilebilir. (v1'de 'parity' tum n'e bakiyordu, iki etki
# karisiyordu -- 02_PROJECT_STATUS.md 5.7 madde 2.)
#
# `qglobal` BILEREK istisna: n ile zorlasir, "EC referansi" olarak raporlanir.

LABELS = ['lin3', 'quad3', 'parity3', 'trig3', 'qlocal', 'qglobal']


def make_labels(X, kind, n, seed, states=None, n_fixed=3):
    rng = np.random.default_rng(seed + 5000)
    k = min(n_fixed, X.shape[1])
    x = X[:, :k]
    if kind == 'lin3':
        vals = np.sin(x) @ rng.normal(size=k)
    elif kind == 'quad3':
        vals = np.sin(x[:, 0])*np.sin(x[:, 1]) + 0.5*np.cos(x[:, 2 % k])
    elif kind == 'parity3':
        # ZATEN +-1 degerli -> medyan-merkezleme UYGULANMAZ (self-test yakaladi:
        # medyan +-1 cikinca sign(vals-medyan) her seyi tek sinifa cevirir).
        y = np.prod(np.sign(np.sin(x)), axis=1).astype(float)
        y[y == 0] = 1.0
        return y
    elif kind == 'trig3':
        vals = np.sin(x[:, 0] + x[:, 1])*np.cos(x[:, 2 % k])
    elif kind == 'qlocal':                         # ilk 2 qubitte etkili gozlenebilir
        assert states is not None
        d = 2**n
        A = rng.normal(size=(4, 4)) + 1j*rng.normal(size=(4, 4))
        O = (A + A.conj().T)/2
        s = to_np(states).reshape(-1, 4, d // 4)
        vals = np.real(np.einsum('mir,ij,mjr->m', s.conj(), O, s))
    elif kind == 'qglobal':
        # GLOBAL gozlenebilir: O = sum_j c_j P_j, P_j = UNIFORM RASTGELE AGIRLIK-n Pauli.
        #
        # ONCEKI SURUM 2^n x 2^n YOGUN MATRIS URETIYORDU -> n=16'da 68 GB, Colab
        # oturumu cokuyordu; n=12'de de tohum basina ~3 x (M x 4096 x 4096) einsum
        # ile surenin cogunu yiyordu. Bu surum matematiksel olarak ayni sinifta
        # (tum qubitlere dokunan gozlenebilir) ama maliyeti O(R M n 2^n).
        assert states is not None
        R = 8
        # agirlik-n Pauli: her qubitte X/Y/Z (identity YOK) -> basamak in {1,2,3}
        idx = np.array([sum(int(dd)*(4**k) for k, dd in enumerate(rng.integers(1, 4, n)))
                        for _ in range(R)], dtype=np.int64)
        cj = rng.normal(size=R)
        # R cekilis CAKISABILIR (n=8'de ~%0.4). `pool_layout_from_indices` unique
        # uyguladigi icin sutun sayisi R'den az olur -> eski surumde IndexError.
        # Cakisan terimlerin katsayilarini TOPLA: sum_j c_j P_j degismez.
        uniq, inv = np.unique(idx, return_inverse=True)
        c_agg = np.zeros(len(uniq))
        np.add.at(c_agg, inv, cj)
        lay = pool_layout_from_indices(uniq, n)
        E = to_np(pauli_expect_fast(states, n, lay))   # sutunlar artan Pauli indeksi
        vals = E @ c_agg                               # == uniq sirasi, yeniden siralama YOK
    else:
        raise ValueError(kind)
    y = np.sign(vals - np.median(vals))
    y[y == 0] = 1.0
    return y


# ============================================================================
# 2) KOL SECIMI  A1-A5
# ============================================================================

def capacity(E_sel):
    """gamma_i = E_x[psi_i(x)^2]  -- ETIKETSIZ kapasite (bizim kriterimiz)."""
    return to_np(xp.mean(E_sel**2, axis=0))


def select_random(n_pool, p, rng, n_draws=5):
    """A1 -- Gan et al.'in fiilen yaptigi: uniform rastgele."""
    return [rng.choice(n_pool, size=min(p, n_pool), replace=False) for _ in range(n_draws)]


def select_hbody(w_pool, p, rng, n_draws=5):
    """A2_low -- DUSUK AGIRLIK ONCELIKLI rastgele secim: agirlik<=H operatorleri UNIFORM
    ele alir, en kucuk H'den baslar. KAPASITE BILGISI KULLANMAZ.

    ⚠️ 2026-10-15 DUZELTME: Bu, Gan+'in H-body LPQK'si DEGILDIR (2311.13552 Eq. 42-43:
    agirligi TAM H olan Pauli'ler, I yok; deneyde tam 2-body havuzdan rastgele p).
    Faz B-D ve T1.1/FM kosu 1'de "A2_hbody" adiyla bu fonksiyon kullanildi ve Gan+'in
    deney rakibinden GUCLU cikti (runs/rival_check_v1). Gan+'a sadik kol: `select_hbody_exact`.

    v1 HATASI: kapasite-sirali listeden `cand[:p]` aliyordu -> A2 gizlice kapasite
    kullaniyordu ve H=H_max'te A3 ile OZDESLESIYORDU (hata #12).
    """
    w = np.asarray(w_pool)
    for H in range(1, int(w.max()) + 1):
        cand = np.where(w <= H)[0]
        if len(cand) >= p:
            if len(cand) == p:
                return [cand]
            return [rng.choice(cand, size=p, replace=False) for _ in range(n_draws)]
    cand = np.arange(len(w))
    return [rng.choice(cand, size=min(p, len(cand)), replace=False) for _ in range(n_draws)]


def select_hbody_exact(w_pool, h, p, rng, n_draws=5):
    """A2_hb{h} -- Gan+'a SADIK H-body (2311.13552 Eq. 42-43, Sec. IV): agirligi TAM `h`
    olan operatorlerden UNIFORM rastgele p. I (agirlik 0) hicbir zaman girmez.

    Gan+ deneyde `h = 2`, "averaged over 10 different random selections of p features".
    Havuz `p`'den kucukse BOS liste doner (kol o satirda tanimsiz); esitse tek secim.
    """
    cand = np.where(np.asarray(w_pool) == h)[0]
    if len(cand) < p:
        return []
    if len(cand) == p:
        return [cand]
    return [rng.choice(cand, size=p, replace=False) for _ in range(n_draws)]


def select_topp(gamma, p):
    """A3 -- BIZIM kriterimiz: kapasiteye gore top-p (veri-bagimsiz, etiketsiz)."""
    return np.argsort(gamma)[::-1][:p]


def weight_normalized_capacity(gamma, w):
    """gamma~_i = gamma_i / E[gamma | agirlik(i)]  -- AGIRLIK YANLILIGI duzeltmesi.

    TARAMA BULGUSU (runs/screen_v1): 157 sinyalli hucrede top-16'nin agirlik profili
        BIZ (gamma):  w1:0.74  w2:0.19  w3:0.01  w4:0.00
        ORACLE (KTA): w1:0.11  w2:0.42  w3:0.31  w4:0.16
    Sinyal w2-w4'te, biz w1'e bakiyoruz. Sebep: yapili durumlarda dusuk agirlikli
    gozlenebilirlerin beklenen degerleri DOGAL OLARAK buyuk; gamma mutlak buyuklugu
    olcuyor, bilgi tasiyip tasimadigini degil. Ayrica bu, A3'un A2 (H-body, o da
    dusuk agirlik seciyor) ile ozdeslesmesini ve ona kaybetmesini acikliyor.

    Duzeltme: her operatoru KENDI AGIRLIK SINIFINDAKI tipik degere gore sirala.
    Etiketsiz kalir (sadece gamma ve agirlik kullanir) -> A5'in aksine ust sinir degil.
    """
    g = np.asarray(gamma, dtype=float)
    w = np.asarray(w)
    out = np.zeros_like(g)
    for ww in np.unique(w):
        m = (w == ww)
        mu = g[m].mean()
        out[m] = g[m]/mu if mu > 0 else 0.0
    return out


def select_topp_wnorm(gamma, w, p):
    """A6 -- agirlik-normalize kapasiteye gore top-p."""
    return np.argsort(weight_normalized_capacity(gamma, w))[::-1][:p]


# --- KAPI 3: HOPE'un gercek kurali  argmin J_k / dP_k -------------------------

def select_greedy_residual(E_sel, p, cost=None, block=4096):
    """A7/A8 -- ARTIK kapasiteye gore acgozlu secim (HOPE'un rate-distortion kurali).

    `gamma_i = E[psi_i^2]` bir operatorun MUTLAK katkisini olcer. HOPE'un kurali ise
    marjinaldir: bir birimin degeri, ZATEN TUTULANLARIN USTUNE ne kattigidir.
    Burada bu, secilenlerin gerdigi altuzaya dik bilesenin kapasitesidir:

        gamma_i^res = || psi_i - P_S psi_i ||^2 / M ,   S = simdiye kadar secilenler

    Her adimda `argmax gamma^res` secilir (= `argmin J_k/dP_k`, dP_k = 1 operator).
    `argmax gamma` bunun S=bos ozel halidir -- yani A3, bu kuralin ILK ADIMIDIR ve
    redundansi hic gormez. Tarama v2 bu korlugun bedelini olctu: A6 agirlik profilini
    duzeltti ama ortusme %6'da kaldi (bkz. 02_PROJECT_STATUS.md 6.995).

    `cost` verilirse marjinal kapasite ONA BOLUNUR (A8): tam `J_k/dP_k` formu,
    `dP_k` = operatorun olcum maliyeti (classical shadow: 3^w).
    UYARI: 3^w cok hizli buyudugu icin A8'in dusuk agirliga cokmesi BEKLENIR --
    yani A3'un hatasini tekrarlayabilir. Ikisi de olculup raporlanir.

    ETIKETSIZ kalir: yalnizca secim setindeki `psi` degerlerini kullanir.
    """
    R = xp.array(E_sel, copy=True)                    # (M, K) artiklar
    M, K = R.shape
    nrm = xp.sum(R**2, axis=0) / M
    c = None if cost is None else to_xp(np.asarray(cost, dtype=np.float64), FDT)
    chosen = []
    for _ in range(min(p, K)):
        score = nrm if c is None else nrm / xp.maximum(c, 1e-30)
        j = int(to_np(xp.argmax(score)))
        if float(to_np(nrm[j])) <= 1e-24:             # kalan her sey zaten kapsanmis
            break
        chosen.append(j)
        q = R[:, j] / xp.sqrt(xp.sum(R[:, j]**2))
        for s in range(0, K, block):                  # bellek dostu izdusum
            sl = slice(s, min(s+block, K))
            R[:, sl] -= xp.outer(q, q @ R[:, sl])
        nrm = xp.sum(R**2, axis=0) / M
        nrm[to_xp(np.asarray(chosen))] = 0.0
    # v2 (2026-10-15): rank < p ise kalan yer secilmemislerden ham kapasite sirasiyla
    # doldurulur. Eskiden `break` p'den kisa liste donduruyordu -> kollar ust uste
    # konamayip Faz B v2 dustu (n=14, depth=1, ent=0, H=2: rank 8). Doldurulanlar
    # secilenlerin gerdigi altuzayda (artik ~0), kernel'e yeni yon katmaz.
    if len(chosen) < min(p, K):
        cap = to_np(xp.sum(E_sel**2, axis=0)).astype(float)
        cap[np.asarray(chosen, dtype=int)] = -np.inf
        chosen += [int(j) for j in np.argsort(-cap, kind='stable')[:min(p, K) - len(chosen)]]
    return np.asarray(chosen)


def select_kta(E_sel, y_sel, p):
    """A5 -- DENETIMLI ust sinir (kernel-target alignment).
    KTA_i = (sum_m psi_i(x_m) y_m)^2 / sum_m psi_i(x_m)^2.  Etiket KULLANIR."""
    y = to_xp(np.asarray(y_sel, dtype=np.float64), FDT)
    num = (y[None, :] @ E_sel).ravel()**2
    den = xp.sum(E_sel**2, axis=0) + 1e-30
    return to_np(xp.argsort(num/den)[::-1][:p])


# ============================================================================
# 3) MERGE -- genellestirilmis Eckart-Young, SADECE etiketsiz sette ogrenilir
# ============================================================================

def greedy_merge(E_sel, idx, keep):
    """En korele cifti tekrar tekrar birlestirir; `keep` yon kalana dek.

    G = [[g_i, m_ij],[m_ij, g_j]] -> birlesik kapasite = lam1, distorsiyon = lam2,
    yon = ust ozvektor.  Doner: (C, distorsiyon)  --  C: (len(idx), keep) katsayi
    matrisi; ozellikler F = E[:, idx] @ C.

    KRITIK: C SADECE etiketsiz secim setinden ogrenilir, sonra tr/te'ye UYGULANIR.
    """
    E = E_sel[:, to_xp(np.asarray(idx))]
    G = to_np(E.T @ E) / E.shape[0]                  # (K,K) Gram
    K = G.shape[0]
    C = np.eye(K)
    dist = 0.0
    while C.shape[1] > keep:
        Gc = C.T @ G @ C                             # mevcut yonlerin Gram'i
        d = np.sqrt(np.maximum(np.diag(Gc), 1e-30))
        R = np.abs(Gc / np.outer(d, d))              # normalize korelasyon
        np.fill_diagonal(R, -1.0)
        i, j = np.unravel_index(np.argmax(R), R.shape)
        sub = Gc[np.ix_([i, j], [i, j])]
        ev, V = np.linalg.eigh(sub)                  # artan sirada
        dist += float(ev[0])                         # lam2 = kaybedilen kapasite
        new = C[:, i]*V[0, 1] + C[:, j]*V[1, 1]      # lam1 yonu
        C = np.delete(C, [i, j], axis=1)
        C = np.column_stack([C, new])
    return C, dist


# ============================================================================
# 4) KERNEL + KRR  (ozayrisimla lambda taramasi ~bedava)
# ============================================================================

def kernel_pair(F_a, F_b, diag_a=None, diag_b=None):
    """Birim kosegene normalize kernel. F: (M, p) ozellikler."""
    da = xp.maximum(xp.sum(F_a**2, 1), 1e-30) if diag_a is None else diag_a
    db = xp.maximum(xp.sum(F_b**2, 1), 1e-30) if diag_b is None else diag_b
    return (F_a @ F_b.T) / xp.sqrt(xp.outer(da, db))


LAMS = (1e-4, 1e-3, 1e-2, 1e-1, 1.0)


def krr_multi(Ktr, Kte, Ytr, Yte, lams=LAMS):
    """Kernel ridge, TUM etiketler tek cagride, `lambda` EGZAKT LOO-CV ile secilir.

    Neden boyle: kernel etiketten BAGIMSIZ, yani 6 etiket ayni ayrisimi paylasabilir.
    Ktr = U diag(ev) U^T BIR KEZ hesaplanir; sonra her lambda ve her etiket icin
    cozum sadece O(M^2) matris-vektor.

    LOO kapali formu (kat dongusu YOK, hem daha hizli hem daha iyi istatistik):
        H = U diag(ev/(ev+lam)) U^T,   h_i = H_ii
        yhat_i^{-i} = (yhat_i - h_i y_i) / (1 - h_i)

    Ilk surumde her (etiket, lambda, kat) icin ayri `solve` vardi -> ~112k cagri,
    pilot saatlerce surdu. Bu surum ayni sonucu ~2 ozayrisimla veriyor.

    Doner: {etiket_adi: (dogruluk_yuzde, correct_vec)}
    """
    Ktr = Ktr if Ktr.ndim == 3 else Ktr[None]      # (B,M,M)
    Kte = Kte if Kte.ndim == 3 else Kte[None]      # (B,Mte,M)
    names = list(Ytr.keys())
    Y = to_xp(np.column_stack([Ytr[k] for k in names]).astype(np.float64), FDT)   # (M,L)
    B, L = Ktr.shape[0], len(names)

    ev, U = xp.linalg.eigh(Ktr)                    # YIGIN halinde: tek cagri
    ev = xp.clip(ev.real, 0, None)                 # (B,M)
    UtY = xp.matmul(xp.swapaxes(U, 1, 2), Y[None])  # (B,M,L)  = U^T Y
    U2 = U**2
    best_acc = xp.full((B, L), -1.0, dtype=FDT)
    best_lam = np.zeros((B, L))
    for lam in lams:
        d = 1.0/(ev + lam)
        f = (ev*d)[:, :, None]                     # (B,M,1)
        yhat = xp.matmul(U, f*UtY)
        h = xp.matmul(U2, (ev*d)[:, :, None])
        loo = (yhat - h*Y[None]) / xp.maximum(1.0 - h, 1e-6)
        acc = xp.mean(xp.sign(loo) == Y[None], axis=1)     # (B,L)
        upd = acc > best_acc
        best_lam = np.where(to_np(upd), lam, best_lam)
        best_acc = xp.where(upd, acc, best_acc)

    lam_x = to_xp(best_lam, FDT)                   # (B,L)
    d = 1.0/(ev[:, :, None] + lam_x[:, None, :])   # (B,M,L)
    A = xp.matmul(U, d*UtY)                        # (B,M,L)  katsayilar
    P = to_np(xp.sign(xp.matmul(Kte, A)))          # (B,Mte,L)
    out = {}
    for b in range(B):
        for j, k in enumerate(names):
            c = (P[b, :, j] == np.asarray(Yte[k])).astype(np.float64)
            out[(b, k) if B > 1 else k] = (float(c.mean()*100), c)
    return out


def krr_accuracy(Ktr, Kte, ytr, yte, lams=LAMS):
    """Tek etiketlik ince sarmalayici (dogrulama testleri icin)."""
    r = krr_multi(Ktr, Kte, {'y': ytr}, {'y': yte}, lams)
    return r['y']


# ============================================================================
# 5) KATMAN D -- ogrenilebilirlik on-kosulu  (ON-KAYITLI)
# ============================================================================
# Faz A/A2/A2R2'nin tamami bu adima bagli. "w<=2 kernel konsantre olmuyor" tek
# basina UCUZ bir sonuc -- sabit bir kernel de konsantre olmaz. Bulgunun anlam
# kazanmasi icin ayni kernel'in HALA OGRENDIGI gosterilmelidir.

D_LIVE   = 5.0    # bir hucrenin karar tablosuna girmesi icin: en iyi kol >= sans+5
D_A0DEAD = 3.0    # "A0 olu" esigi: A0 < sans+3


def layer_d(acc_by_arm, chance=50.0):
    """A0 x Lego 2x2 tablosu. Doner: (durum, canli_mi, altin_mi)

    |                | Lego canli                  | Lego olu        |
    |----------------|-----------------------------|-----------------|
    | **A0 olu**     | `EC-ACTIVE-LEGO-SURVIVES` ✅ | `DEAD`          |
    | **A0 canli**   | `EC-MILD`                   | `LEGO-LOSES` ⚠  |

    `EC-ACTIVE-LEGO-SURVIVES` -> ALTIN, EC-mitigasyon iddiasi icin GECERLI tek rejim.
    `EC-MILD`     -> A0 saglikli; buradaki kazanc EC-mitigasyon KANITI DEGIL.
    `LEGO-LOSES`  -> tam kernel'de sinyal VAR, biz kaybediyoruz. **Bizim ALEYHIMIZE
                     sonuc, durustce raporlanmali.** (v1'in yaniltici 'EC-ACTIVE-ALL-DEAD'
                     etiketi bu durumu gizliyordu.)
    `DEAD`        -> hicbir kol ogrenmiyor, hucre bilgi tasimiyor -> dislanir.
    `AMBIGUOUS`   -> A0, [sans+3, sans+5) bandinda; ne olu ne canli sayilir, dislanir.
    """
    a0 = acc_by_arm.get('A0', 0.0)
    lego = max(v for k, v in acc_by_arm.items() if k != 'A0')
    a0_dead = a0 < chance + D_A0DEAD
    a0_alive = a0 >= chance + D_LIVE
    lego_alive = lego >= chance + D_LIVE
    if not (a0_alive or lego_alive):
        return 'DEAD', False, False
    if a0_dead and lego_alive:
        return 'EC-ACTIVE-LEGO-SURVIVES', True, True
    if a0_alive and lego_alive:
        return 'EC-MILD', True, False
    if a0_alive and not lego_alive:
        return 'LEGO-LOSES', True, False
    return 'AMBIGUOUS', False, False


# --- HATA #29 duzeltmesi (2026-08-17) -------------------------------------
# Faz B FULL'de altin hucrelerin %87'sinde `>=sans+5` sartini SADECE `A5_super`
# (etiket kullanan oracle) sagliyordu. Yani sart "bilgi hala orada" diyordu,
# "bizim ETIKETSIZ yontemimiz calisiyor" DEMIYORDU. KATI ALTIN bunu kapatir.
# Faz C'de ON-KAYITLI birincil kapi budur (01_TEST_PROTOCOL.md 6.37).
STRICT_UNSUP = ('A1_random', 'A2_hbody', 'A3_topp', 'A4_merge', 'A6_wnorm', 'A7_resid',
                'A8_cost',   # A8 KAPI 4'te acildi (2026-08-21); eski kosularda bu
                             # anahtar hic uretilmedigi icin Faz A-D sonuclari AYNI
                'A2_hb1', 'A2_hb2', 'A2_low')   # v2 rakipleri (2026-10-15), ayni mantik


def layer_d_strict(acc_by_arm, chance=50.0):
    """Katman D, ama "Lego hayatta" yarisini SADECE ETIKETSIZ bir kol saglayabilir.

    `layer_d` ile ayni 2x2; tek fark `A5_super` (ve baska her denetimli kol)
    hayatta-kalma oyuna KATILMAZ. `A0` her zaman kalir (karsilastirma ekseni).
    """
    sub = {k: v for k, v in acc_by_arm.items() if k == 'A0' or k in STRICT_UNSUP}
    if len(sub) < 2:
        raise ValueError('layer_d_strict: etiketsiz kol yok')
    return layer_d(sub, chance)


def redundancy(E_sel, idx):
    """Secilen `psi` sutunlari arasinda ortalama |Pearson korelasyonu| (kosegen haric).

    A7/A4'un mekanizmasi REDUNDANSA dayanir: artik-kapasite ve merge, ancak
    sutunlar korele oldugunda ham `gamma`'dan farklilasir. Bu olcum E2'nin
    (en guclu etki rejiminin taranmasi) OLCULEBILIR halidir -- tarama v3
    "A7 == A3" dedi ama redundansin O REJIMDE var olup olmadigini olcmemisti.
    """
    F = E_sel[:, to_xp(np.asarray(idx))]
    F = F - xp.mean(F, 0)
    s = xp.sqrt(xp.sum(F**2, 0))
    C = to_np((F.T @ F) / xp.maximum(xp.outer(s, s), 1e-30))
    k = C.shape[0]
    if k < 2:
        return 0.0
    return float(np.abs(C[~np.eye(k, dtype=bool)]).mean())


# ============================================================================
# 6) ISTATISTIK -- 01_TEST_PROTOCOL.md esikleri, DEGISTIRILMEDI
# ============================================================================

WIN_THR, ELIM_THR = 1.0, 0.3


def paired_bootstrap(A, B, n_boot=4000, seed=0):
    """A, B: (n_seed, n_test) dogru/yanlis matrisleri. Tohum duzeyinde BLOK
    yeniden orneklemeli eslestirilmis bootstrap."""
    A = np.asarray(A, float); B = np.asarray(B, float)
    D = A - B
    rng = np.random.default_rng(seed)
    ns, nt = D.shape
    point = float(D.mean()*100)
    st = np.empty(n_boot)
    for t in range(n_boot):
        s = rng.integers(0, ns, ns); i = rng.integers(0, nt, nt)
        ws = np.bincount(s, minlength=ns)
        wi = np.bincount(i, minlength=nt)
        st[t] = (ws @ D @ wi) / (ns * nt) * 100.0
    return point, float(np.percentile(st, 2.5)), float(np.percentile(st, 97.5))


def verdict(point, lo, hi, win=WIN_THR, elim=ELIM_THR):
    """Iki esik ailesi (hata #3): kazanc ve eleme esikleri TERS yone calisir."""
    if lo > 0 and point >= win:
        return 'KAZANC'
    if hi < elim:
        return 'ELEME-ADAYI'
    return 'SONUCSUZ'


COMPARISONS = {
    # --- protokolun ana sarti: A3/A4 hem A1'i hem A2'yi yensin, >=2 etikette ---
    'A3 top-p vs A1 rand':  ('A3_topp',  'A1_random'),
    'A3 top-p vs A2 hbody': ('A3_topp',  'A2_hbody'),
    'A4 merge vs A1 rand':  ('A4_merge', 'A1_random'),
    'A4 merge vs A2 hbody': ('A4_merge', 'A2_hbody'),   # v1'de EKSIKTI (hata #13)
    'A4 merge vs A3 top-p': ('A4_merge', 'A3_topp'),
    # --- Kapi 3 kollari (tarama v3'ten sonra eklendi; ELEME icin E2 eksik) ---
    'A7 resid vs A2 hbody': ('A7_resid', 'A2_hbody'),
    'A7 resid vs A1 rand':  ('A7_resid', 'A1_random'),
    'A7 resid vs A3 top-p': ('A7_resid', 'A3_topp'),
    'A7 resid vs A4 merge': ('A7_resid', 'A4_merge'),   # redundansa iki farkli cevap
    'A6 wnorm vs A3 top-p': ('A6_wnorm', 'A3_topp'),
    # --- Faz C: EC-mitigasyon iddiasinin ASIL cumlesi (2026-08-17) ---
    # "A0 olu, biz canli" bir REJIM etiketiydi (Katman D); bu ise DOGRUDAN,
    # eslestirilmis, GA'li bir karsilastirma. Iddia icin en gucluk tek satir.
    'A3 top-p vs A0 fidel': ('A3_topp',  'A0'),
    'A7 resid vs A0 fidel': ('A7_resid', 'A0'),
    # --- ust sinir ---
    # HATA #32 (2026-08-19): tek yonlu karsilastirma "A3 anlamli USTUN" vakalarini
    # gorunmez kiliyordu -- elle hesaplamak zorunda kaldim. Simdi iki yonlu.
    'A3 top-p vs A7 resid': ('A3_topp',  'A7_resid'),
    'A5 super vs A3 top-p': ('A5_super', 'A3_topp'),
    'A5 super vs A7 resid': ('A5_super', 'A7_resid'),
}


"""KAPI 4 cekirdegi -- KAYNAK MUHASEBESI.  TEK KAYNAK.

qk_core : devre, Pauli domeni, kernel, EC teshisi (Katman A/B/C)
phaseb  : etiketler, kol secimi (A1-A7), merge, KRR, istatistik, Katman D
gate4   : shot muhasebesi -- shadow varyansi, gurultulu kernel, S*, butce

BEKLENEN ORTAM: qk_core + phaseb_core'un tanimladigi her sey +
`np/xp/GPU/CDT/FDT/to_xp/to_np` (defterin 0. hucresi).

================================ PARA BIRIMI =================================
1 shot = 1 DEVRE CALISTIRMASI. Iki olcum protokolu AYNI birimde sayilir ama
kernel matrisine FARKLI carpanla donusur:

  shadow   (A1-A8): S snapshot x NOKTA sayisi. Tum (x,y) ciftleri
                    post-processing'den bedava gelir  ->  maliyet M ile LINEER.
  fidelity (A0)   : compute-uncompute, her CIFT icin ayri devre
                                                     ->  maliyet M^2 ile.

Bu `M` carpani Kapi 4'un en buyuk tek kalemidir ve hata #17'nin ("A0'in mutlak
shot sayilari GFQK ile ayni para birimi degil") cozumudur: birim ayni, carpan
farkli, ikisi de acikca yazili.

Ikinci eksen (devre derinligi) `twoq_gates` ile ayri raporlanir; ucuncu eksen
(model karmasikligi `p`) kollar arasi kiyasta SABIT tutulur. Protokol geregi
uc eksen BIRLIKTE degil AYRI AYRI raporlanir (01_TEST_PROTOCOL.md, KAPI 4).
==============================================================================
"""

SHADOW_BASE = 3.0


def op_cost(w):
    """`dP_k` -- bir Pauli'nin classical-shadow olcum maliyeti: `3^agirlik`.

    A8'in maliyet vektoru budur (`select_greedy_residual(..., cost=op_cost(w))`,
    yani HOPE'un tam `argmin J_k/dP_k` kurali). Kapi 1-3'te `dP_k = 1` alinmisti;
    protokol A8'i "shot muhasebesi Kapi 4'un konusu" diye burada aciyor.
    """
    return SHADOW_BASE ** np.asarray(w, dtype=float)


def dir_var(w, C=None, model='l1'):
    """Yonlerin TEK SNAPSHOT varyansi. Doner `(p,)`.

    `C=None`   -> ciplak Pauli'ler: `v_k = 3^{w_k}`.
    `C` verili -> birlestirilmis yon `f_k = sum_i C_ik P_i` (A4):
      'l1' (BIRINCIL, muhafazakar): shadow normu alt-toplamsal,
           `Var <= (sum_i |C_ik| 3^{w_i/2})^2`
      'l2' (IYIMSER alt sinir): bilesen tahminleri BAGIMSIZmis gibi,
           `Var = sum_i C_ik^2 3^{w_i}`

    Ikisi de hesaplanip raporlanir; `l1/l2` orani merge'in SHOT EKSENINDEKI
    bedelidir (olcut G2). Protokolun 2. sorusu -- "bilesen Pauli'leri ayri ayri
    mi gerekiyor?" -- cevabi: shadow altinda EK SNAPSHOT gerekmez (ayni anlik
    goruntuler tum Pauli'leri besler), ama birlesik yonun VARYANSI buyuyebilir.
    Yani merge'in maliyeti "sifir" degil: "ekstra devre yok, ekstra varyans var".
    """
    c = op_cost(w)
    if C is None:
        return c
    A = np.abs(np.asarray(C, dtype=float))
    if model == 'l1':
        return (A.T @ np.sqrt(c)) ** 2
    if model == 'l2':
        return (A.T ** 2) @ c
    raise ValueError(model)


def sigma2_arm(F, v):
    """`S=1` shot varyansi, birim-kosegene normalize kernel icin. `qk_core.
    kernel_gfqk` ile AYNI form: `2 * sum_k v_k <f_k^2> / <diag>^2`.

    F: `(M, p)` yon degerleri · v: `(p,)` tek-snapshot varyanslari.
    `S` shot'ta varyans `sigma2/S`, dolayisiyla `S_req = sigma2 / Var[k~]`
    (`1/Var` DEGIL -- hata #15).
    """
    mf2 = to_np(xp.mean(F ** 2, 0)).astype(float)
    dg = float(to_np(xp.mean(xp.sum(F ** 2, 1))))
    return float(2.0 * np.sum(np.asarray(v, dtype=float) * mf2) / max(dg, 1e-30) ** 2)


def noisy_features(F, v, S, rng):
    """`f_hat = f + N(0, v/S)`. Doner `(F_hat, diag_hat)`; diag YANLILIK DUZELTILMIS.

    `E[sum_k f_hat_k^2] = sum_k f_k^2 + sum_k v_k/S` -> duzeltilmezse dusuk
    normlu noktalar yapay olarak birbirine benzer gorunur (hata #16 ile ayni
    aile: gurultu ekseni ters donerdi). Cikarilir, pozitife kirpilir.

    BILINEN BASITLESTIRME (on-kayitli): ayni shadow kumesi TUM operatorleri
    besledigi icin `eps`'ler gercekte KORELELIDIR; burada bagimsiz alinir. Bu
    hata TUM kollarda ayni yonde -> `S*` ORANLARI icin zararsiz, MUTLAK `S*`
    icin iyimser. Kapi 4'un karari oranlar uzerinden verilir.
    """
    sd = np.sqrt(np.asarray(v, dtype=float) / float(S))
    G = F + to_xp(rng.standard_normal(size=tuple(F.shape)) * sd[None, :], FDT)
    return G, xp.maximum(xp.sum(G ** 2, 1) - float(np.sum(sd ** 2)), 1e-30)


def noisy_fidelity(K, S, rng, symmetric=False):
    """A0'in gurultu modeli: compute-uncompute BINOM tahmin edici.

    `k_hat ~ Binom(S, k)/S` -> `Var = k(1-k)/S`. Shadow'daki `3^w` yuku YOK;
    A0'in bedeli varyansta degil, CIFT SAYISINDA (bkz. `budget`).

    `symmetric=True`: egitim blogu icin. Girdi bloklari bagimsiz gurultulendigi
    icin `K_hat` simetrik cikmaz; `krr_multi` `eigh` kullandigindan simetrize
    edilir. Bu ayni zamanda fiziksel olarak da dogru: `(x,y)` ve `(y,x)` AYNI
    devredir, iki kez olculmez -- yani simetrizasyon shot da kazandirmaz,
    `budget('fidelity', ...)` zaten ust ucgeni sayar.
    """
    Kc = np.clip(to_np(K).astype(float), 0.0, 1.0)
    Kh = rng.binomial(int(S), Kc) / float(int(S))
    if symmetric:
        Kh = (Kh + Kh.T) / 2.0
    return to_xp(Kh, FDT)


def s_star(S_grid, accs, target, interp=True):
    """Hedefi saglayan ve DAHA BUYUK her `S`'te de saglayan en kucuk `S`.

    Monotonluk sarti, tek bir gurultu dalgalanmasinin `S*`'i yapay olarak
    kucultmesini engeller (kol karsilastirmasinda bu dogrudan sahte kazanc
    olurdu). `S_grid` ARTAN olmali. Hicbiri saglamiyorsa `inf`.

    `interp=True` (varsayilan): kesisme `log10(S)`'te DOGRUSAL ara degerle
    bulunur. GEREKCE: grid adimi yarim dekad (~3.16x); grid'e yapistirilmis
    `S*` yalnizca `{1, 3.16, 10, ...}` oranlarini uretebilir, yani "2x tasarruf"
    gibi bir esik OLCULEMEZ hale gelir. Ara deger cozunurlugu surekli yapar.
    Gridin ILK noktasinda zaten saglaniyorsa `S_grid[0]` doner (yani "<= S_min",
    ust sinir olarak okunur); `inf` girdisinde ara deger yapilmaz.
    """
    a = list(accs)
    for i, S in enumerate(S_grid):
        if not all(x >= target for x in a[i:]):
            continue
        if i == 0 or not interp or np.isinf(S):
            return float(S)
        s0, a0, a1 = S_grid[i - 1], a[i - 1], a[i]      # a0 < target <= a1
        if a1 <= a0:
            return float(S)
        f = (target - a0) / (a1 - a0)
        return float(10.0 ** (np.log10(s0) + f * (np.log10(S) - np.log10(s0))))
    return float('inf')


def budget(protocol, S, m_tr, m_te, m_sel=0):
    """Tum kernel matrisini kurmak icin TOPLAM devre calistirmasi.

    `m_sel` = kriter (`gamma`) icin olculen SECIM seti. A1/A2 kriter kullanmaz
    -> `m_sel=0`; A3/A4/A7/A8 kullanir. Ayni anlik goruntuler havuzdaki TUM
    `M_H` operatoru besledigi icin ek maliyet operator sayisinda LOGARITMIK
    (bkz. 01_TEST_PROTOCOL.md 2.5) -- yani `m_sel` NOKTA sayisiyla dogrusal,
    havuz buyuklugunden bagimsiz sayilir. Kriter maliyetinin karari cevirip
    cevirmedigi G5'te ayrica raporlanir.
    """
    if protocol == 'shadow':
        return float(S) * (m_tr + m_te + m_sel)
    if protocol == 'fidelity':
        return float(S) * (m_tr * (m_tr - 1) / 2.0 + m_te * m_tr)
    raise ValueError(protocol)


def twoq_gates(n, depth, ent, protocol):
    """Devre basina 2-qubit kapi sayisi (ikinci maliyet ekseni).

    ZZ-feature map: katman basina `n(n-1)/2` ZZ dondurmesi, `ent=0` ise hic.
    Fidelity compute-uncompute -> 2 kat. Shadow'un rastgele olcum katmani TEK
    QUBIT'tir, 2q sayisina katki YOK.
    """
    g = depth * (n * (n - 1) // 2) if ent > 0 else 0
    return 2 * g if protocol == 'fidelity' else g


def weight_mix(w, idx):
    """Secimin ortalama agirligi + `3^w` cinsinden ortalama maliyeti.

    G3'un teshisi: A8'in dusuk agirliga COKUP cokmedigi (yani A2'ye yakinsayip
    yakinsamadigi) buradan gorulur -- `select_greedy_residual` docstring'inde
    on-kayitli olarak beklenen sonuc budur.
    """
    ww = np.asarray(w, dtype=float)[np.asarray(idx, dtype=int)]
    return float(ww.mean()), float(op_cost(ww).mean())




def enc(X, n, depth, ent, bw):
    return encode(X, n, depth, ent, bw=bw, norm=NORM)


def pauli_expect_batched(X, n, depth, ent, bw, layout, batch=512):
    """Noktalar uzerinde PARCALI -- tum durumlari ayni anda bellekte tutmaz."""
    out = []
    for i in range(0, len(X), batch):
        psi = enc(X[i:i+batch], n, depth, ent, bw)
        out.append(pauli_expect_fast(psi, n, layout))
        del psi; free_mem()
    return xp.concatenate(out, axis=0)


def fidelity_blocks(Xa, Xb, n, depth, ent, bw, batch=512):
    def st(X):
        return [enc(X[i:i+batch], n, depth, ent, bw) for i in range(0, len(X), batch)]
    Sa, Sb = st(Xa), st(Xb)
    K = xp.concatenate([xp.concatenate([xp.abs(a.conj() @ b.T)**2 for b in Sb], 1)
                        for a in Sa], 0)
    del Sa, Sb; free_mem()
    return K
```

## Bolum 3 - DOGRULAMA (gecmezse ASAGI DEVAM ETME)

```python
def _validate():
    rng = np.random.default_rng(0); n = 6; M = 400
    X = rng.uniform(0, 2*np.pi, size=(M, n))
    lay = pool_layout(n, 3)
    E = pauli_expect_batched(X, n, 2, 1.0, 0.3, lay, batch=64)
    wp = np.asarray(lay[3])

    # [1] parcali Pauli yolu == tek seferlik  (onceki fazlarin testi, korunuyor)
    Ea = pauli_expect_fast(enc(X, n, 2, 1.0, 0.3), n, lay)
    assert np.abs(to_np(E-Ea)).max() < 1e-4
    print('[1] OK  parcali Pauli yolu == tek seferlik')

    # [2] MUHASEBE TUTARLILIGI: sigma2_arm == kernel_gfqk'nin s2'si.
    #     Iki BAGIMSIZ yol ayni sayiyi vermeli; vermezse Kapi 4'un butun
    #     shot sayilari Faz A-D'nin S_req'iyle ayni para biriminde DEGIL demektir.
    for H in (2, 3):
        m = wp <= H
        _, s2_ref = kernel_gfqk(E, wp, H, 0.0)
        s2_new = sigma2_arm(E[:, to_xp(m)], op_cost(wp[m]))
        assert abs(s2_new - s2_ref)/s2_ref < 1e-4, (H, s2_new, s2_ref)
    print('[2] OK  sigma2_arm == kernel_gfqk.s2  (H=2,3)')

    # [3] shot gurultusu 1/S olcekleniyor ve sigma2_arm'i tutturuyor
    #     -> ANALITIK S_req ile AMPIRIK S* ayni birimde (G4'un on kosulu)
    idx = select_topp(capacity(E), 16)
    F = E[:, to_xp(np.asarray(idx))]; v = dir_var(wp[idx])
    Kc = to_np(kernel_pair(F, F)); off = ~np.eye(M, dtype=bool)
    for S in (1e3, 1e4):
        d = []
        for _ in range(20):
            G, dg = noisy_features(F, v, S, rng)
            d.append(to_np(kernel_pair(G, G, dg, dg))[off] - Kc[off])
        r = float(np.var(np.concatenate(d))) / (sigma2_arm(F, v)/S)
        assert 0.5 < r < 2.0, (S, r)
    print(f'[3] OK  ampirik/analitik shot varyansi = {r:.2f}')

    # [4] kosegen yanliligi CIKIYOR (duzeltilmezse dusuk normlu noktalar yapay
    #     olarak benzer gorunur -- hata #16 ile ayni aile)
    d0 = to_np(xp.sum(F**2, 1)); acc = np.zeros_like(d0)
    for _ in range(200): acc += to_np(noisy_features(F, v, 5e4, rng)[1])
    rel = float(np.abs(acc/200 - d0).mean()/d0.mean())
    assert rel < 0.02, rel
    print(f'[4] OK  kosegen yanliligi duzeltildi (bagil sapma {rel:.1e})')

    # [5] merge varyans modelleri: C=I'de l1==l2==3^w, genelde l1>=l2
    assert np.allclose(dir_var(wp[idx], np.eye(16), 'l1'), 3.0**wp[idx])
    Cm, _ = greedy_merge(E, select_topp(capacity(E), 32), 16)
    v1 = dir_var(wp[select_topp(capacity(E), 32)], Cm, 'l1')
    v2 = dir_var(wp[select_topp(capacity(E), 32)], Cm, 'l2')
    assert np.all(v1 >= v2 - 1e-9)
    print(f'[5] OK  merge varyansi l1>=l2  (medyan bedel {np.median(v1/v2):.2f}x)')

    # [6] A0'in BINOM modeli: Var == k(1-k)/S; egitim blogu simetrik
    Kf, _ = kernel_fidelity(enc(X[:120], n, 2, 1.0, 0.3))
    K0 = to_np(Kf)
    d = [to_np(noisy_fidelity(Kf, 1e3, rng)) - K0 for _ in range(40)]
    r = float(np.var(np.concatenate([x.ravel() for x in d]))) / (np.mean(K0*(1-K0))/1e3)
    assert 0.5 < r < 2.0, r
    Ks = to_np(noisy_fidelity(Kf, 1e3, rng, symmetric=True))
    assert np.allclose(Ks, Ks.T)
    print(f'[6] OK  A0 binom varyansi k(1-k)/S (oran {r:.2f}), egitim blogu simetrik')

    # [7] A8 GERCEKTEN maliyet-farkinda: pahali ama guclu yonu ATLAR
    Eo = np.linalg.qr(rng.normal(size=(M, 4)))[0]*np.sqrt(M)
    Eo = to_xp(Eo*np.array([3.0, 2.9, 2.8, 2.7]), FDT)
    w4 = np.array([4, 1, 1, 1])
    a7 = set(map(int, select_greedy_residual(Eo, 2)))
    a8 = set(map(int, select_greedy_residual(Eo, 2, cost=op_cost(w4))))
    assert 0 in a7 and 0 not in a8, (a7, a8)
    #     ve maliyet SABIT olunca A8 == A7 (dejenere hal -- kural bozulmamis)
    assert np.array_equal(select_greedy_residual(Eo, 3),
                          select_greedy_residual(Eo, 3, cost=np.ones(4)))
    print('[7] OK  A8 pahali yonu atliyor; sabit maliyette A8 == A7')

    # [8] S* semantigi: gecici sicrama S*'i KUCULTMEMELI + log-ara deger
    g = [10, 100, 1000, 10000]
    assert s_star(g, [60, 40, 40, 60], 55, interp=False) == 10000.0
    assert s_star(g, [40, 40, 60, 60], 55, interp=False) == 1000.0
    assert s_star(g, [40, 40, 40, 40], 55) == float('inf')
    assert s_star(g, [60, 60, 60, 60], 55) == 10.0
    assert abs(s_star(g, [40, 40, 60, 60], 55) - 10**2.75) < 1e-6
    assert s_star([10, 100, np.inf], [40, 40, 60], 55) == float('inf')
    #     ara deger SART: grid adimi ~3.16x, yapistirilmis S* ile "2x" olculemez
    assert (S_GRID[1]/S_GRID[0]) > G1_MIN_RATIO, 'grid adimi esikten kucuk -> interp sart'
    print('[8] OK  s_star monotonluk + log-ara deger')

    # [9] PARA BIRIMI: fidelity M mertebesinde pahali, kriter kalemi eklenebiliyor
    bs = budget('shadow', 1, N_TR, N_TE); bf = budget('fidelity', 1, N_TR, N_TE)
    #  oran = (M_tr(M_tr-1)/2 + M_te·M_tr)/(M_tr+M_te) -- `M` mertebesinde.
    #  Sabit bir esik yerine KAPALI FORM dogrulanir (grid degisince kirilmasin).
    exp = (N_TR*(N_TR-1)/2 + N_TE*N_TR)/(N_TR+N_TE)
    assert abs(bf/bs - exp) < 1e-9 and bf/bs > 10, (bf/bs, exp)
    assert budget('shadow', 1, N_TR, N_TE, m_sel=N_SEL) == bs + N_SEL
    assert twoq_gates(8, 8, 1.0, 'fidelity') == 2*twoq_gates(8, 8, 1.0, 'shadow')
    assert twoq_gates(8, 8, 0.0, 'shadow') == 0
    print(f'[9] OK  fidelity/shadow devre orani = {bf/bs:.0f}x  (M={N_TR+N_TE})')

    # [10] A8 KATI ALTIN oyununa GIRIYOR (STRICT_UNSUP'a eklendi)
    a = {'A0': 51.0, 'A1_random': 51.0, 'A2_hbody': 51.0, 'A3_topp': 51.0,
         'A4_merge': 51.0, 'A7_resid': 51.0, 'A8_cost': 57.0, 'A5_super': 58.0}
    assert layer_d_strict(a, CHANCE)[2] is True, 'A8 etiketsiz sayilmiyor!'
    assert layer_d_strict(dict(a, A8_cost=51.0), CHANCE)[2] is False
    print('[10] OK  A8 etiketsiz kol olarak KATI ALTIN sayiliyor')

    print('\nDOGRULAMA 10/10 GECTI')


_validate()
```

## Bolum 4 - KOSU: kol secimi x shot taramasi

Is birimi = `(n, depth, ent, bw, rep, jitter, tohum)`. `psi` BIR KEZ hesaplanir;
tum `H`, `p`, kol, etiket ve **shot degeri** onu paylasir.

Her `(H, p)` icin once secim yapilir (gurultusuz `psi` uzerinde -- kriterin
kendi shot maliyeti G5'te ayri kalem), sonra ayni oznitelikler `S_GRID`'in her
degerinde gurultulendirilip KRR kosulur.

Kaydedilenler:
- `acc[H|p|etiket|kol|S]` -- tohum ortalamasi dogruluk
- `sig[H|p|kol]` -- `sigma^2` (S=1 shot varyansi) ve `S_req` (analitik vekil)
- `mix[H|p|kol]` -- ortalama agirlik + ortalama `3^w` maliyeti (G3 teshisi)
- `ec[...]` -- A0'in `diagnose`/`layers_ac` durumu (devralinan EC kapisinin
  ucuz DOGRULAMASI; `EC-LEARN` cikmayan birimler ayri raporlanir)

```python
import pickle, shutil, pathlib
# v3 (2026-10-15): PARCALI KOSU. Tohum basina checkpoint (`{CKPT}/{birim}__sNN.pkl`); zaman butcesi
# (`HOPE_BUDGET_H`, varsayilan 11.25 sa) dolunca yeni tohuma baslanmaz. Sonraki parca onceki parcanin
# ciktisini girdi alir (Kaggle: `kernel_sources`/dataset -> /kaggle/input; Colab: Drive klasoru ayni kalir)
# ve kaldigi tohumdan devam eder. Tohum akisi `seed`'e bagli -> parcali kosu tek parcayla AYNI sayilari verir.
RUN_BUDGET_H = float(os.environ.get('HOPE_BUDGET_H', 11.25))
DEADLINE = T_NB0 + RUN_BUDGET_H*3600
SEED_EST = {}            # n -> tohum suresi x 1.15 (en kotusu)
INCOMPLETE = []
for f in pathlib.Path('/kaggle/input').rglob('*'):
    if not f.is_file():
        continue
    if f.suffix == '.pkl' and '__s' in f.name:
        dst = os.path.join(CKPT, f.name)
    elif f.parent.name == CKPT:
        dst = os.path.join(CKPT, f.name)
    elif f.name.startswith('pre_M') and 'PRE_CKPT' in globals():
        dst = os.path.join(PRE_CKPT, f.name)
    else:
        continue
    if not os.path.exists(dst):
        shutil.copy(f, dst)


def _seed_est(n):
    return SEED_EST.get(n, max(SEED_EST.values(), default=0.0))


def run_unit(n, depth, ent, bw, rep, jit, seeds, log=None):
    # v3: agirlik <= 2 flip'lerinin HEPSI, yalniz 3-4 flip'liler kirpilir (denetim md. 4)
    layout = pool_layout(n, max(H_LIST), max_flips=max_flips_for(n), seed=SEED0, keep_low=2)
    w_pool = np.asarray(layout[3])
    accs_all = defaultdict(list); corr_all = defaultdict(list)
    sig_all  = defaultdict(list); mix_all  = defaultdict(list); ec_all = defaultdict(list)
    ukey = unit_key(n, depth, ent, bw, rep, jit)

    for si, seed in enumerate(seeds):
        sp = os.path.join(CKPT, f'{ukey}__s{si:02d}.pkl')     # v3: tohum checkpoint'i
        if os.path.exists(sp):
            with open(sp, 'rb') as fh:
                _ld = pickle.load(fh)
            for k, v in _ld[0].items(): accs_all[k] += v
            for k, v in _ld[1].items(): corr_all[k] += v
            for k, v in _ld[2].items(): sig_all[k] += v
            for k, v in _ld[3].items(): mix_all[k] += v
            for k, v in _ld[4].items(): ec_all[k] += v
            continue
        if time.time() + _seed_est(n) > DEADLINE:
            return None                                    # sonraki parcaya kalir
        ts = time.time()
        accs = defaultdict(list); corr = defaultdict(list)
        sig  = defaultdict(list); mix  = defaultdict(list); ec = defaultdict(list)
        _rj = 0 if (rep == 1 and jit == 0.0) else 101*rep + int(37*jit) + 7
        rng  = np.random.default_rng(SEED0 + 9973*seed + n + int(1000*bw) + _rj)
        # SHOT gurultusu AYRI akistan: veri/secim akisi bozulmasin diye
        nrng = np.random.default_rng(SEED0 + 7717*seed + 13*n + int(100*bw) + _rj)

        Xs = sample_X(rng, N_SEL, n, rep, jit)
        Xt = sample_X(rng, N_TR,  n, rep, jit)
        Xe = sample_X(rng, N_TE,  n, rep, jit)
        Es = pauli_expect_batched(Xs, n, depth, ent, bw, layout)
        Et = pauli_expect_batched(Xt, n, depth, ent, bw, layout)
        Ee = pauli_expect_batched(Xe, n, depth, ent, bw, layout)

        Ys, Yt, Ye = {}, {}, {}
        for kind in LABELS_C:
            if kind == 'qlocal':
                ns_ = min(N_SEL, 512)
                st_s = enc(Xs[:ns_], n, depth, ent, bw)
                st_t = enc(Xt, n, depth, ent, bw); st_e = enc(Xe, n, depth, ent, bw)
                Ys[kind] = make_labels(Xs[:ns_], kind, n, seed, states=st_s)
                Yt[kind] = make_labels(Xt, kind, n, seed, states=st_t)
                Ye[kind] = make_labels(Xe, kind, n, seed, states=st_e)
                del st_s, st_t, st_e; free_mem()
            else:
                Ys[kind] = make_labels(Xs, kind, n, seed)
                Yt[kind] = make_labels(Xt, kind, n, seed)
                Ye[kind] = make_labels(Xe, kind, n, seed)

        # ---------- A0: H/p'den BAGIMSIZ, shot taramasi BINOM modelle ----------
        Ktt = fidelity_blocks(Xt, Xt, n, depth, ent, bw)
        Ket = fidelity_blocks(Xe, Xt, n, depth, ent, bw)
        # devralinan EC kapisinin UCUZ dogrulamasi (kernel zaten elimizde)
        d0 = diagnose(Ktt, budgets=(BUDGET,))
        st0, a_shot, a_learn = layers_ac(d0, BUDGET, MEAN_GATE, EFFRANK_GATE)
        ec['status'].append(st0); ec['effrank_frac'].append(d0['effrank_frac'])
        ec['offdiag_mean'].append(d0['offdiag_mean']); ec['s_req_A0'].append(d0['s_req'])
        for S in S_GRID:
            if np.isinf(S):
                kt, ke = Ktt, Ket
            else:
                kt = noisy_fidelity(Ktt, S, nrng, symmetric=True)
                ke = noisy_fidelity(Ket, S, nrng)
            r0 = krr_multi(kt, ke, Yt, Ye)
            for kind, (a, c) in r0.items():
                for H in H_LIST:
                    for p in P_LIST:
                        accs[f'{H}|{p}|{kind}|A0|{S}'].append(a)
                        if np.isinf(S):
                            corr[f'{H}|{p}|{kind}|A0'].append(c)
            del kt, ke; free_mem()
        del Ktt, Ket; free_mem()

        # ---------- Lego kollari ----------
        for H in H_LIST:
            mi = np.where((w_pool <= H) & (w_pool >= 1))[0]   # v3: I hicbir havuzda yok
            if len(mi) < min(P_LIST):
                continue
            ix_H = to_xp(mi)
            Es_H, Et_H, Ee_H = Es[:, ix_H], Et[:, ix_H], Ee[:, ix_H]
            gam = capacity(Es_H); wH = w_pool[mi]; cst = op_cost(wH)

            for p in P_LIST:
                if p > len(mi):
                    continue
                # sels: (tag, sutun indeksleri, C ya da None)
                sels = []
                for idx in select_random(len(mi), p, rng, N_DRAWS):
                    sels.append(('A1_random', np.asarray(idx), None))
                for idx in select_hbody(wH, p, rng, N_DRAWS):          # eski kol, yeni adi
                    sels.append(('A2_low', np.asarray(idx), None))
                for h, tag in ((1, 'A2_hb1'), (2, 'A2_hb2')):         # Gan+'a sadik (tam h)
                    for idx in select_hbody_exact(wH, h, p, rng, N_DRAWS):
                        sels.append((tag, np.asarray(idx), None))
                s3 = select_topp(gam, p)
                sels.append(('A3_topp', np.asarray(s3), None))
                s7 = select_greedy_residual(Es_H, p)
                sels.append(('A7_resid', np.asarray(s7), None))
                # A8 = HOPE'un TAM kurali: dP_k = 3^w (shadow olcum maliyeti)
                s8 = select_greedy_residual(Es_H, p, cost=cst)
                sels.append(('A8_cost', np.asarray(s8), None))
                cand = select_topp(gam, min(2*p, len(mi)))
                Cm, dist = greedy_merge(Es_H, cand, p)
                sels.append(('A4_merge', np.asarray(cand), Cm))
                # v3: oracle (etiket basina) gurultulu taramaya girer; eskiden yalniz S=inf
                for kind in LABELS_C:
                    ns_ = min(len(Ys[kind]), Es_H.shape[0])
                    sels.append(('A5_super|' + kind,
                                 np.asarray(select_kta(Es_H[:ns_], Ys[kind][:ns_], p)), None))

                # ---- oznitelikler + her kolun TEK SNAPSHOT varyans vektoru ----
                packs = []
                for tag, idx, Cc in sels:
                    ii = to_xp(idx)
                    if Cc is None:
                        Ft, Fe_ = Et_H[:, ii], Ee_H[:, ii]
                        v1 = dir_var(wH[idx]); v2 = v1
                    else:
                        Cx = to_xp(Cc, FDT)
                        Ft, Fe_ = Et_H[:, ii] @ Cx, Ee_H[:, ii] @ Cx
                        v1 = dir_var(wH[idx], Cc, 'l1')     # BIRINCIL (muhafazakar)
                        v2 = dir_var(wH[idx], Cc, 'l2')     # iyimser alt sinir
                    packs.append((tag, Ft, Fe_, v1))
                    s2 = sigma2_arm(Ft, v1)
                    K0 = kernel_pair(Ft, Ft)
                    off = K0[~xp.eye(K0.shape[0], dtype=bool)]
                    vr = float(to_np(off.var()))
                    sig[f'{H}|{p}|{tag}|sigma2'].append(s2)
                    sig[f'{H}|{p}|{tag}|sigma2_l2'].append(sigma2_arm(Ft, v2))
                    sig[f'{H}|{p}|{tag}|s_req'].append(s2/vr if vr > 0 else np.inf)
                    mw, mc = weight_mix(wH, idx)
                    mix[f'{H}|{p}|{tag}|w'].append(mw)
                    mix[f'{H}|{p}|{tag}|cost'].append(mc)
                    del K0, off; free_mem()

                sig[f'{H}|{p}|A4_merge|dist'].append(float(dist))
                sig[f'{H}|{p}|A3_topp|rho'].append(redundancy(Es_H, s3))
                sig[f'{H}|{p}|overlap_A7_A8'].append(
                    len(set(map(int, s7)) & set(map(int, s8)))/float(p))

                # ---- SHOT TARAMASI ----
                for S in S_GRID:
                    Gt, Ge, dt, de, tg = [], [], [], [], []
                    for (tag, Ft, Fe_, v) in packs:
                        if np.isinf(S):
                            gt, dtt = Ft, xp.maximum(xp.sum(Ft**2, 1), 1e-30)
                            ge, dee = Fe_, xp.maximum(xp.sum(Fe_**2, 1), 1e-30)
                        else:
                            gt, dtt = noisy_features(Ft, v, S, nrng)
                            ge, dee = noisy_features(Fe_, v, S, nrng)
                        Gt.append(gt); Ge.append(ge); dt.append(dtt); de.append(dee)
                        tg.append(tag)
                    F3, E3 = xp.stack(Gt), xp.stack(Ge)
                    if NOISY_KERNEL == 'RAW' or np.isinf(S):   # v3: gurultulu norm (hep pozitif)
                        D3 = xp.maximum(xp.sum(F3**2, 2), 1e-30)
                        D4 = xp.maximum(xp.sum(E3**2, 2), 1e-30)
                    else:
                        D3, D4 = xp.stack(dt), xp.stack(de)
                    F3t = xp.swapaxes(F3, 1, 2)
                    Kt = xp.matmul(F3, F3t)/xp.sqrt(D3[:, :, None]*D3[:, None, :])
                    Ke = xp.matmul(E3, F3t)/xp.sqrt(D4[:, :, None]*D3[:, None, :])
                    R = krr_multi(Kt, Ke, Yt, Ye)
                    for b, tag in enumerate(tg):
                        if tag.startswith('A5_super|'):
                            kind = tag.split('|')[1]
                            a, c = R[(b, kind)]
                            accs[f'{H}|{p}|{kind}|A5_super|{S}'].append(a)
                            if np.isinf(S):
                                corr[f'{H}|{p}|{kind}|A5_super'].append(c)
                        else:
                            for kind in LABELS_C:
                                a, c = R[(b, kind)]
                                accs[f'{H}|{p}|{kind}|{tag}|{S}'].append(a)
                                if np.isinf(S):
                                    corr[f'{H}|{p}|{kind}|{tag}'].append(c)
                    del F3, E3, D3, D4, Kt, Ke, R; free_mem()
                del packs; free_mem()
            del Es_H, Et_H, Ee_H; free_mem()
        del Es, Et, Ee; free_mem()
        with open(sp + '.tmp', 'wb') as fh:
            pickle.dump((dict(accs), dict(corr), dict(sig), dict(mix), dict(ec),), fh)
        os.replace(sp + '.tmp', sp)                         # yarim dosya kalmaz
        for k, v in accs.items(): accs_all[k] += v
        for k, v in corr.items(): corr_all[k] += v
        for k, v in sig.items(): sig_all[k] += v
        for k, v in mix.items(): mix_all[k] += v
        for k, v in ec.items(): ec_all[k] += v
        SEED_EST[n] = max(SEED_EST.get(n, 0.0), 1.15*(time.time() - ts))
        if log and (si == 0 or (si+1) % 5 == 0):
            print(f'      tohum {si+1}/{len(seeds)}  ({time.time()-log:.0f}s)', flush=True)

    accs, corr, sig, mix, ec = accs_all, corr_all, sig_all, mix_all, ec_all
    # A1/A2 birden cok cekimlidir -> tohum icinde de ortalanir
    out_a = {k: float(np.mean(v)) for k, v in accs.items()}
    out_s = {k: float(np.mean(v)) for k, v in sig.items()}
    out_m = {k: float(np.mean(v)) for k, v in mix.items()}
    out_e = {'status': Counter(ec['status']).most_common(1)[0][0],
             'effrank_frac': float(np.mean(ec['effrank_frac'])),
             'offdiag_mean': float(np.mean(ec['offdiag_mean'])),
             's_req_A0': float(np.mean(ec['s_req_A0']))}
    out_c = {}
    for k, v in corr.items():
        nd = len(v)//len(seeds)
        out_c[k] = np.array(v).reshape(len(seeds), nd, -1).mean(1)
    return out_a, out_s, out_m, out_e, out_c


RES_A, RES_S, RES_M, RES_E, RES_C = {}, {}, {}, {}, {}
seeds = list(range(N_SEED))
t00 = time.time()
for ui, (n, d, e, bw, rep, jit) in enumerate(ARM_UNITS):
    key = unit_key(n, d, e, bw, rep, jit); path = os.path.join(CKPT, key + '.npz')
    if os.path.exists(path):
        z = np.load(path, allow_pickle=True)
        RES_A[key], RES_S[key] = z['a'].item(), z['s'].item()
        RES_M[key], RES_E[key] = z['m'].item(), z['e'].item()
        RES_C[key] = z['c'].item()
        print(f'[{ui+1}/{len(ARM_UNITS)}] {key}: checkpoint')
        continue
    t0 = time.time()
    out = run_unit(n, d, e, bw, rep, jit, seeds, log=t0)
    if out is None:
        INCOMPLETE.append(key); continue
    a, s, m, ee, c = out
    np.savez_compressed(path, a=a, s=s, m=m, e=ee, c=c)
    RES_A[key], RES_S[key], RES_M[key], RES_E[key], RES_C[key] = a, s, m, ee, c
    el = time.time()-t00
    print(f'[{ui+1}/{len(ARM_UNITS)}] {key}  {time.time()-t0:6.1f}s  ({ee["status"]})  '
          f'(toplam {el/60:.1f}dk, tahmini kalan '
          f'{el/(ui+1)*(len(ARM_UNITS)-ui-1)/60:.1f}dk)', flush=True)
if INCOMPLETE:
    print(f'\nPARCA BITTI (butce {RUN_BUDGET_H} sa doldu)  {(time.time()-t00)/60:.1f} dk')
    for k in INCOMPLETE:
        nd = len([f for f in os.listdir(CKPT) if f.startswith(k + '__s') and f.endswith('.pkl')])
        print(f'  {k}: {nd}/{len(seeds)} tohum')
    print('Kalan birimler icin sonraki parcayi bu ciktiyla kos. Analiz hucreleri ATLANIYOR.')
    get_ipython().input_transformers_cleanup.append(lambda lines: ['pass\n'])
else:
    print(f'\nKOSU BITTI  {(time.time()-t00)/60:.1f} dk')

# devralinan EC kapisinin dogrulamasi
bad = [k for k, v in RES_E.items() if v['status'] != 'EC-LEARN']
print(f'\nEC kapisi (Faz D\'den devralindi, burada dogrulandi): '
      f'{len(RES_E)-len(bad)}/{len(RES_E)} birim EC-LEARN')
for k in bad:
    print(f'  ! {k}: {RES_E[k]["status"]}  effrank/M={RES_E[k]["effrank_frac"]:.3f} '
          f'-> satirlari AYRI raporlanir')
```

## Bolum 5 - Satirlar + `S*` hesabi

Her satir = `(birim, H, p, etiket)`. Katman D **gurultusuz** dogruluklardan
hesaplanir (rejim etiketi bir hucre ozelligidir, shot butcesinin fonksiyonu
degildir -- bu ayrim on-kayitli).

`S*` iki hedefte hesaplanir:
- **birincil** `tau = A2(gurultusuz) - 1.0` -> "H-body ile ayni dogruluk"
- **ikincil** `tau2 = sans + 5` -> "kullanilabilir siniflandiriciya en ucuz yol"

`S*` grid noktasina **yapistirilmaz**, `log10(S)`'te ara deger ile bulunur.
Grid adimi yarim dekad (`~3.16x`); yapistirilsaydi oranlar yalnizca
`{1, 3.16, 10, ...}` degerlerini alabilir ve on-kayitli `2x` esigi
**olculemez** olurdu. (Bu, defter yazilirken fark edilip duzeltildi;
Bolum 3'te `[8]` bunu ayrica assert eder.)

```python
FIN = [s for s in S_GRID if not np.isinf(s)]


def _acc(key, H, p, kind, arm, S):
    return RES_A[key].get(f'{H}|{p}|{kind}|{arm}|{S}', np.nan)


rows = []
for (n, d, e, bw, rep, jit) in ARM_UNITS:
    key = unit_key(n, d, e, bw, rep, jit)
    if key not in RES_A:
        continue
    for H in H_LIST:
        for p in P_LIST:
            for kind in LABELS_C:
                clean = {a: _acc(key, H, p, kind, a, np.inf) for a in ARMS_RAW}
                # `A2_hb1` p > |1-body| iken tanimsiz (NaN) -- satiri dusurmez
                if any(np.isnan(v) for a, v in clean.items() if a != 'A2_hb1'):
                    continue
                # v3 birincil rakip: Gan+ H-body ailesinin bu satirdaki GUCLUSU (test dogrulugu ile,
                # rakip lehine -- muhafazakar; makalede yazilir)
                HREF = ('A2_hb1' if np.isfinite(clean['A2_hb1']) and clean['A2_hb1'] > clean['A2_hb2']
                        else 'A2_hb2')
                clean['A2_hbody'] = clean[HREF]
                cfin = {a: v for a, v in clean.items() if np.isfinite(v)}
                st,  live,  gold  = layer_d(cfin, CHANCE)
                sts, lives, golds = layer_d_strict(cfin, CHANCE)
                tau = clean['A2_hbody'] - TAU_DELTA
                r = dict(n=n, depth=d, ent=e, bw=bw, rep=rep, jitter=jit, key=key,
                         H=H, p=p, label=kind, status=st, live=live, gold=gold,
                         status_strict=sts, live_strict=lives, gold_strict=golds,
                         ec=RES_E[key]['status'], tau=tau,
                         rho=RES_S[key].get(f'{H}|{p}|A3_topp|rho', np.nan),
                         merge_dist=RES_S[key].get(f'{H}|{p}|A4_merge|dist', np.nan),
                         overlap_A7_A8=RES_S[key].get(f'{H}|{p}|overlap_A7_A8', np.nan))
                r['hbody_ref'] = HREF
                for a in ARMS_RAW + ['A2_hbody']:
                    src = HREF if a == 'A2_hbody' else a
                    curve = [_acc(key, H, p, kind, src, S) for S in S_GRID]
                    r[f'acc_{a}'] = clean[a]
                    r[f'S_{a}']   = s_star(S_GRID, curve, tau)
                    r[f'S2_{a}']  = s_star(S_GRID, curve, TAU2)
                    r[f'curve_{a}'] = curve
                for a in ARMS_SHOT + ['A2_hb1', 'A2_hb2']:
                    src = HREF if a == 'A2_hbody' else a
                    r[f'sig_{a}']  = RES_S[key].get(f'{H}|{p}|{src}|sigma2', np.nan)
                    r[f'sreq_{a}'] = RES_S[key].get(f'{H}|{p}|{src}|s_req', np.nan)
                    r[f'w_{a}']    = RES_M[key].get(f'{H}|{p}|{src}|w', np.nan)
                    r[f'c_{a}']    = RES_M[key].get(f'{H}|{p}|{src}|cost', np.nan)
                r['sig_A4_l2'] = RES_S[key].get(f'{H}|{p}|A4_merge|sigma2_l2', np.nan)
                rows.append(r)

LIVE = [r for r in rows if r['live'] and r['ec'] == 'EC-LEARN']
OFFEC = [r for r in rows if r['live'] and r['ec'] != 'EC-LEARN']
GOLD = [r for r in LIVE if r['gold_strict']]
print(f'satir: {len(rows)}   CANLI (EC-LEARN): {len(LIVE)}   '
      f'KATI ALTIN: {len(GOLD)}   EC disi ama canli (ayri raporlanir): {len(OFFEC)}')
print('durum dagilimi:', Counter(r['status'] for r in rows).most_common())
print('etiketler (canli):', Counter(r['label'] for r in LIVE).most_common())
print('birincil rakip (Gan+ H-body ailesinin guclusu), canli:', Counter(r['hbody_ref'] for r in LIVE).most_common())

# S* = inf ne siklikta? (hedefe hicbir butcede ulasilamiyor)
print('\nS* = inf orani (canli satirlarda):')
for a in ARMS:
    v = [r[f'S_{a}'] for r in LIVE]
    ninf = sum(np.isinf(x) for x in v)
    med = np.median([x for x in v if not np.isinf(x)]) if ninf < len(v) else np.nan
    print(f'  {a:<11} inf: {ninf:>4}/{len(v):<4}  sonlu medyan S* = '
          + ('-' if np.isnan(med) else f'{med:.0e}'))
```

## Bolum 6 - **G1 (BIRINCIL)**: sabit dogrulukta shot

`oran = S*(A2_hbody) / S*(kol)`. `>1` = kol AYNI dogruluga **daha az shot**la
ulasiyor. `S*` sonsuz olan satirlarda oran `RATIO_CAP`'e kirpilir ve kirpilan
satir sayisi **ayrica** yazilir (sessizce sonsuza kacmaz).

Ayni tabloda **toplam devre** karsiligi da verilir -- `S*` shot/nokta, butce ise
protokolun carpaniyla toplam calistirma (A0 icin `M^2`, shadow icin `M`).

```python
def boot_median(v, n_boot=N_BOOT_G, seed=0):
    v = np.asarray(v, dtype=float)
    if len(v) == 0:
        return np.nan, np.nan, np.nan
    rng = np.random.default_rng(seed)
    st = np.median(v[rng.integers(0, len(v), (n_boot, len(v)))], axis=1)
    return float(np.median(v)), float(np.percentile(st, 2.5)), float(np.percentile(st, 97.5))


def ratio(num, den):
    """S*(A2)/S*(kol), inf'ler tavana kirpilir. Doner (oran, kirpildi_mi)."""
    if np.isinf(num) and np.isinf(den):
        return 1.0, True
    if np.isinf(num):
        return RATIO_CAP, True
    if np.isinf(den):
        return 1.0/RATIO_CAP, True
    return num/den, False


def g1_table(rowset, field='S_', title=''):
    print(f'\n=== {title} ===')
    print(f'{"kol":<11} {"n":>4} {"medyan oran":>12} {"%95 GA":>18} '
          f'{"kirpik":>7} {"etiket":>7} {"KARAR":>10}')
    out = {}
    for a in ['A3_topp', 'A4_merge', 'A7_resid', 'A8_cost', 'A0']:
        rs = [ratio(r[field+'A2_hbody'], r[field+a]) for r in rowset]
        v = [x for x, _ in rs]; clip = sum(1 for _, c in rs if c)
        m, lo, hi = boot_median(v)
        labs = set()
        for r, (x, _) in zip(rowset, rs):
            if x > 1.0 and r['label'] not in UNFAIR:
                labs.add(r['label'])
        ok = (m >= G1_MIN_RATIO and lo > G1_MIN_LO and
              len(labs) >= G1_MIN_LABELS and len(v) >= G1_MIN_ROWS)
        out[a] = dict(n=len(v), median=m, lo=lo, hi=hi, clipped=clip,
                      labels=sorted(labs), passed=bool(ok))
        print(f'{a:<11} {len(v):>4} {m:>12.2f} [{lo:>7.2f},{hi:>7.2f}] '
              f'{clip:>7} {len(labs):>7} {"GECTI" if ok else "-":>10}')
    return out


G1 = g1_table(LIVE, 'S_', f'G1 BIRINCIL -- hedef: A2 gurultusuz - {TAU_DELTA} puan')
G1_TAU2 = g1_table(LIVE, 'S2_', f'G1 IKINCIL -- hedef: sans+5 = {TAU2}')

# --- DOZ AYRIMI (on-kayitli): Faz D, yuksek redundansta A3'un COKTUGUNU buldu
#     (A3-A2 isaret degistiriyor: +2.44 -> -2.02, 00_RESEARCH_PLAN.md md.7c).
#     Hedef A2'ye gore tanimli oldugu icin o rejimde S*(A3)=inf BEKLENIR.
#     Toplu tablo bunu gizlerdi -> dozlar AYRI raporlanir.
G1_DOSE = {}
for (rp, jt) in DOSES:
    sub = [r for r in LIVE if (r['rep'], r['jitter']) == (rp, jt)]
    if sub:
        G1_DOSE[f'rep{rp}_jit{jt}'] = g1_table(
            sub, 'S_', f'G1 -- doz rep={rp} jit={jt}  (rho '
                       f'{np.median([r["rho"] for r in sub]):.3f})')
if GOLD:
    G1_GOLD = g1_table(GOLD, 'S_', 'G1 -- KATI ALTIN alt kumesi (ikincil)')
else:
    G1_GOLD = {}; print('\n(KATI ALTIN satir yok -- ikincil tablo atlandi)')

G1_PASS = any(G1[a]['passed'] for a in ('A3_topp', 'A4_merge', 'A7_resid', 'A8_cost'))
print(f'\nG1 KARARI: {"GECTI" if G1_PASS else "GECMEDI"}  '
      f'(gecen kol: {[a for a in G1 if G1[a]["passed"]] or "yok"})')
if not G1_PASS:
    print('  -> secim avantaji SHOT eksenine tasinmiyor. Dogruluk kazanci (Faz B-D)')
    print('     ayakta kalir; "ayni dogruluk daha az kaynakla" cumlesi YAZILMAZ.')

# --- TOPLAM DEVRE: para birimi carpanlariyla ---
print(f'\n=== TOPLAM DEVRE CALISTIRMASI (medyan, canli satirlar) ===')
print(f'{"kol":<11} {"protokol":>9} {"S*":>10} {"kriter":>8} {"toplam devre":>14} '
      f'{"2q/devre":>9} {"toplam 2q":>12}')
BUD = {}
for a in ARMS:
    prot = 'fidelity' if a == 'A0' else 'shadow'
    v = [r[f'S_{a}'] for r in LIVE if not np.isinf(r[f'S_{a}'])]
    if not v:
        print(f'{a:<11} {prot:>9} {"inf":>10}   -- hicbir butcede hedefe ulasmadi --')
        continue
    S = float(np.median(v))
    ms = N_SEL if a in CRIT_ARMS else 0
    nn = int(np.median([r['n'] for r in LIVE])); dd = int(np.median([r['depth'] for r in LIVE]))
    tot = budget(prot, S, N_TR, N_TE, ms)
    g2q = twoq_gates(nn, dd, 1.0, prot)
    BUD[a] = dict(protocol=prot, s_star=S, m_sel=ms, circuits=tot,
                  twoq_per_circuit=g2q, twoq_total=tot*g2q)
    print(f'{a:<11} {prot:>9} {S:>10.0e} {ms:>8} {tot:>14.3e} {g2q:>9} {tot*g2q:>12.3e}')
```

## Bolum 7 - **G2**: merge'in shot bedeli

Protokolun 2. ve 3. sorusu. Cevap iki parcali:

1. **Ek devre gerekmiyor.** Classical shadow'da ayni anlik goruntuler havuzdaki
   TUM Pauli'leri besler; birlesik gozlenebilirin bilesenleri icin ek olcum yok.
2. **Ama varyans bedava degil.** `f_k = sum_i C_ik P_i` icin shadow normu
   alt-toplamsaldir: `Var <= (sum_i |C_ik| 3^{w_i/2})^2` (**l1**, muhafazakar).
   Bilesenler bagimsizmis gibi sayilirsa `sum_i C_ik^2 3^{w_i}` (**l2**, iyimser
   alt sinir). Gercek deger ikisinin arasinda; **karar l1 ile verilir.**

`l1/l2` orani merge'in shot eksenindeki bedelinin buyuklugudur.

```python
print('=== G2: merge varyans bedeli (canli satirlar) ===')
ov  = [r['sig_A4_merge']/r['sig_A4_l2'] for r in LIVE
       if r['sig_A4_l2'] > 0 and np.isfinite(r['sig_A4_merge'])]
vs3 = [r['sig_A4_merge']/r['sig_A3_topp'] for r in LIVE
       if r['sig_A3_topp'] > 0 and np.isfinite(r['sig_A4_merge'])]
m_ov, lo_ov, hi_ov = boot_median(ov)
m_v3, lo_v3, hi_v3 = boot_median(vs3)
print(f'  l1/l2 (merge ic bedeli)      medyan {m_ov:6.2f}x  [{lo_ov:.2f}, {hi_ov:.2f}]')
print(f'  sigma2(A4)/sigma2(A3)  (l1)  medyan {m_v3:6.2f}x  [{lo_v3:.2f}, {hi_v3:.2f}]')

rs = [ratio(r['S_A3_topp'], r['S_A4_merge']) for r in LIVE]
m_s, lo_s, hi_s = boot_median([x for x, _ in rs])
print(f'  S*(A3)/S*(A4)  (>1: merge ucuz)  medyan {m_s:6.2f}x  [{lo_s:.2f}, {hi_s:.2f}]')
d_acc = boot_median([r['acc_A4_merge'] - r['acc_A3_topp'] for r in LIVE])
print(f'  gurultusuz dogruluk A4-A3        medyan {d_acc[0]:+6.2f} '
      f'[{d_acc[1]:+.2f}, {d_acc[2]:+.2f}] puan')
print(f'  merge distorsiyonu (kaybedilen kapasite) medyan '
      f'{np.median([r["merge_dist"] for r in LIVE]):.4f}')

if m_v3 <= 1.0:
    G2_VERDICT = 'MERGE-SHOT-KAZANIYOR'
elif m_s > 1.0:
    G2_VERDICT = 'NET-KAZANC-BEDEL-VAR'
else:
    G2_VERDICT = 'KAYNAKTA-GERI-VERILIYOR'
print(f'\nG2 SONUCU: {G2_VERDICT}')
print({'MERGE-SHOT-KAZANIYOR':
       '  -> merge hem dogrulukta hem shot ekseninde kazandiriyor.',
       'NET-KAZANC-BEDEL-VAR':
       '  -> varyans bedeli var ama S* yine de A3\'ten kucuk: net kazanc, bedel raporlanir.',
       'KAYNAKTA-GERI-VERILIYOR':
       '  -> Kapi 2\'nin dogruluk kazanci shot ekseninde GERI VERILIYOR.\n'
       '     Iddia "dogrulukta kazandirir, kaynakta kazandirmaz" olarak DARALIR.'}[G2_VERDICT])
```

## Bolum 8 - **G3**: `A8_cost` -- HOPE'un tam `argmin J_k/dP_k` kurali

`dP_k = 3^w`. **On-kayitli beklenti** (`select_greedy_residual` docstring'inde,
kosudan once yazili): `3^w` cok hizli buyudugu icin A8'in **dusuk agirliga
cokmesi** ve A2'ye yakinsamasi bekleniyor.

Teshis: secimin ortalama agirligi ve ortalama `3^w` maliyeti, kol basina.
A8'in agirlik profili A2'ninkine yaklasiyorsa beklenti dogrulanmis olur.

```python
print('=== G3: secim profilleri (canli satirlar, medyan) ===')
print(f'{"kol":<11} {"ort.agirlik":>12} {"ort.3^w":>12} {"sigma2":>12} '
      f'{"S* medyan":>11} {"gurultusuz acc":>15}')
PROF = {}
for a in ARMS_SHOT:
    w  = np.median([r[f'w_{a}'] for r in LIVE])
    c  = np.median([r[f'c_{a}'] for r in LIVE])
    s2 = np.median([r[f'sig_{a}'] for r in LIVE])
    fs = [r[f'S_{a}'] for r in LIVE if not np.isinf(r[f'S_{a}'])]
    ac = np.median([r[f'acc_{a}'] for r in LIVE])
    PROF[a] = dict(weight=float(w), cost=float(c), sigma2=float(s2),
                   s_star=float(np.median(fs)) if fs else float('inf'), acc=float(ac))
    print(f'{a:<11} {w:>12.2f} {c:>12.1f} {s2:>12.3e} '
          f'{(f"{np.median(fs):.0e}" if fs else "inf"):>11} {ac:>15.2f}')

print(f'\n  A7/A8 secim ortusmesi: '
      f'{np.median([r["overlap_A7_A8"] for r in LIVE]):.2f}')
w8, w2, w7 = PROF['A8_cost']['weight'], PROF['A2_hbody']['weight'], PROF['A7_resid']['weight']
collapsed = abs(w8 - w2) < abs(w8 - w7)
print(f'  A8 agirlik profili {"A2" if collapsed else "A7"}\'ye daha yakin '
      f'(A8={w8:.2f}, A2={w2:.2f}, A7={w7:.2f}) '
      f'-> on-kayitli cokme beklentisi {"DOGRULANDI" if collapsed else "CIKMADI"}')

rs = [ratio(r['S_A7_resid'], r['S_A8_cost']) for r in LIVE]
m8, lo8, hi8 = boot_median([x for x, _ in rs])
dacc = boot_median([r['acc_A8_cost'] - r['acc_A7_resid'] for r in LIVE])
print(f'\n  S*(A7)/S*(A8)  medyan {m8:.2f}x  [{lo8:.2f}, {hi8:.2f}]  '
      f'(esik {G3_MIN_RATIO})')
print(f'  dogruluk A8-A7 medyan {dacc[0]:+.2f} [{dacc[1]:+.2f}, {dacc[2]:+.2f}] puan  '
      f'(izin verilen kayip {G3_MAX_LOSS})')
G3_PASS = bool(m8 >= G3_MIN_RATIO and lo8 > 1.0 and dacc[0] >= -G3_MAX_LOSS)
print(f'\nG3 KARARI: {"GECTI -- HOPE\'un TAM kurali transfer etti" if G3_PASS else "GECMEDI -- A8 ELENDI (gecerli eleme: kendi ekseninde, kendi rejiminde test edildi)"}')
```

## Bolum 9 - **G4 + G5**: capraz kontrol ve kriter maliyeti

**G4.** Faz A'dan beri Katman A'nin olcutu **analitik** bir vekildi:
`S_req = sigma^2 / Var[k~]`. Kapi 4 ilk kez **ampirik** `S*` uretiyor. Ikisi
korele degilse, vekil ogrenme maliyetini ongormuyor demektir -- bu **bizim
aleyhimize** bir bulgudur ve oyle raporlanir.

**G5.** A3/A4/A7/A8 kriter (`gamma`) icin `M_sel` nokta olcer; A1/A2 olcmez.
Bu kalem butceye eklendiginde G1 karari donuyor mu?

```python
# ---------------- G4: analitik vekil vs ampirik S* ----------------
xs, ys = [], []
for r in LIVE:
    for a in ARMS_SHOT:
        s, sr = r[f'S_{a}'], r[f'sreq_{a}']
        if np.isfinite(s) and np.isfinite(sr) and sr > 0:
            xs.append(np.log10(sr)); ys.append(np.log10(s))
xs, ys = np.array(xs), np.array(ys)
if len(xs) > 10:
    if float(np.std(xs)) == 0 or float(np.std(ys)) == 0:
        cc = lo4 = hi4 = np.nan; G4_PASS = False
        print('=== G4: korelasyon TANIMSIZ (sabit dizi) ===')
    else:
        with np.errstate(invalid='ignore', divide='ignore'):
            rng = np.random.default_rng(7)
            cc_raw = float(np.corrcoef(xs, ys)[0, 1])
            st_raw = np.array([np.corrcoef(*np.array([xs, ys])[:, rng.integers(0, len(xs), len(xs))])[0, 1]
                           for _ in range(1000)])
            st = st_raw[np.isfinite(st_raw)]
            if not np.isfinite(cc_raw) or len(st) < 100:
                cc = lo4 = hi4 = np.nan; G4_PASS = False
                print(f'=== G4: korelasyon TANIMSIZ [nan, nan] (n={len(xs)}) ===')
            else:
                cc = cc_raw
                lo4, hi4 = float(np.percentile(st, 2.5)), float(np.percentile(st, 97.5))
                if not np.isfinite(lo4) or not np.isfinite(hi4):
                    G4_PASS = False
                    print(f'=== G4: korelasyon r={cc:+.3f} [nan, nan] -> TANIMSIZ (n={len(xs)}) ===')
                else:
                    G4_PASS = bool(cc > 0 and lo4 > 0)
                    print(f'=== G4: korelasyon(log10 S_req_analitik, log10 S*_ampirik) ===')
                    print(f'  n={len(xs)}  r = {cc:+.3f}  [{lo4:+.3f}, {hi4:+.3f}]  '
                          f'-> {"GECTI" if G4_PASS else "GECMEDI"}')
                    if not G4_PASS:
                        print('  -> Katman A\'nin ANALITIK olcutu ogrenme maliyetini ONGORMUYOR.')
                        print('     Bizim aleyhimize bulgu; 02_PROJECT_STATUS.md hata siciline yazilir.')
else:
    cc = lo4 = hi4 = np.nan; G4_PASS = False
    print('=== G4: yeterli sonlu S* yok, capraz kontrol yapilamadi ===')

# ---------------- G5: kriter maliyeti butceye eklenince ----------------
# ONEMLI: oranlar G1 ile AYNI sekilde hesaplanir -- SATIR BASINA oran, sonra
# medyan. (Medyanlarin orani ile oranlarin medyani ayni sey degildir; karar
# olcutu iki farkli yontemle hesaplanirsa kiyaslanamaz.)
print(f'\n=== G5: kriter maliyeti (M_sel={N_SEL} nokta) butceye eklenince ===')


def row_budget(r, a):
    S = r[f'S_{a}']
    if np.isinf(S):
        return np.inf
    prot = 'fidelity' if a == 'A0' else 'shadow'
    return budget(prot, S, N_TR, N_TE, N_SEL if a in CRIT_ARMS else 0)


print(f'{"kol":<11} {"medyan S*":>10} {"kritersiz":>12} {"kriterli":>12} {"sisme":>7}')
G5 = {}
for a in ARMS_SHOT:
    v = [r[f'S_{a}'] for r in LIVE if not np.isinf(r[f'S_{a}'])]
    if not v:
        continue
    S = float(np.median(v))
    b0 = budget('shadow', S, N_TR, N_TE, 0)
    b1 = budget('shadow', S, N_TR, N_TE, N_SEL if a in CRIT_ARMS else 0)
    G5[a] = dict(s_star=S, no_crit=b0, with_crit=b1, inflation=b1/b0)
    print(f'{a:<11} {S:>10.0e} {b0:>12.3e} {b1:>12.3e} {b1/b0:>6.2f}x')

print(f'\n  BUTCE orani (satir basina, kriter DAHIL) vs S* orani:')
print(f'{"kol":<11} {"butce orani":>12} {"%95 GA":>18} {"S* orani":>10} {"donuyor?":>9}')
G5_FLIP = []
G5_RAT = {}
for a in ('A3_topp', 'A4_merge', 'A7_resid', 'A8_cost'):
    rs = [ratio(row_budget(r, 'A2_hbody'), row_budget(r, a)) for r in LIVE]
    m, lo, hi = boot_median([x for x, _ in rs])
    was = G1[a]['median']
    flip = (was >= G1_MIN_RATIO) != (m >= G1_MIN_RATIO)
    G5_RAT[a] = dict(median=m, lo=lo, hi=hi, flips=bool(flip))
    if flip:
        G5_FLIP.append(a)
    print(f'{a:<11} {m:>12.2f} [{lo:>7.2f},{hi:>7.2f}] {was:>10.2f} '
          f'{"EVET" if flip else "-":>9}')
print(f'  kriter maliyeti karari DONDURUYOR mu: '
      f'{"EVET -> " + str(G5_FLIP) if G5_FLIP else "HAYIR"}')

# ---------------- Kapi 4 ozet karar tablosu ----------------
print('\n' + '='*66)
print('KAPI 4 KARAR TABLOSU'.center(66))
print('='*66)
print(f'  G1  sabit dogrulukta shot      : {"GECTI" if G1_PASS else "GECMEDI"}'
      f'   (gecen kol: {[a for a in G1 if G1[a]["passed"]] or "yok"})')
print(f'  G2  merge shot bedeli          : {G2_VERDICT}')
print(f'  G3  A8 (tam J_k/dP_k kurali)   : {"GECTI" if G3_PASS else "GECMEDI -> A8 elendi"}')
print(f'  G4  analitik vekil vs ampirik  : {"GECTI" if G4_PASS else "GECMEDI"}  (r={cc:+.3f})')
print(f'  G5  kriter maliyeti kararı     : {"DONDURUYOR " + str(G5_FLIP) if G5_FLIP else "dondurmuyor"}')
print('='*66)
print('  Protokol: uc maliyet ekseni (shot / devre derinligi / p) AYRI raporlanir.')
print('  Birinde kazanip digerinde kaybetmek MESRU bir sonuctur ve oyle yazilir.')
```
